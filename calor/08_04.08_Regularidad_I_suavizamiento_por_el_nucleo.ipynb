## 04.08.2 — Notación de multiíndices

### **Definición 04.08.1 — Multiíndice**

Un **multiíndice** es un vector

$$
\alpha=(\alpha_1,\ldots,\alpha_n)\in\mathbb N_0^n.
$$

Definimos

$$
|\alpha|
=
\alpha_1+\cdots+\alpha_n
$$

y

$$
D_x^\alpha
=
\frac{\partial^{|\alpha|}}
{\partial x_1^{\alpha_1}\cdots\partial x_n^{\alpha_n}}.
$$


## 04.08.3 — Escalamiento de las derivadas del núcleo

Recordemos que

$$
\psi(x,t)
=
\frac{1}{(4\pi t)^{n/2}}
\exp\!\left(-\frac{|x|^2}{4t}\right),
\qquad
t>0.
$$

También puede escribirse como

$$
\psi(x,t)
=
t^{-n/2}\Psi\!\left(\frac{x}{\sqrt t}\right),
$$

donde

$$
\Psi(z)
=
\frac{1}{(4\pi)^{n/2}}
\exp\!\left(-\frac{|z|^2}{4}\right).
$$


### **Lema 04.08.2 — Estructura de las derivadas gaussianas**

Sean

$$
\alpha\in\mathbb N_0^n
$$

y

$$
m\in\mathbb N_0.
$$

Existe un polinomio

$$
P_{\alpha,m}
$$

tal que

$$
D_x^\alpha\partial_t^m\psi(x,t)
=
t^{-\frac{n+|\alpha|+2m}{2}}
P_{\alpha,m}\!\left(\frac{x}{\sqrt t}\right)
\exp\!\left(-\frac{|x|^2}{4t}\right)
$$

para todo

$$
x\in\mathbb R^n,
\qquad
t>0.
$$

### Demostración

La representación autosimilar es

$$
\psi(x,t)
=
t^{-n/2}
\Psi\!\left(\frac{x}{\sqrt t}\right).
$$

Cada derivada espacial introduce un factor

$$
t^{-1/2}
$$

y deriva la función gaussiana respecto de la variable

$$
z=\frac{x}{\sqrt t}.
$$

Las derivadas de una gaussiana son siempre un polinomio multiplicado por la misma gaussiana.

Cada derivada temporal introduce un factor adicional

$$
t^{-1}
$$

y produce nuevamente combinaciones de polinomios en

$$
\frac{x}{\sqrt t}
$$

multiplicados por la gaussiana.

Después de

$$
|\alpha|
$$

derivadas espaciales y

$$
m
$$

derivadas temporales, el factor total de escala es

$$
t^{-n/2}
t^{-|\alpha|/2}
t^{-m}
=
t^{-\frac{n+|\alpha|+2m}{2}}.
$$

Esto prueba la fórmula.


### **Corolario 04.08.3 — Estimaciones en $L^1$ y $L^\infty$**

Para cada

$$
\alpha\in\mathbb N_0^n
$$

y

$$
m\in\mathbb N_0,
$$

existe una constante

$$
C_{\alpha,m,n}>0
$$

tal que

$$
\left\|
D_x^\alpha\partial_t^m\psi(\cdot,t)
\right\|_{L^1(\mathbb R^n)}
\leq
C_{\alpha,m,n}
t^{-\frac{|\alpha|+2m}{2}},
$$

y

$$
\left\|
D_x^\alpha\partial_t^m\psi(\cdot,t)
\right\|_{L^\infty(\mathbb R^n)}
\leq
C_{\alpha,m,n}
t^{-\frac{n+|\alpha|+2m}{2}}.
$$

### Demostración

Para la norma

$$
L^\infty,
$$

la afirmación se obtiene directamente del Lema 04.08.2, pues una función polinomial multiplicada por una gaussiana es acotada.

Para la norma

$$
L^1,
$$

hacemos el cambio de variables

$$
x=\sqrt t\,z.
$$

Entonces

$$
dx=t^{n/2}\,dz.
$$

Por tanto,

$$
\begin{aligned}
\left\|
D_x^\alpha\partial_t^m\psi(\cdot,t)
\right\|_{L^1}
&=
t^{-\frac{n+|\alpha|+2m}{2}}
t^{n/2}
\int_{\mathbb R^n}
\left|
P_{\alpha,m}(z)
\right|
e^{-|z|^2/4}\,dz
\\[2mm]
&=
C_{\alpha,m,n}
t^{-\frac{|\alpha|+2m}{2}}.
\end{aligned}
$$


## 04.08.4 — Suavizamiento instantáneo

### **Teorema 04.08.4 — Regularización instantánea por el núcleo del calor**

Sea

$$
g\in L^\infty(\mathbb R^n).
$$

Para

$$
t>0,
$$

definimos

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)g(y)\,dy.
$$

Entonces

$$
u\in C^\infty(\mathbb R^n\times(0,\infty)).
$$

Además, para todo multiíndice

$$
\alpha
$$

y todo entero

$$
m\geq0,
$$

se cumple

$$
D_x^\alpha\partial_t^m u(x,t)
=
\int_{\mathbb R^n}
D_x^\alpha\partial_t^m\psi(x-y,t)g(y)\,dy.
$$

Finalmente,

$$
\boxed{
\left\|
D_x^\alpha\partial_t^m u(\cdot,t)
\right\|_{L^\infty(\mathbb R^n)}
\leq
C_{\alpha,m,n}
t^{-\frac{|\alpha|+2m}{2}}
\|g\|_{L^\infty(\mathbb R^n)}
}.
$$


### Demostración

Fijemos

$$
\tau>0.
$$

En la región

$$
t\geq\tau,
$$

todas las derivadas del núcleo son integrables respecto de la variable espacial y satisfacen cotas uniformes.

Por el Corolario 04.08.3,

$$
D_x^\alpha\partial_t^m\psi(\cdot,t)
\in L^1(\mathbb R^n).
$$

Como

$$
g\in L^\infty(\mathbb R^n),
$$

la convolución

$$
D_x^\alpha\partial_t^m\psi(\cdot,t)*g
$$

está bien definida.

Los cocientes diferenciales de la convolución pueden dominarse mediante funciones integrables obtenidas a partir de derivadas de orden superior del núcleo.

El teorema de convergencia dominada permite, por tanto, diferenciar bajo el signo integral:

$$
D_x^\alpha\partial_t^m u
=
\left(
D_x^\alpha\partial_t^m\psi
\right)*g.
$$

Como esto vale para todos

$$
\alpha
$$

y

$$
m,
$$

se concluye que

$$
u\in C^\infty(\mathbb R^n\times(0,\infty)).
$$

Finalmente, por la desigualdad de Young,

$$
\begin{aligned}
\left\|
D_x^\alpha\partial_t^m u(\cdot,t)
\right\|_{L^\infty}
&\leq
\left\|
D_x^\alpha\partial_t^m\psi(\cdot,t)
\right\|_{L^1}
\|g\|_{L^\infty}
\\[2mm]
&\leq
C_{\alpha,m,n}
t^{-\frac{|\alpha|+2m}{2}}
\|g\|_{L^\infty}.
\end{aligned}
$$


## 04.08.5 — Compatibilidad con la ecuación del calor

### **Corolario 04.08.5**

Bajo las hipótesis del Teorema 04.08.4,

$$
u_t-\Delta u=0
$$

en

$$
\mathbb R^n\times(0,\infty).
$$

### Demostración

Como

$$
\psi_t-\Delta\psi=0,
$$

y la diferenciación bajo el signo integral está justificada,

$$
\begin{aligned}
u_t-\Delta u
&=
(\psi_t-\Delta\psi)*g
\\
&=
0.
\end{aligned}
$$


## 04.08.6 — Datos iniciales discontinuos

### **Ejemplo 04.08.6 — Regularización de una función indicadora**

En dimensión uno, sea

$$
g(x)=\mathbf 1_{[-a,a]}(x),
\qquad
a>0.
$$

La solución es

$$
u(x,t)
=
\int_{-a}^{a}
\frac{1}{\sqrt{4\pi t}}
\exp\!\left(
-\frac{(x-y)^2}{4t}
\right)\,dy.
$$

Introduciendo la función error

$$
\operatorname{erf}(z)
=
\frac{2}{\sqrt\pi}
\int_0^z e^{-s^2}\,ds,
$$

obtenemos

$$
u(x,t)
=
\frac12
\left[
\operatorname{erf}\!\left(
\frac{x+a}{\sqrt{4t}}
\right)
-
\operatorname{erf}\!\left(
\frac{x-a}{\sqrt{4t}}
\right)
\right].
$$

Aunque

$$
g
$$

es discontinua, para cada

$$
t>0
$$

la función

$$
u(\cdot,t)
$$

es de clase

$$
C^\infty.
$$


### **Observación 04.08.7 — Recuperación del dato en los puntos de salto**

Para

$$
|x|<a,
$$

se tiene

$$
u(x,t)\longrightarrow1
$$

cuando

$$
t\downarrow0.
$$

Para

$$
|x|>a,
$$

se tiene

$$
u(x,t)\longrightarrow0.
$$

En los puntos de discontinuidad,

$$
x=\pm a,
$$

la simetría del núcleo produce

$$
u(\pm a,t)\longrightarrow\frac12.
$$

Por tanto, el suavizamiento para

$$
t>0
$$

no implica convergencia uniforme al dato discontinuo cuando

$$
t\downarrow0.
$$


```python
import math
import numpy as np
import matplotlib.pyplot as plt

a = 1.25
x = np.linspace(-4.0, 4.0, 2200)

def erf_array(values):
    return np.array([math.erf(float(value)) for value in values])

def solution(x, t):
    if t == 0.0:
        return (np.abs(x) < a).astype(float)
    scale = np.sqrt(4.0 * t)
    return 0.5 * (
        erf_array((x + a) / scale)
        - erf_array((x - a) / scale)
    )

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.01, 0.05, 0.18, 0.60]:
    plt.plot(x, solution(x, t), linewidth=2, label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Regularización instantánea de un dato inicial discontinuo")
plt.legend()
plt.tight_layout()
plt.show()
```

## 04.08.7 — Estimaciones $L^p$–$L^q$

### **Proposición 04.08.8 — Efecto regularizante en normas de Lebesgue**

Sean

$$
1\leq p\leq q\leq\infty
$$

y

$$
g\in L^p(\mathbb R^n).
$$

Entonces

$$
u(\cdot,t)=\psi(\cdot,t)*g
$$

satisface

$$
\boxed{
\|u(\cdot,t)\|_{L^q(\mathbb R^n)}
\leq
C_{n,p,q}
t^{-\frac n2\left(\frac1p-\frac1q\right)}
\|g\|_{L^p(\mathbb R^n)}
}.
$$

> **Complemento añadido.**
> Esta estimación cuantifica que la difusión no sólo suaviza derivadas: también mejora la integrabilidad de la solución para tiempos positivos.


### Demostración

Elegimos

$$
r\in[1,\infty]
$$

de modo que

$$
1+\frac1q
=
\frac1p+\frac1r.
$$

La desigualdad de Young para convoluciones da

$$
\|u(\cdot,t)\|_{L^q}
\leq
\|\psi(\cdot,t)\|_{L^r}
\|g\|_{L^p}.
$$

Por el escalamiento del núcleo,

$$
\|\psi(\cdot,t)\|_{L^r}
=
C_{n,r}
t^{-\frac n2\left(1-\frac1r\right)}.
$$

La relación entre los exponentes implica

$$
1-\frac1r
=
\frac1p-\frac1q.
$$

Sustituyendo se obtiene el resultado.


## 04.08.8 — Regularidad del término de Duhamel

### **Proposición 04.08.9 — Suavidad lejos del tiempo inicial**

Sea

$$
h\in C_0^\infty(\mathbb R^n\times(0,\infty)).
$$

Definimos

$$
w(x,t)
=
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)h(y,s)\,dy\,ds.
$$

Entonces

$$
w\in C^\infty(\mathbb R^n\times(0,\infty)).
$$

Además,

$$
w_t-\Delta w=h.
$$

### Observación

La singularidad aparente en

$$
s=t
$$

se controla separando la integral temporal y usando la regularidad de

$$
h.
$$

La versión escrita en las notas supone

$$
h\in C_0^2
$$

y concluye la regularidad necesaria

$$
C_1^2.
$$

La hipótesis

$$
h\in C_0^\infty
$$

permite iterar el argumento y obtener suavidad de todos los órdenes.
