## 04.05.1 — Transcripción literal normalizada de las notas

### **Principio fuerte del máximo**

Sea

$$
\Omega\subset\mathbb R^n
$$

un dominio abierto, conexo y acotado.

Sea

$$
T>0
$$

una constante y sea

$$
c=c(x,t)
$$

continua en

$$
\Omega_T=\Omega\times(0,T],
$$

con

$$
c\geq 0.
$$

Supóngase que

$$
u\in C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T})
$$

satisface

$$
u_t-\Delta u+c\,u\leq 0
$$

en

$$
\Omega_T.
$$

Si existe

$$
(x_*,t_*)\in\Omega_T
$$

tal que

$$
u(x_*,t_*)
=
\sup_{\Omega_{t_*}}u
\geq 0,
$$

donde

$$
\Omega_{t_*}
=
\Omega\times(0,t_*],
$$

entonces

$$
u
$$

es constante en

$$
\Omega_{t_*}.
$$

> **Nota de transcripción.**
> Se conservan los símbolos \(c\), \((x_*,t_*)\) y \(\Omega_{t_*}\) usados en la nota. La frase final manuscrita se lee como “\(u\) es constante en \(\Omega_{t_*}\)”.


## 04.05.2 — Interpretación del resultado

El principio débil afirma que una subsolución no puede superar en el interior los valores permitidos por la frontera parabólica.

El principio fuerte es más rígido.

Si una subsolución alcanza un máximo no negativo en un punto interior del espacio-tiempo, entonces ese máximo no puede ser aislado.

La igualdad se propaga hacia todos los tiempos anteriores:

$$
0<t\leq t_*.
$$

Por tanto, una solución no constante de la ecuación del calor no puede conservar un máximo interior.

Esta dirección temporal es esencial. El resultado concluye constancia en el cilindro pasado

$$
\Omega_{t_*},
$$

no necesariamente para tiempos posteriores a

$$
t_*.
$$


## 04.05.3 — Definiciones auxiliares

### **Definición 04.05.1 — Cilindro pasado de un punto**

Sea

$$
(x_*,t_*)\in\Omega_T.
$$

El conjunto

$$
\Omega_{t_*}
=
\Omega\times(0,t_*]
$$

se denomina **cilindro pasado** asociado al tiempo

$$
t_*.
$$

---

### **Definición 04.05.2 — Conjunto de máximos**

Sea

$$
M
=
\sup_{\Omega_{t_*}}u.
$$

Definimos

$$
E
=
\left\{
(x,t)\in\Omega_{t_*}
:
u(x,t)=M
\right\}.
$$

El principio fuerte afirma que, bajo sus hipótesis,

$$
E=\Omega_{t_*}.
$$


## 04.05.4 — Forma local mediante bolas de calor

La demostración más natural usa la propiedad de las medias parabólica.

### **Lema 04.05.3 — Desigualdad de medias para subsoluciones**

Sea

$$
Q
$$

un cilindro contenido compactamente en

$$
\Omega_T.
$$

Si

$$
v\in C^{2,1}(Q)
$$

satisface

$$
v_t-\Delta v\leq 0
$$

en

$$
Q,
$$

entonces el valor de

$$
v
$$

en un punto interior no puede exceder un promedio positivo de sus valores en una región abierta situada estrictamente en el pasado.

En particular, si

$$
v\leq M
$$

y

$$
v(x_0,t_0)=M,
$$

entonces

$$
v=M
$$

en una vecindad parabólica abierta situada debajo de

$$
(x_0,t_0).
$$

> **Complemento añadido.**
> La fórmula explícita de las bolas de calor y su demostración aparecerán en un notebook posterior. Aquí se utiliza únicamente su consecuencia de positividad: un promedio positivo sólo puede ser igual al máximo cuando la función coincide con el máximo en toda la región promediada.


## 04.05.5 — Reducción del término de orden cero

### **Lema 04.05.4 — Reducción a una subsolución del calor**

Bajo las hipótesis de las notas, sea

$$
M
=
\sup_{\Omega_{t_*}}u
\geq 0.
$$

Definimos

$$
w=u-M.
$$

Entonces

$$
w\leq 0
$$

en

$$
\Omega_{t_*},
$$

y

$$
w(x_*,t_*)=0.
$$

Además,

$$
w_t-\Delta w+c\,w
=
u_t-\Delta u+c\,u-cM.
$$

Como

$$
u_t-\Delta u+c\,u\leq 0,
$$

$$
c\geq 0
$$

y

$$
M\geq 0,
$$

obtenemos

$$
w_t-\Delta w+c\,w\leq 0.
$$

En la región donde

$$
w\leq 0,
$$

se tiene

$$
c\,w\leq 0.
$$

Por consiguiente,

$$
w_t-\Delta w
\leq
-c\,w.
$$

La forma más limpia de aplicar el argumento fuerte consiste en usar directamente la versión de la desigualdad de medias para el operador con término de orden cero no negativo.

Equivalentemente, puede introducirse una transformación exponencial local que preserve el punto de máximo.


## 04.05.6 — Teorema principal

### **Teorema 04.05.5 — Principio fuerte del máximo parabólico**

Sea

$$
\Omega\subset\mathbb R^n
$$

un dominio abierto, conexo y acotado.

Sea

$$
T>0,
$$

y sea

$$
c\in C(\Omega_T)
$$

tal que

$$
c\geq 0
$$

en

$$
\Omega_T.
$$

Supóngase que

$$
u\in C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T})
$$

satisface

$$
u_t-\Delta u+c\,u\leq 0
$$

en

$$
\Omega_T.
$$

Si existe

$$
(x_*,t_*)\in\Omega_T
$$

tal que

$$
u(x_*,t_*)
=
\sup_{\Omega_{t_*}}u
=
M
\geq 0,
$$

entonces

$$
\boxed{
u\equiv M
\quad
\text{en }
\Omega_{t_*}
}.
$$


## 04.05.7 — Demostración por propagación del conjunto de máximos

### Paso 1 — El máximo se propaga localmente hacia el pasado

Definimos

$$
E
=
\left\{
(x,t)\in\Omega_{t_*}
:
u(x,t)=M
\right\}.
$$

Por hipótesis,

$$
(x_*,t_*)\in E.
$$

Sea

$$
(x_0,t_0)\in E.
$$

La versión local de la desigualdad de medias para el operador

$$
\partial_t-\Delta+c
$$

con

$$
c\geq 0
$$

implica que, si

$$
u\leq M
$$

y

$$
u(x_0,t_0)=M,
$$

entonces

$$
u=M
$$

en una vecindad parabólica abierta contenida en el pasado de

$$
(x_0,t_0).
$$

Por tanto, el conjunto de máximos se propaga localmente hacia tiempos menores.

---

### Paso 2 — Propagación a lo largo de caminos espacio-temporales

Fijemos un punto arbitrario

$$
(y,s)\in\Omega_{t_*}.
$$

Como

$$
\Omega
$$

es conexo, existe un camino continuo

$$
\gamma:[0,1]\longrightarrow\Omega
$$

que une

$$
x_*
$$

con

$$
y.
$$

Escogemos una función temporal estrictamente decreciente

$$
\tau:[0,1]\longrightarrow(0,t_*]
$$

tal que

$$
\tau(0)=t_*,
\qquad
\tau(1)=s.
$$

El camino espacio-temporal

$$
\Gamma(\sigma)
=
\bigl(\gamma(\sigma),\tau(\sigma)\bigr)
$$

permanece en

$$
\Omega_{t_*}.
$$

La imagen de

$$
\Gamma
$$

es compacta.

Puede cubrirse por un número finito de vecindades parabólicas en las que el máximo se propaga localmente hacia el pasado.

Comenzando en

$$
(x_*,t_*),
$$

y avanzando a través de esta cadena finita de vecindades, obtenemos

$$
u(y,s)=M.
$$

---

### Paso 3 — Conclusión

El punto

$$
(y,s)\in\Omega_{t_*}
$$

era arbitrario.

Por tanto,

$$
u(y,s)=M
$$

para todo

$$
(y,s)\in\Omega_{t_*}.
$$

En consecuencia,

$$
u\equiv M
$$

en

$$
\Omega_{t_*}.
$$


## 04.05.8 — Consecuencias

### **Corolario 04.05.6 — Alternativa fuerte**

Bajo las hipótesis del Teorema 04.05.5, ocurre exactamente una de las siguientes posibilidades:

1. 
   $$
   u<M
   $$
   en todos los puntos interiores del cilindro pasado;

2. 
   $$
   u\equiv M
   $$
   en
   $$
   \Omega_{t_*}.
   $$

---

### **Corolario 04.05.7 — Estricta preservación de positividad**

Sea

$$
u\in C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T})
$$

una solución de

$$
u_t-\Delta u=0
$$

en

$$
\Omega_T.
$$

Supóngase que

$$
u\geq 0
$$

en

$$
\Omega_T
$$

y que

$$
u
$$

no es idénticamente cero en el cilindro pasado.

Entonces

$$
u(x,t)>0
$$

para todo punto interior que pueda conectarse hacia el pasado con una región donde el dato sea positivo.

---

### **Corolario 04.05.8 — Rigidez de un máximo interior**

Si una solución no constante de

$$
u_t-\Delta u=0
$$

alcanza su máximo absoluto en el interior, dicho máximo sólo puede ocurrir en el tiempo inicial.

En particular, para todo

$$
t>0,
$$

el máximo espacial de una solución no constante es estrictamente menor que el máximo inicial, salvo que la solución sea constante.


## 04.05.9 — Ejemplo explícito

Sea

$$
\Omega=(0,1)
$$

y consideremos

$$
u(x,t)
=
\frac12
+
\frac13 e^{-\pi^2t}\sin(\pi x).
$$

Entonces

$$
u_t-u_{xx}=0.
$$

Para

$$
t=0,
$$

el máximo es

$$
\frac56.
$$

Para todo

$$
t>0,
$$

se tiene

$$
\max_{x\in[0,1]}u(x,t)
=
\frac12+\frac13e^{-\pi^2t}
<
\frac56.
$$

La solución no es constante, y su máximo decrece estrictamente.

En cambio, la función

$$
v(x,t)=M
$$

es constante y conserva su máximo en todos los puntos.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1200)
times = [0.0, 0.04, 0.10, 0.22, 0.45]

def u(x, t):
    return 0.5 + (1.0 / 3.0) * np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)

plt.figure(figsize=(10, 6), dpi=160)

for t in times:
    plt.plot(x, u(x, t), linewidth=2, label=f"t = {t}")

plt.axhline(5.0 / 6.0, linestyle="--", linewidth=1.5, label="máximo inicial")
plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Descenso estricto del máximo para una solución no constante")
plt.legend()
plt.tight_layout()
plt.show()
```
