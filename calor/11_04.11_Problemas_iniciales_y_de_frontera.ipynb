## 04.11.1 — Transcripción literal normalizada de las notas

Sea

$$
\Omega\subset\mathbb R^n
$$

abierto y acotado, con frontera regular por pedazos. Sea

$$
T>0
$$

y definamos

$$
\Omega_T=\Omega\times(0,T].
$$

La frontera parabólica es

$$
\Gamma_T
=
(\partial\Omega\times(0,T])
\cup
(\overline\Omega\times\{0\}).
$$

Las notas consideran el problema

$$
\begin{cases}
u_t-\Delta u=f,
&
\text{en } \Omega_T,
\\[2mm]
u=g,
&
\text{en } \Gamma_T,
\end{cases}
$$

con

$$
f\in C(\overline{\Omega_T}),
\qquad
g\in C(\Gamma_T).
$$

En las notas se afirma, como consecuencia del principio del máximo, que si existe una solución clásica, entonces es única.


## 04.11.2 — Datos iniciales y laterales

### **Definición 04.11.1 — Dato inicial**

El dato inicial es una función

$$
u_0:\overline\Omega\to\mathbb R
$$

prescrita mediante

$$
u(x,0)=u_0(x).
$$

### **Definición 04.11.2 — Dato de Dirichlet**

El dato lateral de Dirichlet es

$$
u(x,t)=g_D(x,t)
$$

sobre

$$
\partial\Omega\times(0,T].
$$

### **Definición 04.11.3 — Dato de Neumann**

El dato lateral de Neumann es

$$
\frac{\partial u}{\partial\nu}(x,t)=g_N(x,t),
$$

donde

$$
\nu
$$

es la normal exterior unitaria.


## 04.11.3 — Problema inicial de Dirichlet

### **Definición 04.11.4 — Problema inicial de Dirichlet**

Dadas

$$
f,
\qquad
u_0,
\qquad
g_D,
$$

el problema inicial de Dirichlet es

$$
\begin{cases}
u_t-\Delta u=f,
&
\text{en } \Omega\times(0,T],
\\[2mm]
u=g_D,
&
\text{sobre } \partial\Omega\times(0,T],
\\[2mm]
u(\cdot,0)=u_0,
&
\text{en } \Omega.
\end{cases}
$$

Una solución clásica pertenece a

$$
C^{2,1}(\Omega\times(0,T])
\cap
C(\overline\Omega\times[0,T]).
$$


## 04.11.4 — Condiciones de compatibilidad

### **Proposición 04.11.5 — Compatibilidad de orden cero**

Si existe una solución continua hasta

$$
\partial\Omega\times\{0\},
$$

entonces necesariamente

$$
\boxed{
u_0(x)=g_D(x,0)
\qquad
\forall\,x\in\partial\Omega
}.
$$

### Demostración

En un punto

$$
x_0\in\partial\Omega,
$$

la continuidad exige simultáneamente

$$
u(x_0,0)=u_0(x_0)
$$

y

$$
u(x_0,0)=g_D(x_0,0).
$$

Por tanto,

$$
u_0(x_0)=g_D(x_0,0).
$$

### **Observación 04.11.6 — Compatibilidad de primer orden**

Si se exige que la ecuación sea clásica hasta la esquina parabólica, formalmente debe cumplirse

$$
\boxed{
\partial_t g_D(x,0)
=
\Delta u_0(x)+f(x,0)
}
$$

sobre

$$
\partial\Omega,
$$

si todas estas expresiones tienen sentido.


## 04.11.5 — Unicidad de Dirichlet

### **Teorema 04.11.7 — Unicidad**

Dos soluciones clásicas del mismo problema inicial de Dirichlet coinciden.

### Demostración

Si

$$
w=u-v,
$$

entonces

$$
w_t-\Delta w=0,
$$

$$
w=0
$$

en la frontera lateral y

$$
w(\cdot,0)=0.
$$

El principio débil del máximo aplicado a

$$
w
$$

y a

$$
-w
$$

produce

$$
w=0.
$$


## 04.11.6 — Problema inicial de Neumann

### **Definición 04.11.8 — Problema inicial de Neumann**

El problema inicial de Neumann es

$$
\begin{cases}
u_t-\Delta u=f,
&
\text{en } \Omega\times(0,T],
\\[2mm]
\dfrac{\partial u}{\partial\nu}=g_N,
&
\text{sobre } \partial\Omega\times(0,T],
\\[3mm]
u(\cdot,0)=u_0,
&
\text{en } \Omega.
\end{cases}
$$

### **Proposición 04.11.9 — Compatibilidad inicial**

Si la solución es de clase

$$
C^1
$$

hasta la esquina, entonces

$$
\boxed{
\frac{\partial u_0}{\partial\nu}(x)=g_N(x,0)
}
$$

sobre

$$
\partial\Omega.
$$


## 04.11.7 — Balance de masa

### **Proposición 04.11.10 — Balance de masa**

Definimos

$$
M(t)=\int_\Omega u(x,t)\,dx.
$$

Entonces

$$
\boxed{
M'(t)
=
\int_\Omega f(x,t)\,dx
+
\int_{\partial\Omega}g_N(x,t)\,dS
}.
$$

### Demostración

Integrando la ecuación,

$$
M'(t)
=
\int_\Omega\Delta u\,dx
+
\int_\Omega f\,dx.
$$

El teorema de la divergencia da

$$
\int_\Omega\Delta u\,dx
=
\int_{\partial\Omega}
\frac{\partial u}{\partial\nu}\,dS
=
\int_{\partial\Omega}g_N\,dS.
$$

### **Corolario 04.11.11 — Conservación de masa**

Si

$$
f=0
$$

y

$$
g_N=0,
$$

entonces

$$
M(t)=M(0).
$$

### **Observación 04.11.12**

El problema parabólico de Neumann no requiere una condición integral estática como el problema elíptico. La relación correcta es la ley dinámica de balance.


## 04.11.8 — Unicidad de Neumann

### **Teorema 04.11.13 — Unicidad con dato inicial fijado**

Dos soluciones clásicas del mismo problema inicial de Neumann coinciden.

### Demostración

Para

$$
w=u-v
$$

se tiene

$$
w_t-\Delta w=0,
$$

$$
\frac{\partial w}{\partial\nu}=0,
$$

y

$$
w(\cdot,0)=0.
$$

Multiplicando por

$$
w
$$

e integrando,

$$
\frac{d}{dt}
\left(
\frac12\int_\Omega w^2\,dx
\right)
+
\int_\Omega|\nabla w|^2\,dx
=
0.
$$

La energía inicial es cero, por lo que

$$
w=0.
$$


## 04.11.9 — Homogeneización de la frontera

### **Proposición 04.11.14 — Levantamiento de Dirichlet**

Supongamos que

$$
\ell=g_D
$$

sobre

$$
\partial\Omega\times[0,T].
$$

Definimos

$$
v=u-\ell.
$$

Entonces

$$
v=0
$$

en la frontera lateral y

$$
\boxed{
v_t-\Delta v
=
f-\ell_t+\Delta\ell
},
$$

con dato inicial

$$
\boxed{
v(\cdot,0)
=
u_0-\ell(\cdot,0)
}.
$$


## 04.11.10 — Levantamiento explícito en un intervalo

Sea

$$
\Omega=(0,L)
$$

y supongamos

$$
u(0,t)=g_0(t),
\qquad
u(L,t)=g_L(t).
$$

Definimos

$$
\boxed{
\ell(x,t)
=
\left(1-\frac{x}{L}\right)g_0(t)
+
\frac{x}{L}g_L(t)
}.
$$

Entonces

$$
\ell_{xx}=0,
$$

y

$$
v=u-\ell
$$

satisface frontera homogénea:

$$
v(0,t)=v(L,t)=0.
$$

La ecuación transformada es

$$
v_t-v_{xx}=f-\ell_t.
$$


### **Ejemplo 04.11.15 — Frontera dependiente del tiempo**

Considérese

$$
\begin{cases}
u_t-u_{xx}=x e^{-t},
&
0<x<1,\quad t>0,
\\[2mm]
u(0,t)=0,
\\[2mm]
u(1,t)=1-e^{-t},
\\[2mm]
u(x,0)=\sin(\pi x).
\end{cases}
$$

Tomamos

$$
\ell(x,t)=x(1-e^{-t}).
$$

Entonces

$$
\ell_t=x e^{-t},
\qquad
\ell_{xx}=0.
$$

La función

$$
v=u-\ell
$$

satisface

$$
v_t-v_{xx}=0,
$$

$$
v(0,t)=v(1,t)=0,
$$

y

$$
v(x,0)=\sin(\pi x).
$$

La solución elemental es

$$
v(x,t)=e^{-\pi^2t}\sin(\pi x).
$$

Por tanto,

$$
\boxed{
u(x,t)
=
x(1-e^{-t})
+
e^{-\pi^2t}\sin(\pi x)
}.
$$

> **Aclaración.**
> La solución se reconoce directamente porque es un solo modo. La construcción general mediante series se realizará en el notebook siguiente.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1400)

def solution(x, t):
    return (
        x * (1.0 - np.exp(-t))
        + np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)
    )

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.08, 0.25, 0.60, 1.25]:
    plt.plot(x, solution(x, t), linewidth=2, label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Problema de Dirichlet con frontera dependiente del tiempo")
plt.legend()
plt.tight_layout()
plt.show()
```

## 04.11.11 — Condiciones de Robin

### **Definición 04.11.16 — Condición de Robin**

Una condición de Robin tiene la forma

$$
a(x,t)u
+
b(x,t)
\frac{\partial u}{\partial\nu}
=
g_R(x,t).
$$

Dirichlet corresponde a

$$
b=0,
$$

y Neumann a

$$
a=0.
$$
