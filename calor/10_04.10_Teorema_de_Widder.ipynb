## 04.10.1 — Transcripción literal normalizada de las notas

### **Teorema 04.10.1 — Teorema de Widder**

En las notas aparece el siguiente enunciado.

Sea

$$
u\in C^{2,1}(\Omega_T)
$$

solución de

$$
\begin{cases}
u_t-\Delta u=f,
&
x\in\mathbb R^n,\quad t>0,
\\[2mm]
u=g,
&
t=0.
\end{cases}
$$

Si

$$
u\geq0,
$$

entonces

$$
u
$$

es única.

---

### **Lema 04.10.2**

Las notas registran además:

sea

$$
u\in
C^\infty(\mathbb R^n\times(0,\infty))
\cap
C(\mathbb R^n\times[0,\infty))
$$

una solución no negativa de

$$
u_t=\Delta u.
$$

Entonces

$$
u(x,t)
\geq
\int_{\mathbb R^n}
\psi(x-y,t)u(y,0)\,dy.
$$

> **Nota de transcripción.**
> En el sistema manuscrito del teorema se lee el término \(f\), pero la hoja no muestra hipótesis adicionales sobre \(f\). El lema escrito inmediatamente debajo es homogéneo. La formulación clásica del teorema de Widder se refiere precisamente a soluciones no negativas de la ecuación homogénea del calor. Ambas capas se mantienen separadas en este notebook.


## 04.10.2 — Aclaración sobre el alcance del enunciado manuscrito

La no negatividad por sí sola no produce, a partir de las hipótesis visibles, una afirmación completa de unicidad para un problema no homogéneo arbitrario

$$
u_t-\Delta u=f.
$$

Para una fuente se necesitan hipótesis adicionales y una fórmula de representación que incluya el término de Duhamel.

La versión estándar que completa el subtema oficial es:

> toda solución no negativa de la ecuación homogénea del calor posee una representación por el núcleo del calor y una medida inicial no negativa.

Cuando la solución tiene un dato inicial continuo

$$
g,
$$

la medida inicial es

$$
g(y)\,dy,
$$

y la representación determina la solución de manera única.


## 04.10.3 — El lema de cota inferior

### **Lema 04.10.3 — Cota inferior por el dato inicial**

Sea

$$
u\in
C^{2,1}(\mathbb R^n\times(0,T))
\cap
C(\mathbb R^n\times[0,T))
$$

una solución no negativa de

$$
u_t-\Delta u=0.
$$

Entonces, para todo

$$
x\in\mathbb R^n
$$

y

$$
0<t<T,
$$

se cumple

$$
\boxed{
u(x,t)
\geq
\int_{\mathbb R^n}
\psi(x-y,t)u(y,0)\,dy
}.
$$


### Demostración

Fijemos

$$
R>0
$$

y tomemos una función

$$
\eta_R\in C_c^\infty(\mathbb R^n)
$$

tal que

$$
0\leq\eta_R\leq1,
$$

$$
\eta_R=1
\qquad
\text{en } B_R(0),
$$

y

$$
\operatorname{supp}\eta_R
\subset B_{2R}(0).
$$

Definimos

$$
v_R(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)
\eta_R(y)u(y,0)\,dy.
$$

La función

$$
v_R
$$

satisface

$$
(v_R)_t-\Delta v_R=0
$$

y tiene dato inicial

$$
v_R(x,0)
=
\eta_R(x)u(x,0)
\leq
u(x,0).
$$

Como el dato inicial de

$$
v_R
$$

tiene soporte compacto, se cumple

$$
v_R(x,t)\longrightarrow0
$$

uniformemente para

$$
0\leq t\leq t_0
$$

cuando

$$
|x|\longrightarrow\infty,
$$

para cada

$$
t_0<T.
$$

Fijemos

$$
t_0<T
$$

y sea

$$
\varepsilon>0.
$$

Existe

$$
M>0
$$

tal que

$$
v_R(x,t)\leq\varepsilon
$$

si

$$
|x|=M,
\qquad
0\leq t\leq t_0.
$$

Como

$$
u\geq0,
$$

tenemos

$$
v_R-u-\varepsilon\leq0
$$

sobre la frontera lateral del cilindro

$$
B_M(0)\times(0,t_0].
$$

En el tiempo inicial,

$$
v_R(x,0)-u(x,0)-\varepsilon\leq0.
$$

El principio débil del máximo aplicado a

$$
v_R-u-\varepsilon
$$

da

$$
v_R(x,t)\leq u(x,t)+\varepsilon
$$

en el cilindro.

Haciendo

$$
\varepsilon\downarrow0,
$$

obtenemos

$$
v_R(x,t)\leq u(x,t).
$$

Finalmente, como

$$
\eta_R\uparrow1
$$

y todos los integrandos son no negativos, el teorema de convergencia monótona produce

$$
\int_{\mathbb R^n}
\psi(x-y,t)u(y,0)\,dy
\leq
u(x,t).
$$


## 04.10.4 — Traza inicial como medida

### **Definición 04.10.4 — Traza inicial de Widder**

Sea

$$
u\geq0
$$

una solución calórica en

$$
\mathbb R^n\times(0,T).
$$

Una medida de Borel no negativa

$$
\mu
$$

se denomina **traza inicial de Widder** de

$$
u
$$

si

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,d\mu(y)
$$

para

$$
0<t<T.
$$


## 04.10.5 — Teorema de representación

### **Teorema 04.10.5 — Teorema de representación de Widder**

Sea

$$
T\in(0,\infty].
$$

Supongamos que

$$
u\in C^{2,1}(\mathbb R^n\times(0,T))
$$

satisface

$$
u_t-\Delta u=0
$$

y

$$
u\geq0
$$

en

$$
\mathbb R^n\times(0,T).
$$

Entonces existe una única medida de Borel no negativa

$$
\mu
$$

sobre

$$
\mathbb R^n
$$

tal que

$$
\boxed{
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,d\mu(y)
}
$$

para todo

$$
x\in\mathbb R^n,
\qquad
0<t<T.
$$

La medida satisface la condición de integrabilidad necesaria para que la integral sea finita:

$$
\int_{\mathbb R^n}
e^{-\frac{|y|^2}{4t}}
\,d\mu(y)
<
\infty
$$

para cada

$$
0<t<T.
$$


## 04.10.6 — Lectura guiada de la demostración de Widder

La demostración completa es medida-teórica. Su estructura es la siguiente.

### Paso 1 — Medidas en tiempos positivos

Para

$$
\tau>0,
$$

definimos

$$
d\mu_\tau(y)
=
u(y,\tau)\,dy.
$$

El lema anterior, aplicado después de trasladar el origen temporal, da

$$
u(x,t)
\geq
\int_{\mathbb R^n}
\psi(x-y,t-\tau)
\,d\mu_\tau(y),
$$

si

$$
0<\tau<t<T.
$$

---

### Paso 2 — Compacidad local de las medidas

Fijemos

$$
(x_0,t_0)\in\mathbb R^n\times(0,T).
$$

Para

$$
0<\tau<t_0/2,
$$

la desigualdad anterior implica

$$
u(x_0,t_0)
\geq
\int_{\mathbb R^n}
\psi(x_0-y,t_0-\tau)
\,d\mu_\tau(y).
$$

Sobre cada compacto

$$
K\subset\mathbb R^n,
$$

el núcleo posee una cota inferior positiva uniforme:

$$
\psi(x_0-y,t_0-\tau)\geq c_K>0.
$$

Por tanto,

$$
\mu_\tau(K)
\leq
\frac{u(x_0,t_0)}{c_K}.
$$

La familia

$$
\{\mu_\tau\}
$$

es localmente acotada.

---

### Paso 3 — Extracción de una medida límite

Por compacidad débil de medidas en compactos, existe una sucesión

$$
\tau_j\downarrow0
$$

y una medida no negativa

$$
\mu
$$

tal que

$$
\mu_{\tau_j}
\rightharpoonup
\mu
$$

vagamente.

---

### Paso 4 — Paso al límite en la representación

Las cotas gaussianas obtenidas en el Paso 2 controlan las colas de las integrales.

Al hacer

$$
\tau_j\downarrow0
$$

en las fórmulas con núcleo desplazado se obtiene

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,d\mu(y).
$$

El paso esencial de Widder consiste en demostrar que no queda un resto calórico no negativo con traza inicial nula.

---

### Paso 5 — Unicidad de la medida

Supongamos que

$$
\mu
$$

y

$$
\nu
$$

producen la misma solución.

Para toda función de prueba

$$
\varphi\in C_c^\infty(\mathbb R^n),
$$

se tiene

$$
\int_{\mathbb R^n}
\varphi(x)
u(x,t)\,dx
=
\int_{\mathbb R^n}
(\psi(\cdot,t)*\varphi)(y)
\,d\mu(y).
$$

Cuando

$$
t\downarrow0,
$$

$$
\psi(\cdot,t)*\varphi
\longrightarrow
\varphi
$$

uniformemente.

Por tanto,

$$
\int_{\mathbb R^n}
\varphi\,d\mu
=
\int_{\mathbb R^n}
\varphi\,d\nu.
$$

Luego

$$
\mu=\nu.
$$

> **Complemento añadido.**
> El Paso 4 contiene el núcleo profundo del teorema. La positividad permite usar convergencia monótona, compactitud de medidas y estimaciones gaussianas; esos mecanismos no están disponibles para las soluciones de signo cambiante de Tychonoff.


## 04.10.7 — Dato inicial continuo

### **Corolario 04.10.6 — Representación por el dato inicial**

Supongamos, además de las hipótesis del Teorema 04.10.5, que

$$
u
$$

se extiende continuamente a

$$
\mathbb R^n\times[0,T)
$$

y escribamos

$$
g(x)=u(x,0).
$$

Entonces

$$
d\mu(y)=g(y)\,dy,
$$

y por tanto

$$
\boxed{
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)g(y)\,dy
}.
$$

### Demostración

Para toda

$$
\varphi\in C_c^\infty(\mathbb R^n),
$$

la continuidad hasta

$$
t=0
$$

implica

$$
\lim_{t\downarrow0}
\int_{\mathbb R^n}
\varphi(x)u(x,t)\,dx
=
\int_{\mathbb R^n}
\varphi(x)g(x)\,dx.
$$

Por otro lado, la representación de Widder da

$$
\lim_{t\downarrow0}
\int_{\mathbb R^n}
\varphi(x)u(x,t)\,dx
=
\int_{\mathbb R^n}
\varphi(y)\,d\mu(y).
$$

En consecuencia,

$$
\int_{\mathbb R^n}
\varphi\,d\mu
=
\int_{\mathbb R^n}
\varphi(y)g(y)\,dy
$$

para toda función de prueba.

Por tanto,

$$
d\mu=g\,dy.
$$


## 04.10.8 — Unicidad en la clase no negativa

### **Corolario 04.10.7 — Unicidad de Widder**

Sea

$$
g\in C(\mathbb R^n)
$$

con

$$
g\geq0.
$$

Dentro de la clase de soluciones no negativas que admiten traza inicial continua

$$
g,
$$

el problema

$$
\begin{cases}
u_t-\Delta u=0,
&
x\in\mathbb R^n,\quad t>0,
\\[2mm]
u(x,0)=g(x),
&
x\in\mathbb R^n,
\end{cases}
$$

tiene a lo sumo una solución.

Esa solución es

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)g(y)\,dy.
$$


## 04.10.9 — Relación con Tychonoff

### **Corolario 04.10.8 — No existen soluciones de Tychonoff no negativas**

Sea

$$
u\geq0
$$

una solución de

$$
u_t-\Delta u=0
$$

en

$$
\mathbb R^n\times(0,\infty),
$$

continua hasta

$$
t=0,
$$

con

$$
u(x,0)=0.
$$

Entonces

$$
u\equiv0.
$$

### Demostración

Por el Corolario 04.10.6,

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)u(y,0)\,dy
=
0.
$$

Por tanto, toda solución no trivial de Tychonoff debe cambiar de signo o escapar de las hipótesis de representación positiva.

---

### **Observación 04.10.9**

La no unicidad de Tychonoff y el teorema de Widder no se contradicen.

Tychonoff muestra que la regularidad clásica, sin ninguna condición adicional, no basta para la unicidad global.

Widder muestra que la hipótesis

$$
u\geq0
$$

es suficientemente fuerte para recuperar una representación y, por tanto, unicidad.


## 04.10.10 — Ejemplo gaussiano

Sea

$$
g(x)=e^{-x^2}
$$

en dimensión uno.

La representación de Widder es

$$
u(x,t)
=
\int_{\mathbb R}
\frac{1}{\sqrt{4\pi t}}
e^{-\frac{(x-y)^2}{4t}}
e^{-y^2}\,dy.
$$

Completando cuadrados se obtiene

$$
\boxed{
u(x,t)
=
\frac{1}{\sqrt{1+4t}}
\exp\!\left(
-\frac{x^2}{1+4t}
\right)
}.
$$

La solución permanece no negativa y queda completamente determinada por el dato inicial.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4.0, 4.0, 1800)

def widder_gaussian(x, t):
    denominator = 1.0 + 4.0 * t
    return np.exp(-(x**2) / denominator) / np.sqrt(denominator)

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.08, 0.25, 0.60, 1.20]:
    plt.plot(
        x,
        widder_gaussian(x, t),
        linewidth=2,
        label=f"t = {t}",
    )

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Representación de Widder para un dato inicial gaussiano")
plt.legend()
plt.tight_layout()
plt.show()
```
