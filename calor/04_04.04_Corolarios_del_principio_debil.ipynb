## 04.04.3 — Principio de comparación sin término de orden cero

### **Proposición 04.04.4 — Principio de comparación básico**

Sean

$$
u,v\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T).
$$

Supóngase que

$$
u_t-\Delta u
\leq
v_t-\Delta v
\qquad
\text{en } \Omega_T,
$$

y que

$$
u\leq v
\qquad
\text{en } \Gamma_T.
$$

Entonces

$$
u\leq v
\qquad
\text{en } \Omega_T.
$$

### Demostración

Definimos

$$
w=u-v.
$$

Entonces

$$
w\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T),
$$

$$
w_t-\Delta w
=
(u_t-\Delta u)-(v_t-\Delta v)
\leq 0
\qquad
\text{en } \Omega_T,
$$

y además

$$
w\leq 0
\qquad
\text{en } \Gamma_T.
$$

Aplicando el principio débil del máximo al campo

$$
w,
$$

obtenemos

$$
\max_{\overline{\Omega_T}}w
=
\max_{\Gamma_T}w
\leq 0.
$$

Por tanto,

$$
w\leq 0
\qquad
\text{en } \Omega_T,
$$

es decir,

$$
u\leq v
\qquad
\text{en } \Omega_T.
$$


## 04.04.4 — Unicidad del problema de Dirichlet parabólico

### **Corolario 04.04.5 — Unicidad**

Sean

$$
f\in C(\overline{\Omega_T}),
\qquad
g\in C(\Gamma_T).
$$

Si el problema

$$
\begin{cases}
u_t-\Delta u=f
&
\text{en } \Omega_T,
\\[2mm]
u=g
&
\text{en } \Gamma_T
\end{cases}
$$

admite una solución clásica

$$
u\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T),
$$

entonces dicha solución es única.

### Demostración

Sean

$$
u_1
\qquad
\text{y}
\qquad
u_2
$$

dos soluciones del problema.

Definimos

$$
w=u_1-u_2.
$$

Entonces

$$
w_t-\Delta w=0
\qquad
\text{en } \Omega_T,
$$

y

$$
w=0
\qquad
\text{en } \Gamma_T.
$$

Aplicando la Proposición 04.04.4 a

$$
w
\qquad
\text{y}
\qquad
0,
$$

se obtiene

$$
w\leq 0
\qquad
\text{en } \Omega_T.
$$

Repitiendo el argumento con

$$
-w,
$$

resulta

$$
-w\leq 0
\qquad
\text{en } \Omega_T.
$$

Luego

$$
w=0
\qquad
\text{en } \Omega_T,
$$

y por tanto

$$
u_1=u_2.
$$


## 04.04.5 — Variante con término de orden cero no negativo

### **Teorema 04.04.6 — Principio de comparación con $h\geq 0$**

Sea

$$
h\in C(\Omega_T)
$$

con

$$
h\geq 0
\qquad
\text{en } \Omega_T.
$$

Supóngase que

$$
u\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T)
$$

satisface

$$
u_t-\Delta u+h\,u\leq 0
\qquad
\text{en } \Omega_T,
$$

y

$$
u\leq 0
\qquad
\text{en } \Gamma_T.
$$

Entonces

$$
u\leq 0
\qquad
\text{en } \Omega_T.
$$

### Demostración

Sea

$$
A=\{(x,t)\in \Omega_T: u(x,t)>0\}.
$$

Si

$$
A=\varnothing,
$$

no hay nada que probar.

Supongamos entonces que

$$
A\neq \varnothing.
$$

En

$$
A
$$

tenemos

$$
u>0,
\qquad
h\geq 0,
$$

y por consiguiente

$$
h\,u\geq 0.
$$

De la desigualdad

$$
u_t-\Delta u+h\,u\leq 0
$$

se deduce

$$
u_t-\Delta u\leq 0
\qquad
\text{en } A.
$$

Además, en la frontera parabólica de

$$
A,
$$

la función

$$
u
$$

satisface

$$
u\leq 0.
$$

En efecto, sobre la parte de la frontera contenida en

$$
\Gamma_T
$$

esto es una hipótesis, y sobre la parte interior de la frontera de

$$
A
$$

se tiene necesariamente

$$
u=0
$$

por definición de

$$
A.
$$

Aplicando el principio débil del máximo a

$$
u
$$

en el dominio

$$
A,
$$

obtenemos

$$
u\leq 0
\qquad
\text{en } A,
$$

lo cual contradice la definición de

$$
A.
$$

Concluimos que

$$
A=\varnothing,
$$

y por tanto

$$
u\leq 0
\qquad
\text{en } \Omega_T.
$$


## 04.04.6 — Variante con coeficiente acotado inferiormente

### **Teorema 04.04.7 — Principio de comparación con $h\geq -c_0$**

Sea

$$
h\in C(\Omega_T)
$$

y supóngase que existe

$$
c_0>0
$$

tal que

$$
h(x,t)\geq -c_0
\qquad
\text{para todo } (x,t)\in \Omega_T.
$$

Si

$$
u\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T)
$$

satisface

$$
u_t-\Delta u+h\,u\leq 0
\qquad
\text{en } \Omega_T,
$$

y

$$
u\leq 0
\qquad
\text{en } \Gamma_T,
$$

entonces

$$
u\leq 0
\qquad
\text{en } \Omega_T.
$$

### Demostración

Definimos

$$
v(x,t)=e^{-c_0 t}u(x,t).
$$

Como

$$
u(x,t)=e^{c_0 t}v(x,t),
$$

obtenemos

$$
u_t=e^{c_0 t}(v_t+c_0 v),
\qquad
\Delta u=e^{c_0 t}\Delta v.
$$

Sustituyendo en la desigualdad de hipótesis:

$$
e^{c_0 t}(v_t+c_0 v)-e^{c_0 t}\Delta v+h\,e^{c_0 t}v\leq 0.
$$

Dividiendo por

$$
e^{c_0 t}>0,
$$

resulta

$$
v_t-\Delta v+(h+c_0)v\leq 0.
$$

Por la hipótesis

$$
h\geq -c_0,
$$

se tiene

$$
h+c_0\geq 0
\qquad
\text{en } \Omega_T.
$$

Además, como

$$
u\leq 0
\qquad
\text{en } \Gamma_T,
$$

también

$$
v=e^{-c_0 t}u\leq 0
\qquad
\text{en } \Gamma_T.
$$

Aplicando el Teorema 04.04.6 a

$$
v,
$$

concluimos que

$$
v\leq 0
\qquad
\text{en } \Omega_T.
$$

Multiplicando por

$$
e^{c_0 t}>0,
$$

obtenemos finalmente

$$
u\leq 0
\qquad
\text{en } \Omega_T.
$$


## 04.04.7 — Estimación de dependencia respecto de los datos

### **Corolario 04.04.8 — Dependencia continua en norma supremo**

Sean

$$
u
\qquad
\text{y}
\qquad
v
$$

dos soluciones clásicas de

$$
u_t-\Delta u=f,
\qquad
v_t-\Delta v=g
$$

en

$$
\Omega_T,
$$

con datos de frontera parabólica

$$
u=\varphi,
\qquad
v=\psi
\qquad
\text{en } \Gamma_T.
$$

Si existen constantes

$$
M_f,
\qquad
M_g
$$

tales que

$$
|f-g|\leq M
\qquad
\text{en } \Omega_T,
$$

entonces la diferencia

$$
w=u-v
$$

satisface

$$
w_t-\Delta w=f-g.
$$

En particular, cuando

$$
f=g,
$$

se obtiene la estimación de estabilidad

$$
\|u-v\|_{L^\infty(\Omega_T)}
\leq
\|\,\varphi-\psi\,\|_{L^\infty(\Gamma_T)}.
$$

### Demostración

Si

$$
f=g,
$$

entonces

$$
w_t-\Delta w=0
\qquad
\text{en } \Omega_T,
$$

y

$$
w=\varphi-\psi
\qquad
\text{en } \Gamma_T.
$$

Aplicando el principio de comparación a

$$
w
\qquad
\text{y las constantes } \pm \|\varphi-\psi\|_{L^\infty(\Gamma_T)},
$$

se deduce

$$
-\|\varphi-\psi\|_{L^\infty(\Gamma_T)}
\leq
w
\leq
\|\varphi-\psi\|_{L^\infty(\Gamma_T)}.
$$

Tomando supremo, concluimos la estimación.


## 04.04.8 — Visualización estática

La siguiente celda muestra dos soluciones explícitas del problema del calor en el intervalo

$$
(0,1)
$$

con frontera nula y datos iniciales ordenados.

La gráfica ilustra el principio de comparación:

si inicialmente

$$
u(x,0)\leq v(x,0),
$$

entonces

$$
u(x,t)\leq v(x,t)
$$

para todo tiempo positivo.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1200)

def u_profile(x, t):
    return 0.55 * np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)

def v_profile(x, t):
    return (
        0.90 * np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)
        + 0.12 * np.exp(-(4.0 * np.pi**2) * t) * np.sin(2.0 * np.pi * x)
    )

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.08, 0.18, 0.32, 0.45]:
    plt.plot(x, u_profile(x, t), linewidth=2, label=f"u, t={t}")
    plt.plot(x, v_profile(x, t), linewidth=2, linestyle="--", label=f"v, t={t}")

plt.xlabel("x")
plt.ylabel("amplitud")
plt.title("Principio de comparación para dos soluciones explícitas")
plt.legend(ncol=2, fontsize=8)
plt.tight_layout()
plt.show()
```
