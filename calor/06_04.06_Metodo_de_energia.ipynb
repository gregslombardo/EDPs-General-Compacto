## 04.06.1 — Transcripción literal normalizada de las notas

### **Método de energía**

En las notas aparece el siguiente planteamiento.

Sea

$$
\Omega\subset\mathbb R^n
$$

abierto y acotado, con

$$
\partial\Omega\in C^1
$$

por pedazos. Sea

$$
\Omega_T=\Omega\times(0,T].
$$

Considérese el problema

$$
\begin{cases}
u_t-\Delta u=f
&
\text{en } \Omega_T,
\\[2mm]
u=g
&
\text{en } \Gamma_T
=
\overline{\Omega_T}\setminus\Omega_T,
\end{cases}
$$

con

$$
f\in C(\overline{\Omega_T}),
\qquad
g\in C(\Gamma_T).
$$

> **Nota de transcripción.**
> El bloque visible en la nota establece el dominio, la regularidad de la frontera y el problema de contorno. A partir de ese bloque se desarrolla el método de energía como herramienta de unicidad y estimación.


## 04.06.2 — Idea del método

El método de energía consiste en multiplicar la ecuación por la propia solución, o por la diferencia de dos soluciones, e integrar sobre el dominio espacial.

De esta manera aparece naturalmente la cantidad

$$
E(t)
=
\frac12
\int_{\Omega}
u(x,t)^2\,dx,
$$

o, para la diferencia

$$
w=u-v,
$$

la energía relativa

$$
E_w(t)
=
\frac12
\int_{\Omega}
w(x,t)^2\,dx.
$$

El principio básico es que, para el problema homogéneo con datos de frontera homogéneos, la energía no aumenta.

En muchos casos, de hecho, se verifica

$$
E'(t)\leq 0.
$$


## 04.06.3 — Preliminares analíticos

### **Lema 04.06.1 — Fórmula de Green para el Laplaciano**

Sea

$$
\Omega\subset\mathbb R^n
$$

un dominio acotado con frontera regular por pedazos. Si

$$
w\in C^2(\Omega)\cap C^1(\overline\Omega),
$$

entonces

$$
\int_{\Omega}
w\,\Delta w\,dx
=
\int_{\partial\Omega}
w\,\frac{\partial w}{\partial \nu}\,dS
-
\int_{\Omega}
|\nabla w|^2\,dx.
$$

En particular, si

$$
w=0
\qquad
\text{sobre } \partial\Omega,
$$

se obtiene

$$
\int_{\Omega}
w\,\Delta w\,dx
=
-
\int_{\Omega}
|\nabla w|^2\,dx.
$$


### **Lema 04.06.2 — Derivada temporal de la energía**

Sea

$$
w\in C^1([0,T];L^2(\Omega)).
$$

Entonces la función

$$
E_w(t)
=
\frac12
\int_{\Omega}
w(x,t)^2\,dx
$$

es diferenciable y satisface

$$
E_w'(t)
=
\int_{\Omega}
w(x,t)\,w_t(x,t)\,dx.
$$


## 04.06.4 — Identidad de energía homogénea

### **Proposición 04.06.3 — Identidad de energía**

Sea

$$
w\in C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T})
$$

una solución de

$$
\begin{cases}
w_t-\Delta w=0
&
\text{en } \Omega_T,
\\[2mm]
w=0
&
\text{en } \partial\Omega\times(0,T].
\end{cases}
$$

Entonces, para todo

$$
t\in(0,T],
$$

se cumple

$$
\frac{d}{dt}
\left(
\frac12
\int_{\Omega}
w(x,t)^2\,dx
\right)
+
\int_{\Omega}
|\nabla w(x,t)|^2\,dx
=
0.
$$

### Demostración

Multiplicamos la ecuación

$$
w_t-\Delta w=0
$$

por

$$
w
$$

e integramos en

$$
\Omega.
$$

Se obtiene

$$
\int_{\Omega}
w\,w_t\,dx
-
\int_{\Omega}
w\,\Delta w\,dx
=
0.
$$

Por el Lema 04.06.2,

$$
\int_{\Omega}
w\,w_t\,dx
=
\frac{d}{dt}
\left(
\frac12
\int_{\Omega}
w^2\,dx
\right).
$$

Por el Lema 04.06.1 y la condición de frontera

$$
w=0
\qquad
\text{en } \partial\Omega,
$$

tenemos

$$
\int_{\Omega}
w\,\Delta w\,dx
=
-
\int_{\Omega}
|\nabla w|^2\,dx.
$$

Sustituyendo, resulta

$$
\frac{d}{dt}
\left(
\frac12
\int_{\Omega}
w^2\,dx
\right)
+
\int_{\Omega}
|\nabla w|^2\,dx
=
0.
$$


## 04.06.5 — Consecuencia inmediata: la energía decrece

### **Corolario 04.06.4 — Monotonía de la energía**

Bajo las hipótesis de la Proposición 04.06.3, la energía

$$
E_w(t)
=
\frac12
\int_{\Omega}
w(x,t)^2\,dx
$$

es no creciente en

$$
[0,T].
$$

Además,

$$
E_w(t)
+
\int_0^t
\int_{\Omega}
|\nabla w(x,s)|^2\,dx\,ds
=
E_w(0).
$$

### Demostración

Basta integrar la identidad de energía entre

$$
0
\quad\text{y}\quad
t.
$$


## 04.06.6 — Unicidad por método de energía

### **Teorema 04.06.5 — Unicidad del problema de Dirichlet parabólico**

Sea

$$
\Omega\subset\mathbb R^n
$$

abierto y acotado, con frontera regular por pedazos.

Sean

$$
f\in C(\overline{\Omega_T}),
\qquad
g\in C(\Gamma_T).
$$

Si el problema

$$
\begin{cases}
u_t-\Delta u=f
&
\text{en } \Omega_T,
\\[2mm]
u=g
&
\text{en } \Gamma_T
\end{cases}
$$

admite una solución clásica en

$$
C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T}),
$$

entonces tal solución es única.

### Demostración

Supongamos que

$$
u
\qquad\text{y}\qquad
v
$$

son dos soluciones.

Definimos

$$
w=u-v.
$$

Entonces

$$
w_t-\Delta w=0
\qquad
\text{en } \Omega_T,
$$

y, como ambas soluciones comparten los mismos datos sobre

$$
\Gamma_T,
$$

se tiene

$$
w=0
\qquad
\text{en } \Gamma_T.
$$

En particular,

$$
w(x,0)=0
\qquad
\text{en } \Omega,
$$

y

$$
w=0
\qquad
\text{en } \partial\Omega\times(0,T].
$$

Aplicando la Proposición 04.06.3, obtenemos

$$
\frac{d}{dt}
\left(
\frac12
\int_{\Omega}
w(x,t)^2\,dx
\right)
+
\int_{\Omega}
|\nabla w(x,t)|^2\,dx
=
0.
$$

Como

$$
E_w(0)=0,
$$

el Corolario 04.06.4 implica

$$
E_w(t)=0
\qquad
\text{para todo } t\in[0,T].
$$

Por tanto,

$$
w(\cdot,t)\equiv 0
\qquad
\text{en } \Omega
$$

para todo

$$
t,
$$

es decir,

$$
u\equiv v.
$$


## 04.06.7 — Dependencia continua respecto del término fuente

### **Proposición 04.06.6 — Estimación con fuente**

Sea

$$
w\in C^{2,1}(\Omega_T)\cap C(\overline{\Omega_T})
$$

tal que

$$
\begin{cases}
w_t-\Delta w=r
&
\text{en } \Omega_T,
\\[2mm]
w=0
&
\text{en } \partial\Omega\times(0,T],
\\[2mm]
w(\cdot,0)=w_0
&
\text{en } \Omega.
\end{cases}
$$

Entonces, para todo

$$
t\in[0,T],
$$

se cumple

$$
\frac{d}{dt}
\left(
\frac12\|w(\cdot,t)\|_{L^2(\Omega)}^2
\right)
+
\|\nabla w(\cdot,t)\|_{L^2(\Omega)}^2
=
\int_{\Omega}
r(x,t)\,w(x,t)\,dx.
$$

En consecuencia, por Cauchy–Schwarz y Young,

$$
\frac{d}{dt}
\|w(\cdot,t)\|_{L^2(\Omega)}^2
\leq
\|r(\cdot,t)\|_{L^2(\Omega)}^2
+
\|w(\cdot,t)\|_{L^2(\Omega)}^2.
$$

### Demostración

Se multiplica la ecuación por

$$
w
$$

y se repite el mismo cálculo de la Proposición 04.06.3, ahora con el término fuente. Luego se usa

$$
2ab\leq a^2+b^2.
$$


### **Corolario 04.06.7 — Estabilidad en $L^2$**

Bajo las hipótesis de la Proposición 04.06.6,

$$
\|w(\cdot,t)\|_{L^2(\Omega)}^2
\leq
e^t
\left(
\|w_0\|_{L^2(\Omega)}^2
+
\int_0^t
\|r(\cdot,s)\|_{L^2(\Omega)}^2\,ds
\right).
$$

> **Complemento añadido.**
> Esta estimación se deduce por el lema de Grönwall y es la forma básica de dependencia continua de la solución respecto de los datos.


## 04.06.8 — Ejemplo explícito

Sea

$$
\Omega=(0,1)
$$

y sea

$$
u(x,t)=e^{-\pi^2 t}\sin(\pi x).
$$

Entonces

$$
u_t-u_{xx}=0,
\qquad
u(0,t)=u(1,t)=0.
$$

La energía asociada es

$$
E(t)
=
\frac12
\int_0^1
u(x,t)^2\,dx
=
\frac12
e^{-2\pi^2 t}
\int_0^1
\sin^2(\pi x)\,dx
=
\frac14
e^{-2\pi^2 t}.
$$

Por tanto,

$$
E'(t)
=
-
\frac{\pi^2}{2}
e^{-2\pi^2 t}
<
0.
$$

Esto ilustra de manera explícita el decaimiento monótono de la energía.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0.0, 0.6, 600)
E = 0.25 * np.exp(-2.0 * (np.pi**2) * t)

plt.figure(figsize=(10, 5.5), dpi=160)
plt.plot(t, E, linewidth=2)
plt.xlabel("t")
plt.ylabel("E(t)")
plt.title("Decaimiento de la energía para un modo térmico")
plt.tight_layout()
plt.show()
```
