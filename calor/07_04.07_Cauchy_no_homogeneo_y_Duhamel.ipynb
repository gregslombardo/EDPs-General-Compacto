## 04.07.1 — Transcripción literal normalizada de las notas

### **Teorema 04.07.1**

En las notas aparece primero un teorema para el problema de Cauchy no homogéneo con dato inicial nulo.

Sea

$$
h\in C_0^2(\mathbb R^n\times(0,\infty))
$$

dada. Definimos

$$
u(x,t)
=
\int_0^t
\int_{\mathbb R^n}
\frac{1}{\left(4\pi(t-s)\right)^{n/2}}
\exp\!\left(
-\frac{|x-y|^2}{4(t-s)}
\right)
h(y,s)\,dy\,ds.
$$

Entonces:

**a)**

$$
u\in C_1^2(\mathbb R^n\times(0,\infty)).
$$

**b)**

$$
u_t(x,t)-\Delta u(x,t)=h(x,t)
\qquad
\forall\, x\in\mathbb R^n,\ t>0.
$$

**c)**

$$
\lim_{(x,t)\to(x_0,0^+)}u(x,t)=0
\qquad
\forall\,x_0\in\mathbb R^n.
$$

---

### **Corolario 04.07.2**

En las notas aparece después el siguiente corolario.

Sean

$$
h\in C_0^2(\mathbb R^n\times(0,\infty))
$$

y

$$
g\in C(\mathbb R^n),
$$

con

$$
g
$$

acotada. Entonces

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,g(y)\,dy
+
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy\,ds
$$

satisface:

**a)**

$$
u\in C_1^2(\mathbb R^n\times(0,\infty)).
$$

**b)**

$$
u_t(x,t)-\Delta u(x,t)=h(x,t)
\qquad
\forall\,(x,t)\in\mathbb R^n\times(0,\infty).
$$

**c)**

$$
\lim_{(x,t)\to(x_0,0^+)}u(x,t)=g(x_0)
\qquad
\forall\,x_0\in\mathbb R^n.
$$

> **Nota de transcripción.**
> Se ha normalizado la notación usando
> $$
> \psi(x,t)=\frac{1}{(4\pi t)^{n/2}}e^{-\frac{|x|^2}{4t}}
> $$
> para el núcleo del calor, tal como aparece en los primeros notebooks.


## 04.07.2 — Interpretación

El primer término de la fórmula representa la evolución libre del dato inicial

$$
g.
$$

El segundo término representa la superposición temporal de las respuestas elementales creadas por la fuente

$$
h(y,s)
$$

en cada instante

$$
s<t.
$$

Esa superposición es precisamente el **principio de Duhamel**:

cada contribución creada en el tiempo

$$
s
$$

evoluciona desde entonces hasta el tiempo

$$
t
$$

mediante el semigrupo del calor.


## 04.07.3 — Definición del operador de Duhamel

### **Definición 04.07.3 — Potencial de Duhamel**

Para una fuente

$$
h:\mathbb R^n\times(0,\infty)\to\mathbb R,
$$

definimos el **potencial de Duhamel** por

$$
(\mathcal D h)(x,t)
=
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy\,ds.
$$

Cuando la integral está bien definida, el principio de Duhamel afirma que

$$
\mathcal D h
$$

es la solución del problema de Cauchy no homogéneo con dato inicial nulo.


## 04.07.4 — Resultado principal con dato inicial nulo

### **Teorema 04.07.4 — Fórmula de Duhamel con dato inicial nulo**

Sea

$$
h\in C_0^2(\mathbb R^n\times(0,\infty)).
$$

Defínase

$$
u(x,t)
=
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy\,ds.
$$

Entonces:

1. 
   $$
   u\in C_1^2(\mathbb R^n\times(0,\infty));
   $$

2. 
   $$
   u_t-\Delta u=h
   \qquad
   \text{en } \mathbb R^n\times(0,\infty);
   $$

3. 
   $$
   \lim_{(x,t)\to(x_0,0^+)}u(x,t)=0
   \qquad
   \forall\,x_0\in\mathbb R^n.
   $$


### Demostración

Escribimos

$$
u(x,t)
=
\int_0^t
v_s(x,t)\,ds,
$$

donde, para cada

$$
s<t,
$$

definimos

$$
v_s(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy.
$$

Para cada

$$
s
$$

fijo, la función

$$
v_s(\cdot,t)
$$

es una convolución espacial del núcleo del calor con la función

$$
h(\cdot,s).
$$

Por el resultado del Notebook 01, para

$$
t>s,
$$

se tiene

$$
\partial_t v_s - \Delta v_s = 0.
$$

Además, cuando

$$
t\downarrow s,
$$

la propiedad de aproximación de la identidad da

$$
v_s(x,t)\longrightarrow h(x,s).
$$

Derivando respecto de

$$
t
$$

bajo el signo integral y usando la regla de Leibniz, obtenemos

$$
u_t(x,t)
=
v_t(x,t)
+
\int_0^t
\partial_t v_s(x,t)\,ds,
$$

donde el término de borde es precisamente

$$
v_t(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,0^+)\,h(y,t)\,dy
=
h(x,t).
$$

Como

$$
\partial_t v_s=\Delta v_s,
$$

se concluye que

$$
u_t(x,t)
=
h(x,t)
+
\Delta
\int_0^t
v_s(x,t)\,ds
=
h(x,t)+\Delta u(x,t).
$$

Por tanto,

$$
u_t-\Delta u=h.
$$

La regularidad

$$
u\in C_1^2
$$

se deduce de la suavidad de

$$
\psi
$$

para

$$
t>s
$$

y del soporte compacto temporal-espacial de

$$
h.
$$

Finalmente, para el comportamiento cuando

$$
t\downarrow 0,
$$

se estima

$$
|u(x,t)|
\leq
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,|h(y,s)|\,dy\,ds
\leq
t\,\sup |h|.
$$

Por tanto,

$$
u(x,t)\to 0
\qquad
\text{uniformemente en } x
$$

cuando

$$
t\downarrow 0.
$$


## 04.07.5 — Corolario con dato inicial general

### **Corolario 04.07.5 — Fórmula general de Cauchy**

Sean

$$
h\in C_0^2(\mathbb R^n\times(0,\infty))
$$

y

$$
g\in C(\mathbb R^n),
$$

con

$$
g
$$

acotada. Definimos

$$
u(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,g(y)\,dy
+
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy\,ds.
$$

Entonces:

1. 
   $$
   u\in C_1^2(\mathbb R^n\times(0,\infty));
   $$

2. 
   $$
   u_t-\Delta u=h
   \qquad
   \text{en } \mathbb R^n\times(0,\infty);
   $$

3. 
   $$
   \lim_{(x,t)\to(x_0,0^+)}u(x,t)=g(x_0)
   \qquad
   \forall\,x_0\in\mathbb R^n.
   $$


### Demostración

Descomponemos

$$
u=v+w,
$$

donde

$$
v(x,t)
=
\int_{\mathbb R^n}
\psi(x-y,t)\,g(y)\,dy
$$

y

$$
w(x,t)
=
\int_0^t
\int_{\mathbb R^n}
\psi(x-y,t-s)\,h(y,s)\,dy\,ds.
$$

El Notebook 01 muestra que

$$
v_t-\Delta v=0
$$

y que

$$
v(x,t)\to g(x_0)
$$

cuando

$$
(x,t)\to(x_0,0^+).
$$

El Teorema 04.07.4 muestra que

$$
w_t-\Delta w=h
$$

y

$$
w(x,t)\to 0
$$

cuando

$$
(x,t)\to(x_0,0^+).
$$

Sumando ambos resultados se obtiene el corolario.


## 04.07.6 — Comentario añadido para cerrar el temario

En formulación abstracta, el principio de Duhamel puede escribirse como

$$
u(t)
=
e^{t\Delta}g
+
\int_0^t
e^{(t-s)\Delta}h(s)\,ds.
$$

Esta expresión es la versión de semigrupo de la fórmula integral de las notas.

Aunque la notación de semigrupo no aparece explícitamente en la hoja manuscrita, es el lenguaje conceptual estándar que conecta esta sección con análisis funcional y problemas más generales.


## 04.07.7 — Ejemplo explícito en un intervalo

Como complemento añadido, consideremos el problema

$$
\begin{cases}
u_t-u_{xx}=e^{-t}\sin(\pi x),
&
0<x<1,\ t>0,
\\[2mm]
u(0,t)=u(1,t)=0,
&
t>0,
\\[2mm]
u(x,0)=0,
&
0<x<1.
\end{cases}
$$

Buscamos una solución de la forma

$$
u(x,t)=a(t)\sin(\pi x).
$$

Sustituyendo, obtenemos la ecuación para la amplitud

$$
a'(t)+\pi^2 a(t)=e^{-t},
\qquad
a(0)=0.
$$

Por variación de constantes,

$$
a(t)
=
e^{-\pi^2 t}
\int_0^t
e^{(\pi^2-1)s}\,ds
=
\frac{e^{-t}-e^{-\pi^2 t}}{\pi^2-1}.
$$

Por tanto,

$$
u(x,t)
=
\frac{e^{-t}-e^{-\pi^2 t}}{\pi^2-1}\sin(\pi x).
$$

Este ejemplo muestra con claridad la idea de Duhamel: la fuente excita un modo espacial y la amplitud temporal resulta de una convolución en el tiempo.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1200)

def a(t):
    return (np.exp(-t) - np.exp(-(np.pi**2) * t)) / (np.pi**2 - 1.0)

def u(x, t):
    return a(t) * np.sin(np.pi * x)

plt.figure(figsize=(10, 6), dpi=160)
for t in [0.0, 0.10, 0.25, 0.50, 0.90]:
    plt.plot(x, u(x, t), linewidth=2, label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Respuesta forzada tipo Duhamel para un modo senoidal")
plt.legend()
plt.tight_layout()
plt.show()
```

## 04.07.8 — Escena Manim sin compilar

La escena `DuhamelScene` visualiza un caso con dato inicial nulo y fuente activada después de un tiempo pequeño.

La intención es mostrar:

1. el perfil inicial nulo, visible durante más de cuatro segundos;
2. la aparición gradual de la respuesta forzada;
3. la interpretación de la solución como suma temporal de contribuciones.

La configuración es

$$
2560\times1440
$$

y

$$
120\ \mathrm{fps}.
$$


```python
from pathlib import Path
import subprocess
import sys

HERE = Path.cwd()

if HERE.name != "07_04.07_Cauchy_no_homogeneo_y_Duhamel":
    candidate = (
        HERE
        / "04_Ecuacion_del_calor_REVISADO_01_a_07"
        / "07_04.07_Cauchy_no_homogeneo_y_Duhamel"
    )
    if candidate.exists():
        HERE = candidate

SCRIPT = HERE / "manim" / "07_04_07_cauchy_no_homogeneo_y_duhamel.py"
SCENE = "DuhamelScene"

base = [
    sys.executable,
    "-m",
    "manim",
    "--format=mp4",
]

try:
    subprocess.run(
        base + ["--renderer=opengl", str(SCRIPT), SCENE],
        check=True,
        cwd=HERE,
    )
except subprocess.CalledProcessError:
    print("OpenGL no completó el render; se intenta Cairo.")
    subprocess.run(
        base + ["--renderer=cairo", str(SCRIPT), SCENE],
        check=True,
        cwd=HERE,
    )
```
