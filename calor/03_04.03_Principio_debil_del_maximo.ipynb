## 04.03.1 — Transcripción literal normalizada

### **Principio Débil del máximo**

Sea

$$
\Omega\subset\mathbb R^n
$$

abierto y acotado, con frontera regular por pedazos.

Se define

$$
\Omega_T=\Omega\times(0,T],
\qquad
T>0,
$$

y la frontera parabólica

$$
\Gamma_T
=
\left(\partial\Omega\times(0,T]\right)
\cup
\left(\overline\Omega\times\{0\}\right).
$$

En las notas se enuncia:

si

$$
u\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T)
$$

y

$$
u_t\leq\Delta u
$$

en

$$
\Omega_T,
$$

entonces

$$
\max_{\overline{\Omega_T}}u
=
\max_{\Gamma_T}u.
$$


## 04.03.2 — Definiciones

### **Definición 04.03.1 — Cilindro parabólico**

El conjunto

$$
\Omega_T=\Omega\times(0,T]
$$

se denomina **cilindro parabólico**.

---

### **Definición 04.03.2 — Frontera parabólica**

El conjunto

$$
\Gamma_T
=
\left(\partial\Omega\times(0,T]\right)
\cup
\left(\overline\Omega\times\{0\}\right)
$$

se denomina **frontera parabólica**.

---

### **Definición 04.03.3 — Clase parabólica**

La notación

$$
u\in C^{2,1}(\Omega_T)
$$

significa que existen y son continuas en

$$
\Omega_T
$$

las derivadas

$$
u_t,
\qquad
u_{x_i},
\qquad
u_{x_i x_j}.
$$


## 04.03.3 — Teorema principal

### **Teorema 04.03.4 — Principio débil del máximo**

Sea

$$
\Omega\subset\mathbb R^n
$$

abierto y acotado, con frontera regular por pedazos.

Sea

$$
T>0.
$$

Supongamos que

$$
u\in C(\overline{\Omega_T})\cap C^{2,1}(\Omega_T)
$$

y que

$$
u_t-\Delta u\leq0
$$

en

$$
\Omega_T.
$$

Entonces

$$
\boxed{
\max_{\overline{\Omega_T}}u
=
\max_{\Gamma_T}u
}.
$$


## 04.03.4 — Demostración

Para

$$
\varepsilon>0,
$$

definimos

$$
v(x,t)=u(x,t)-\varepsilon t.
$$

Entonces

$$
v_t-\Delta v
=
u_t-\Delta u-\varepsilon
<
0.
$$

Supongamos, por contradicción, que

$$
v
$$

alcanza un máximo absoluto en un punto interior

$$
(x_0,t_0)\in\Omega_T.
$$

En dicho punto se cumple

$$
\nabla v(x_0,t_0)=0
$$

y

$$
D^2v(x_0,t_0)\leq0.
$$

Por consiguiente,

$$
\Delta v(x_0,t_0)\leq0.
$$

Respecto de la variable temporal:

- si
  $$
  0<t_0<T,
  $$
  entonces
  $$
  v_t(x_0,t_0)=0;
  $$

- si
  $$
  t_0=T,
  $$
  entonces la derivada lateral satisface
  $$
  v_t(x_0,T)\geq0.
  $$

En ambos casos,

$$
v_t(x_0,t_0)-\Delta v(x_0,t_0)\geq0.
$$

Esto contradice

$$
v_t-\Delta v<0.
$$

Por tanto, el máximo de

$$
v
$$

se alcanza en

$$
\Gamma_T.
$$

Al hacer

$$
\varepsilon\downarrow0,
$$

obtenemos

$$
\max_{\overline{\Omega_T}}u
=
\max_{\Gamma_T}u.
$$


## 04.03.5 — Ejemplo explícito

Sea

$$
\Omega=(0,1)
$$

y

$$
u(x,t)=e^{-\pi^2 t}\sin(\pi x).
$$

Entonces

$$
u_t-u_{xx}=0,
$$

$$
u(0,t)=u(1,t)=0,
$$

y

$$
u(x,0)=\sin(\pi x).
$$

Además,

$$
\max_{x\in[0,1]}u(x,t)
=
e^{-\pi^2 t}.
$$

Por tanto,

$$
\max_{\overline{\Omega_T}}u=1
$$

y dicho máximo se alcanza en

$$
\overline\Omega\times\{0\}
\subset\Gamma_T.
$$


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1000)

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.05, 0.12, 0.25, 0.45]:
    u = np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)
    plt.plot(x, u, linewidth=2, label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Disminución del máximo de un modo térmico")
plt.legend()
plt.tight_layout()
plt.show()
```
