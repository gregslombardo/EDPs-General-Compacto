## 04.12.1 — Problema homogéneo de Dirichlet

Sean

$$
L>0,
\qquad
k>0.
$$

Consideramos

$$
\begin{cases}
u_t=k u_{xx},
&
0<x<L,\quad t>0,
\\[2mm]
u(0,t)=u(L,t)=0,
&
t>0,
\\[2mm]
u(x,0)=f(x),
&
0<x<L.
\end{cases}
$$

Buscamos soluciones separadas de la forma

$$
u(x,t)=X(x)T(t).
$$


## 04.12.2 — Separación de variables

Sustituyendo en la ecuación,

$$
X(x)T'(t)=kX''(x)T(t).
$$

Después de dividir,

$$
\frac{T'(t)}{kT(t)}
=
\frac{X''(x)}{X(x)}
=
-\lambda.
$$

Aparecen

$$
X''+\lambda X=0
$$

y

$$
T'+k\lambda T=0.
$$


## 04.12.3 — Espectro de Dirichlet

### **Proposición 04.12.1 — Valores propios de Dirichlet**

El problema

$$
\begin{cases}
X''+\lambda X=0,
\\[2mm]
X(0)=X(L)=0
\end{cases}
$$

tiene soluciones no triviales únicamente para

$$
\boxed{
\lambda_n
=
\left(
\frac{n\pi}{L}
\right)^2,
\qquad
n=1,2,\ldots
}
$$

con funciones propias

$$
\boxed{
X_n(x)
=
\sin\!\left(
\frac{n\pi x}{L}
\right)
}.
$$

### Demostración

Para

$$
\lambda=0
$$

las soluciones son lineales y las condiciones de frontera sólo permiten la solución trivial.

Para

$$
\lambda<0
$$

las soluciones son combinaciones exponenciales y nuevamente las condiciones homogéneas fuerzan trivialidad.

Para

$$
\lambda=\mu^2>0,
$$

$$
X(x)=A\cos(\mu x)+B\sin(\mu x).
$$

La condición

$$
X(0)=0
$$

implica

$$
A=0,
$$

y

$$
X(L)=0
$$

exige

$$
\sin(\mu L)=0.
$$

Por tanto,

$$
\mu L=n\pi.
$$


## 04.12.4 — Evolución temporal de cada modo

Para

$$
\lambda_n
=
\left(
\frac{n\pi}{L}
\right)^2,
$$

se tiene

$$
T_n'(t)
+
k\left(
\frac{n\pi}{L}
\right)^2T_n(t)
=
0.
$$

Por tanto,

$$
\boxed{
T_n(t)
=
C_n
\exp\!\left[
-k
\left(
\frac{n\pi}{L}
\right)^2t
\right]
}.
$$

Cada modo separado es

$$
u_n(x,t)
=
C_n
e^{-k(n\pi/L)^2t}
\sin\!\left(
\frac{n\pi x}{L}
\right).
$$


## 04.12.5 — Serie seno de Fourier

### **Definición 04.12.2 — Coeficientes seno**

Para

$$
f\in L^2(0,L),
$$

definimos

$$
\boxed{
b_n
=
\frac{2}{L}
\int_0^L
f(x)
\sin\!\left(
\frac{n\pi x}{L}
\right)\,dx
}.
$$

La serie seno asociada es

$$
f(x)
\sim
\sum_{n=1}^{\infty}
b_n
\sin\!\left(
\frac{n\pi x}{L}
\right).
$$


### **Teorema 04.12.3 — Solución de Dirichlet por series**

Sea

$$
f\in L^2(0,L).
$$

Definimos

$$
\boxed{
u(x,t)
=
\sum_{n=1}^{\infty}
b_n
e^{-k(n\pi/L)^2t}
\sin\!\left(
\frac{n\pi x}{L}
\right)
}.
$$

Entonces, para cada

$$
t>0,
$$

la serie define una función suave que satisface

$$
u_t=k u_{xx},
$$

$$
u(0,t)=u(L,t)=0,
$$

y

$$
u(\cdot,t)\longrightarrow f
$$

en

$$
L^2(0,L)
$$

cuando

$$
t\downarrow0.
$$

### Demostración

Por Parseval,

$$
\sum_{n=1}^{\infty}|b_n|^2<\infty.
$$

Para

$$
t\geq\tau>0,
$$

el factor

$$
e^{-k(n\pi/L)^2t}
$$

domina cualquier potencia de

$$
n.
$$

La serie y todas sus derivadas convergen uniformemente, por lo que puede derivarse término a término.

La convergencia inicial en

$$
L^2
$$

se obtiene de Parseval y de

$$
e^{-k(n\pi/L)^2t}\to1.
$$


## 04.12.6 — Recuperación del dato inicial

### **Corolario 04.12.4**

Si

$$
f
$$

es continua por pedazos, entonces

$$
u(x,t)
\longrightarrow
\frac{f(x^-)+f(x^+)}{2}
$$

en cada punto interior.

En un punto de continuidad,

$$
u(x,t)\longrightarrow f(x).
$$

### **Observación 04.12.5**

Si

$$
f(0)\neq0
$$

o

$$
f(L)\neq0,
$$

no puede haber convergencia uniforme en todo el intervalo porque

$$
u(0,t)=u(L,t)=0
$$

para todo

$$
t>0.
$$


## 04.12.7 — Decaimiento modal

### **Proposición 04.12.6 — Amortiguamiento de frecuencias altas**

El modo

$$
n
$$

decae mediante

$$
e^{-k(n\pi/L)^2t}.
$$

Su escala temporal es

$$
\boxed{
\tau_n
=
\frac{L^2}{k n^2\pi^2}
}.
$$

Los modos de alta frecuencia desaparecen antes.

### **Corolario 04.12.7 — Dominio del primer modo**

Si

$$
b_1\neq0,
$$

entonces

$$
u(x,t)
\sim
b_1
e^{-k(\pi/L)^2t}
\sin\!\left(
\frac{\pi x}{L}
\right)
$$

cuando

$$
t\to\infty.
$$


## 04.12.8 — Energía y Parseval

### **Proposición 04.12.8 — Identidad modal de energía**

Se cumple

$$
\|u(\cdot,t)\|_{L^2(0,L)}^2
=
\frac{L}{2}
\sum_{n=1}^{\infty}
|b_n|^2
e^{-2k(n\pi/L)^2t}.
$$

Por tanto,

$$
\boxed{
\|u(\cdot,t)\|_{L^2}
\leq
e^{-k\pi^2t/L^2}
\|f\|_{L^2}
}.
$$


## 04.12.9 — Problema homogéneo de Neumann

Consideramos

$$
\begin{cases}
u_t=k u_{xx},
&
0<x<L,\quad t>0,
\\[2mm]
u_x(0,t)=u_x(L,t)=0,
\\[2mm]
u(x,0)=f(x).
\end{cases}
$$

### **Proposición 04.12.9 — Espectro de Neumann**

Los valores propios son

$$
\lambda_0=0
$$

y

$$
\lambda_n
=
\left(
\frac{n\pi}{L}
\right)^2,
\qquad
n\geq1.
$$

Las funciones propias son

$$
X_0(x)=1
$$

y

$$
X_n(x)
=
\cos\!\left(
\frac{n\pi x}{L}
\right).
$$


## 04.12.10 — Serie coseno y conservación de la media

Definimos

$$
a_0
=
\frac{1}{L}
\int_0^L f(x)\,dx
$$

y

$$
a_n
=
\frac{2}{L}
\int_0^L
f(x)
\cos\!\left(
\frac{n\pi x}{L}
\right)\,dx.
$$

La solución es

$$
\boxed{
u(x,t)
=
a_0
+
\sum_{n=1}^{\infty}
a_n
e^{-k(n\pi/L)^2t}
\cos\!\left(
\frac{n\pi x}{L}
\right)
}.
$$

El modo constante no decae. En consecuencia,

$$
\frac{1}{L}
\int_0^L u(x,t)\,dx
=
a_0.
$$

### **Corolario 04.12.10 — Equilibrio de Neumann**

Cuando

$$
t\to\infty,
$$

$$
u(x,t)
\longrightarrow
\frac{1}{L}
\int_0^L f(y)\,dy.
$$


## 04.12.11 — Fuente después de homogeneizar

Para

$$
v_t-kv_{xx}=F(x,t)
$$

con frontera homogénea, escribimos

$$
v(x,t)
=
\sum_{n=1}^{\infty}
c_n(t)
\sin\!\left(
\frac{n\pi x}{L}
\right).
$$

Cada coeficiente satisface

$$
c_n'(t)
+
k\left(
\frac{n\pi}{L}
\right)^2c_n(t)
=
F_n(t).
$$

Por variación de constantes,

$$
\boxed{
c_n(t)
=
b_n e^{-k(n\pi/L)^2t}
+
\int_0^t
e^{-k(n\pi/L)^2(t-s)}
F_n(s)\,ds
}.
$$

> **Complemento añadido.**
> Esta fórmula combina separación de variables y Duhamel.


## 04.12.12 — Ejemplo de cuatro modos

Para

$$
L=1,
\qquad
k=1,
$$

sea

$$
f(x)
=
\sin(\pi x)
+
0.45\sin(2\pi x)
-
0.30\sin(3\pi x)
+
0.20\sin(5\pi x).
$$

Entonces

$$
\begin{aligned}
u(x,t)
={}&
e^{-\pi^2t}\sin(\pi x)
+
0.45e^{-4\pi^2t}\sin(2\pi x)
\\
&
-
0.30e^{-9\pi^2t}\sin(3\pi x)
+
0.20e^{-25\pi^2t}\sin(5\pi x).
\end{aligned}
$$


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.0, 1.0, 1800)

coefficients = {
    1: 1.00,
    2: 0.45,
    3: -0.30,
    5: 0.20,
}

def solution(x, t):
    value = np.zeros_like(x)
    for mode, coefficient in coefficients.items():
        value += (
            coefficient
            * np.exp(-((mode * np.pi) ** 2) * t)
            * np.sin(mode * np.pi * x)
        )
    return value

plt.figure(figsize=(10, 6), dpi=160)

for t in [0.0, 0.005, 0.02, 0.08, 0.25]:
    plt.plot(x, solution(x, t), linewidth=2, label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Decaimiento modal de una serie seno finita")
plt.legend()
plt.tight_layout()
plt.show()
```
