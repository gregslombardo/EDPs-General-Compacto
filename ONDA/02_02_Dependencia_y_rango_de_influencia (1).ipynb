# 2.3.1. Caracterización de las soluciones

## **Teorema 2.3.1 (Caracterización de las soluciones unidimensionales; teorema de las notas).**

Sea

$$
u\in C^2(\mathbb R\times(0,\infty)).
$$

Entonces

$$
u_{tt}-c^2u_{xx}=0
$$

si y sólo si existen funciones $V,W\in C^2(\mathbb R)$ tales que

$$
\boxed{
u(x,t)=V(x+ct)+W(x-ct).
}
$$

### Demostración

Introducimos

$$
\xi=x+ct,
\qquad
\eta=x-ct,
$$

y escribimos

$$
u(x,t)=U(\xi,\eta).
$$

Entonces

$$
u_{tt}-c^2u_{xx}
=
-4c^2U_{\xi\eta}.
$$

Por tanto,

$$
u_{tt}-c^2u_{xx}=0
\quad\Longleftrightarrow\quad
U_{\xi\eta}=0.
$$

Integrando sucesivamente,

$$
U(\xi,\eta)=V(\xi)+W(\eta).
$$

Al regresar a las variables originales,

$$
u(x,t)=V(x+ct)+W(x-ct).
$$

La recíproca se obtiene derivando directamente.

$\square$

## **Observación 2.3.2 (No unicidad de la descomposición).**

Si $K\in\mathbb R$, entonces

$$
V(x+ct)+W(x-ct)
=
\bigl(V(x+ct)+K\bigr)
+
\bigl(W(x-ct)-K\bigr).
$$

La descomposición queda determinada salvo el traslado de una constante.

# 2.3.2. Paralelogramo característico

## **Definición 2.3.3 (Paralelogramo característico).**

Un cuadrilátero del plano $(x,t)$ se denomina **paralelogramo característico** si sus lados están contenidos alternadamente en rectas de las familias

$$
x+ct=\text{constante},
$$

y

$$
x-ct=\text{constante}.
$$

En las coordenadas $(\xi,\eta)$, este paralelogramo se transforma en un rectángulo.

## **Teorema 2.3.4 (Teorema del paralelogramo característico; d’Alembert).**

Sea

$$
u\in C^2(\mathbb R\times(0,\infty)).
$$

Entonces $u$ satisface

$$
u_{tt}-c^2u_{xx}=0
$$

si y sólo si para cada paralelogramo característico con vértices $A,B,C,D$, dispuestos como en las notas, se cumple

$$
\boxed{
u(A)+u(C)=u(B)+u(D).
}
$$

### Demostración

Si

$$
u(x,t)=V(\xi)+W(\eta),
$$

y los vértices corresponden a

$$
A=(\xi_2,\eta_2),
\qquad
B=(\xi_2,\eta_1),
$$

$$
C=(\xi_1,\eta_1),
\qquad
D=(\xi_1,\eta_2),
$$

entonces ambos lados de la identidad son iguales a

$$
V(\xi_1)+V(\xi_2)+W(\eta_1)+W(\eta_2).
$$

Recíprocamente, la identidad implica

$$
U(\xi+h,\eta+k)-U(\xi+h,\eta)-U(\xi,\eta+k)+U(\xi,\eta)=0.
$$

Dividiendo entre $hk$ y haciendo $h,k\to0$,

$$
U_{\xi\eta}=0.
$$

En consecuencia,

$$
u_{tt}-c^2u_{xx}=0.
$$

$\square$

## **Corolario 2.3.5 (Diferencia mixta discreta).**

El teorema del paralelogramo es la forma finita de

$$
U_{\xi\eta}=0.
$$

En notación de diferencias,

$$
\boxed{
\Delta_\xi\Delta_\eta U=0.
}
$$

# 2.3.3. Visualización con Manim

La escena:

1. muestra las dos familias características;
2. deja visible el paralelogramo durante varios segundos;
3. escribe los valores de $u$ en los cuatro vértices;
4. deriva visualmente la identidad;
5. desplaza lentamente el paralelogramo;
6. conecta la identidad discreta con $U_{\xi\eta}=0$.

El video no se compila en esta entrega. El script queda preparado en:

`manim/paralelogramo_caracteristico.py`

Para renderizar en alta calidad desde la carpeta de la sección:

```bash
manim -qh --fps 60 manim/paralelogramo_caracteristico.py ParalelogramoCaracteristico
```

No es necesario modificar ninguna bandera dentro del código.

````python
from __future__ import annotations

import os
import tempfile
from pathlib import Path

import numpy as np
from manim import *

# Render de alta calidad. El video NO se compila al crear el notebook.
config.pixel_width = 2560
config.pixel_height = 1440
config.frame_rate = 60
config.background_color = "#0f1115"
config.media_dir = str(Path(__file__).resolve().parent.parent / "animaciones")

# Robustez para Windows/MiKTeX.
try:
    config.no_latex_cleanup = True
except Exception:
    pass

try:
    tex_dir = Path(tempfile.gettempdir()) / f"manim_tex_wave_{os.getpid()}"
    tex_dir.mkdir(parents=True, exist_ok=True)
    config.tex_dir = str(tex_dir)
except Exception:
    pass


class ParalelogramoCaracteristico(Scene):
    def construct(self):
        title = Text(
            "Teorema del paralelogramo característico",
            font_size=46,
        ).to_edge(UP)

        equation = MathTex(
            r"u_{tt}-c^2u_{xx}=0"
        ).next_to(title, DOWN, buff=0.35)

        self.play(Write(title), Write(equation), run_time=2.2)
        self.wait(1.5)

        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[0, 4.5, 1],
            x_length=10.5,
            y_length=5.7,
            axis_config={"include_tip": True},
        ).shift(DOWN * 0.75)

        labels = axes.get_axis_labels(
            MathTex("x"),
            MathTex("t"),
        )

        self.play(Create(axes), Write(labels), run_time=2.5)
        self.wait(1.0)

        c = 1.0
        xi1, xi2 = -2.9, 1.3
        eta1, eta2 = -3.8, -0.9

        def xt(xi, eta):
            x = 0.5 * (xi + eta)
            t = (xi - eta) / (2 * c)
            return axes.c2p(x, t)

        A = xt(xi2, eta2)
        B = xt(xi2, eta1)
        C = xt(xi1, eta1)
        D = xt(xi1, eta2)

        polygon = Polygon(
            A, B, C, D,
            stroke_width=5,
            fill_opacity=0.10,
        )

        dots = VGroup(*[Dot(p, radius=0.075) for p in (A, B, C, D)])
        names = VGroup(
            MathTex("A").next_to(A, UP),
            MathTex("B").next_to(B, RIGHT),
            MathTex("C").next_to(C, DOWN),
            MathTex("D").next_to(D, LEFT),
        )

        family_one = VGroup(
            DashedLine(axes.c2p(-5, 4.0), axes.c2p(3, 0.0)),
            DashedLine(axes.c2p(-3, 4.0), axes.c2p(5, 0.0)),
        )
        family_two = VGroup(
            DashedLine(axes.c2p(-5, 0.0), axes.c2p(3, 4.0)),
            DashedLine(axes.c2p(-3, 0.0), axes.c2p(5, 4.0)),
        )

        self.play(
            Create(family_one),
            Create(family_two),
            run_time=2.5,
        )
        self.play(Create(polygon), FadeIn(dots), Write(names), run_time=2.3)

        # Pausa suficientemente larga para observar la geometría.
        self.wait(3.0)

        char_coords = MathTex(
            r"\xi=x+ct,\qquad \eta=x-ct"
        ).scale(0.82).to_corner(UL).shift(DOWN * 1.25)

        solution_form = MathTex(
            r"u(x,t)=V(\xi)+W(\eta)"
        ).scale(0.82).next_to(char_coords, DOWN, aligned_edge=LEFT)

        self.play(Write(char_coords), Write(solution_form), run_time=2.5)
        self.wait(2.0)

        values = VGroup(
            MathTex(r"u(A)=V(\xi_2)+W(\eta_2)"),
            MathTex(r"u(B)=V(\xi_2)+W(\eta_1)"),
            MathTex(r"u(C)=V(\xi_1)+W(\eta_1)"),
            MathTex(r"u(D)=V(\xi_1)+W(\eta_2)"),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.24).scale(0.68)
        values.to_edge(RIGHT).shift(DOWN * 0.35)

        self.play(LaggedStart(*[Write(v) for v in values], lag_ratio=0.25), run_time=4.5)
        self.wait(2.0)

        identity = MathTex(
            r"\boxed{u(A)+u(C)=u(B)+u(D)}"
        ).scale(1.0).to_edge(DOWN)

        self.play(Write(identity), run_time=2.4)
        self.wait(4.0)

        # Movimiento lento del paralelogramo para mostrar que la identidad
        # no depende de una posición particular.
        shifted = polygon.copy().shift(RIGHT * 1.4 + UP * 0.25)
        shifted_dots = dots.copy().shift(RIGHT * 1.4 + UP * 0.25)
        shifted_names = names.copy().shift(RIGHT * 1.4 + UP * 0.25)

        self.play(
            Transform(polygon, shifted),
            Transform(dots, shifted_dots),
            Transform(names, shifted_names),
            run_time=4.0,
        )
        self.wait(3.0)

        discrete = MathTex(
            r"\Delta_{\xi}\Delta_{\eta}U=0"
        ).scale(0.95).next_to(identity, UP, buff=0.35)

        differential = MathTex(
            r"U_{\xi\eta}=0"
        ).scale(0.95).next_to(discrete, UP, buff=0.3)

        self.play(Write(discrete), run_time=1.8)
        self.play(TransformMatchingTex(discrete.copy(), differential), run_time=2.0)
        self.wait(4.0)
````
