# 2.2.1. Región de dependencia

Después de la fórmula de d’Alembert, las notas fijan un punto $(x,t)$ y escriben

$$
\boxed{
C_{(x,t)}
:=
[x-ct,x+ct]
\subset\mathbb R.
}
\tag{2.6}
$$

Este intervalo se denomina en el manuscrito **región de dependencia de un punto en el espacio-tiempo**.

## **Definición 2.2.1 (Región de dependencia inicial).**

Sea

$$
(x,t)\in\mathbb R\times(0,\infty).
$$

La **región de dependencia inicial** de $(x,t)$ es

$$
\boxed{
C_{(x,t)}
=
[x-ct,x+ct].
}
$$

### Interpretación

La fórmula de d’Alembert muestra que $u(x,t)$ depende únicamente de

$$
f(x-ct),
\qquad
f(x+ct),
$$

y de los valores de $g$ dentro de

$$
[x-ct,x+ct].
$$

Los datos situados fuera de $C_{(x,t)}$ no pueden modificar $u(x,t)$.

## **Proposición 2.2.2 (Dependencia local de los datos).**

Sean

$$
f_1,f_2\in C^2(\mathbb R),
\qquad
g_1,g_2\in C^1(\mathbb R),
$$

y sean $u_1,u_2$ las soluciones correspondientes.

Fijemos

$$
(x_0,t_0)\in\mathbb R\times(0,\infty).
$$

Si

$$
f_1=f_2
\quad\text{en}\quad
[x_0-ct_0,x_0+ct_0]
$$

y

$$
g_1=g_2
\quad\text{en}\quad
[x_0-ct_0,x_0+ct_0],
$$

entonces

$$
u_1(x_0,t_0)=u_2(x_0,t_0).
$$

### Demostración

Por d’Alembert,

$$
\begin{aligned}
u_j(x_0,t_0)
&=
\frac12
\left[
f_j(x_0-ct_0)+f_j(x_0+ct_0)
\right]\\
&\quad+
\frac{1}{2c}
\int_{x_0-ct_0}^{x_0+ct_0}
g_j(y)\,dy,
\qquad j=1,2.
\end{aligned}
$$

Las hipótesis hacen iguales todos los términos.

$\square$

# 2.2.2. Rango de influencia

Las notas fijan ahora

$$
\xi\in\mathbb R
$$

y escriben su rango de influencia como

$$
\boxed{
P_\xi
:=
\left\{
(x,t)\in\mathbb R\times(0,\infty):
\xi-ct\leq x\leq\xi+ct
\right\}.
}
\tag{2.7}
$$

La condición anterior equivale a

$$
|x-\xi|\leq ct.
$$

## **Definición 2.2.3 (Rango de influencia de un punto inicial).**

Sea $\xi\in\mathbb R$. El **rango de influencia** de $\xi$ es

$$
\boxed{
P_\xi
=
\left\{
(x,t)\in\mathbb R\times(0,\infty):
|x-\xi|\leq ct
\right\}.
}
$$

El conjunto $P_\xi$ es el cono futuro con vértice en $(\xi,0)$ y fronteras características

$$
x=\xi-ct,
\qquad
x=\xi+ct.
$$

# 2.2.3. Tiempo mínimo de llegada

Las notas afirman que una perturbación localizada en $\xi$, con condición inicial de soporte cercano a $\xi$, afectará el valor de la solución en $x\in\mathbb R$ sólo a partir del tiempo

$$
\boxed{
t=\frac{|x-\xi|}{c}.
}
\tag{2.8}
$$

## **Proposición 2.2.4 (Tiempo mínimo de influencia).**

Sean $\xi,x\in\mathbb R$.

El punto $(x,t)$ pertenece a $P_\xi$ si y sólo si

$$
t\geq\frac{|x-\xi|}{c}.
$$

En particular, una perturbación localizada en $\xi$ no puede afectar el punto $x$ antes del tiempo

$$
t_{\mathrm{llegada}}
=
\frac{|x-\xi|}{c}.
$$

### Demostración

Por definición,

$$
(x,t)\in P_\xi
\quad\Longleftrightarrow\quad
|x-\xi|\leq ct.
$$

Como $c>0$,

$$
|x-\xi|\leq ct
\quad\Longleftrightarrow\quad
t\geq\frac{|x-\xi|}{c}.
$$

$\square$

## **Resultado añadido 2.2.5 (Velocidad finita de propagación del soporte).**

Sean

$$
f\in C_c^2(\mathbb R),
\qquad
g\in C_c^1(\mathbb R),
$$

y supongamos

$$
\operatorname{supp}f
\cup
\operatorname{supp}g
\subseteq[a,b].
$$

Entonces, para todo $t\geq0$,

$$
\boxed{
\operatorname{supp}u(\cdot,t)
\subseteq
[a-ct,b+ct].
}
$$

### Comentario

Este resultado no aparece como teorema independiente en la página, pero es la formulación rigurosa de los dibujos y del texto sobre el rango de influencia.

### Demostración

Si

$$
x>b+ct,
$$

entonces

$$
x-ct>b
\quad\text{y}\quad
x+ct>b.
$$

Por tanto,

$$
f(x-ct)=f(x+ct)=0,
$$

y todo $[x-ct,x+ct]$ queda fuera del soporte de $g$. La fórmula de d’Alembert da

$$
u(x,t)=0.
$$

El caso $x<a-ct$ es análogo.

$\square$

# 2.2.4. Simulación — cono causal y primera llegada

Se comparan dos problemas cuyos datos iniciales sólo difieren cerca de

$$
\xi=-2.
$$

La diferencia entre sus soluciones permanece fuera del cono

$$
|x-\xi|\leq ct.
$$

Fijaremos un observador en $x_{\mathrm{obs}}=2.5$. La animación marca el instante teórico

$$
t_{\mathrm{llegada}}
=
\frac{|x_{\mathrm{obs}}-\xi|}{c}.
$$

La construcción del mapa espacio-tiempo detecta automáticamente CuPy/CUDA. Si no hay GPU, usa NumPy sin que sea necesario cambiar ninguna opción.

```python
from __future__ import annotations

import shutil
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import Video, display

FIG_DIR = Path("figuras")
ANIM_DIR = Path("animaciones")
FIG_DIR.mkdir(exist_ok=True)
ANIM_DIR.mkdir(exist_ok=True)

GPU_AVAILABLE = False
xp = np

try:
    import cupy as cp
    if cp.cuda.runtime.getDeviceCount() > 0:
        xp = cp
        GPU_AVAILABLE = True
except Exception:
    xp = np
    GPU_AVAILABLE = False

print("Backend numérico:", "CuPy/CUDA" if GPU_AVAILABLE else "NumPy/CPU")

c = 1.0
xi = -2.0
radius = 0.28
x_obs = 2.5
arrival_time = abs(x_obs-xi)/c

x = np.linspace(-7.0, 7.0, 2200)
times = np.linspace(0.0, 6.0, 520)

def bump_numpy(z, center=xi, radius=radius):
    z = np.asarray(z, dtype=float)
    q = (z-center)/radius
    out = np.zeros_like(q)
    mask = np.abs(q) < 1.0
    out[mask] = np.exp(-1.0/(1.0-q[mask]**2))
    if np.any(mask):
        out /= np.exp(-1.0)
    return out

def delta_u(x_values, t):
    return 0.5*(
        bump_numpy(x_values-c*t)
        +
        bump_numpy(x_values+c*t)
    )

print(f"Tiempo teórico de llegada: {arrival_time:.3f}")
```

```python
x0 = 1.1
t0 = 2.2

plt.figure(figsize=(10.5, 7.2))
plt.plot([x0-c*t0, x0], [0.0, t0], linewidth=2.0)
plt.plot([x0+c*t0, x0], [0.0, t0], linewidth=2.0)
plt.plot(
    [x0-c*t0, x0+c*t0],
    [0.0, 0.0],
    linewidth=4.0,
    label=r"$C_{(x_0,t_0)}$",
)
plt.scatter([x0], [t0], s=70)
plt.text(x0+0.08, t0, r"$(x_0,t_0)$")
plt.xlabel("x")
plt.ylabel("t")
plt.title("Región de dependencia de un punto")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

path = FIG_DIR/"02_02_region_de_dependencia.png"
plt.savefig(path, dpi=200, bbox_inches="tight")
plt.show()

print(f"Figura guardada en: {path}")
```

```python
x_backend = xp.asarray(x)
t_backend = xp.asarray(times)
X, T = xp.meshgrid(x_backend, t_backend)

def bump_backend(z, center=xi, radius=radius):
    q = (z-center)/radius
    mask = xp.abs(q) < 1.0
    safe = xp.where(mask, 1.0-q*q, 1.0)
    out = xp.where(mask, xp.exp(-1.0/safe), 0.0)
    return out/xp.exp(-1.0)

U_backend = 0.5*(
    bump_backend(X-c*T)
    +
    bump_backend(X+c*T)
)

U = cp.asnumpy(U_backend) if GPU_AVAILABLE else np.asarray(U_backend)

plt.figure(figsize=(12.8, 7.2))
plt.imshow(
    U,
    origin="lower",
    aspect="auto",
    extent=[x[0], x[-1], times[0], times[-1]],
    interpolation="bilinear",
)
plt.colorbar(label=r"$\Delta u(x,t)$")
plt.plot(xi-c*times, times, "--", linewidth=1.7)
plt.plot(xi+c*times, times, "--", linewidth=1.7)
plt.axvline(x_obs, linestyle=":", linewidth=1.6, label="observador")
plt.scatter([x_obs], [arrival_time], s=65)
plt.xlabel("x")
plt.ylabel("t")
plt.title("Rango de influencia de una perturbación localizada")
plt.legend()
plt.tight_layout()

path = FIG_DIR/"02_02_rango_de_influencia.png"
plt.savefig(path, dpi=200, bbox_inches="tight")
plt.show()

print(f"Figura guardada en: {path}")
```

```python
video_path = ANIM_DIR/"02_02_cono_causal_2K_120fps.mp4"

if not video_path.exists():
    fps = 120
    frames = 180
    animation_times = np.linspace(0.0, 6.0, frames)

    fig, ax = plt.subplots(figsize=(12.8, 7.2), dpi=200)

    line, = ax.plot([], [], linewidth=2.8, label=r"$\Delta u(x,t)$")
    left_boundary, = ax.plot([], [], "--", linewidth=1.2)
    right_boundary, = ax.plot([], [], "--", linewidth=1.2)
    ax.axvline(
        x_obs,
        linestyle=":",
        linewidth=1.8,
        label="observador",
    )
    time_text = ax.text(0.02, 0.93, "", transform=ax.transAxes)
    state_text = ax.text(0.55, 0.93, "", transform=ax.transAxes)

    ax.set_xlim(x[0], x[-1])
    ax.set_ylim(-0.08, 0.62)
    ax.set_xlabel("x")
    ax.set_ylabel(r"$\Delta u(x,t)$")
    ax.set_title("Propagación finita y primera llegada")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper right")

    def init():
        line.set_data([], [])
        left_boundary.set_data([], [])
        right_boundary.set_data([], [])
        time_text.set_text("")
        state_text.set_text("")
        return (
            line,
            left_boundary,
            right_boundary,
            time_text,
            state_text,
        )

    def update(frame):
        t = animation_times[frame]
        line.set_data(x, delta_u(x, t))
        left_boundary.set_data([xi-c*t, xi-c*t], [-0.02, 0.57])
        right_boundary.set_data([xi+c*t, xi+c*t], [-0.02, 0.57])
        time_text.set_text(f"t = {t:.3f}")

        if t < arrival_time:
            state_text.set_text("observador fuera del cono causal")
        else:
            state_text.set_text("el frente causal alcanzó al observador")

        return (
            line,
            left_boundary,
            right_boundary,
            time_text,
            state_text,
        )

    animation = FuncAnimation(
        fig,
        update,
        init_func=init,
        frames=frames,
        interval=1000/fps,
        blit=True,
    )

    if shutil.which("ffmpeg") is None:
        raise RuntimeError("FFmpeg no está disponible.")

    writer = FFMpegWriter(
        fps=fps,
        bitrate=8000,
        metadata={"title": "Cono causal y rango de influencia"},
    )
    animation.save(video_path, writer=writer, dpi=200)
    plt.close(fig)

display(
    Video(
        str(video_path),
        embed=True,
        html_attributes="controls autoplay loop muted",
    )
)
print(f"Animación guardada en: {video_path}")
```
