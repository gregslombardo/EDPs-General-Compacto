# 2.5.1. Problema en la semirrecta

## **Definición 2.5.1 (Problema de Dirichlet homogéneo en la semirrecta).**

Sean

$$
f\in C^2([0,\infty)),
\qquad
g\in C^1([0,\infty)).
$$

Se considera

$$
\begin{cases}
u_{tt}-c^2u_{xx}=0,
&x\in(0,\infty),\ t>0,\\
u(x,0)=f(x),
&x\in[0,\infty),\\
u_t(x,0)=g(x),
&x\in[0,\infty),\\
u(0,t)=0,
&t>0.
\end{cases}
$$

La condición $u(0,t)=0$ representa un extremo fijo.

## **Definición 2.5.2 (Condiciones de compatibilidad en el origen).**

En las notas aparecen

$$
f(0)=0,
\qquad
g(0)=0.
$$

Estas son las condiciones de compatibilidad de orden cero y uno.

### Aclaración

Si se exige una solución clásica de clase $C^2$ hasta la esquina $(0,0)$ y se usa una extensión impar de clase $C^2$, debe imponerse además

$$
f''(0)=0.
$$

Esta condición también aparece explícitamente al final de la página.

# 2.5.2. Extensión impar

## **Definición 2.5.3 (Extensiones impares de los datos).**

Definimos

$$
\widetilde f(x)
=
\begin{cases}
f(x),&x\geq0,\\
-f(-x),&x<0,
\end{cases}
$$

y

$$
\widetilde g(x)
=
\begin{cases}
g(x),&x\geq0,\\
-g(-x),&x<0.
\end{cases}
$$

Entonces

$$
\widetilde f(-x)=-\widetilde f(x),
\qquad
\widetilde g(-x)=-\widetilde g(x).
$$

## **Proposición 2.5.4 (Regularidad de las extensiones impares).**

Supongamos

$$
f\in C^2([0,\infty)),
\qquad
f(0)=f''(0)=0.
$$

Entonces

$$
\widetilde f\in C^2(\mathbb R).
$$

Si

$$
g\in C^1([0,\infty)),
\qquad
g(0)=0,
$$

entonces

$$
\widetilde g\in C^1(\mathbb R).
$$

### Demostración

La continuidad de la extensión impar de $f$ en $0$ exige $f(0)=0$. Las derivadas laterales de primer orden coinciden automáticamente. Para la segunda derivada,

$$
\widetilde f''(0^-)
=
-f''(0),
\qquad
\widetilde f''(0^+)
=
f''(0).
$$

Por tanto, la continuidad de la segunda derivada exige

$$
f''(0)=0.
$$

El argumento para $g$ es análogo, pero sólo hasta orden uno.

$\square$

# 2.5.3. Fórmula por reflexión

## **Teorema 2.5.5 (Método de reflexión impar).**

Bajo las hipótesis anteriores, sea $\widetilde u$ la solución del problema global de Cauchy en $\mathbb R$ con datos $\widetilde f$ y $\widetilde g$.

Entonces

$$
u(x,t)
=
\widetilde u(x,t),
\qquad
x\geq0,
$$

es la única solución clásica del problema en la semirrecta.

De manera explícita,

$$
\boxed{
u(x,t)
=
\frac12
\left[
\widetilde f(x-ct)
+
\widetilde f(x+ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}
\widetilde g(y)\,dy.
}
$$

### Demostración

La solución global existe por la fórmula de d’Alembert. Como los datos extendidos son impares, la unicidad del problema global implica

$$
\widetilde u(-x,t)
=
-\widetilde u(x,t).
$$

Por tanto,

$$
\widetilde u(0,t)=0.
$$

La restricción a $x\geq0$ satisface la ecuación, los datos originales y la condición de frontera.

$\square$

## **Corolario 2.5.6 (Fórmula por regiones de las notas).**

Para $x\geq ct$,

$$
u(x,t)
=
\frac12
\left[
f(x-ct)+f(x+ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}g(y)\,dy.
$$

Para $0\leq x<ct$,

$$
\boxed{
u(x,t)
=
\frac12
\left[
f(x+ct)-f(ct-x)
\right]
+
\frac{1}{2c}
\int_{ct-x}^{ct+x}g(y)\,dy.
}
$$

### Reconstrucción probable

La segunda rama está parcialmente comprimida en el manuscrito. La expresión anterior es la forma obtenida al sustituir las extensiones impares en la fórmula de d’Alembert.

## **Observación 2.5.7 (Reflexión con inversión de signo).**

El término $-f(ct-x)$ representa la imagen reflejada del pulso incidente. El signo negativo es la inversión producida por el extremo fijo.

# 2.5.4. Escena Manim

La escena deja visible el perfil inicial durante cuatro segundos. Después muestra:

- la pared $x=0$;
- la extensión impar;
- el pulso incidente;
- el pulso reflejado con signo opuesto.

El video no se compila en esta entrega.

Script:

`manim/reflexion_impar_semirrecta.py`

Render directo:

```bash
manim -qh --fps 60 manim/reflexion_impar_semirrecta.py ReflexionImparSemirrecta
```

````python
from __future__ import annotations

import os
import tempfile
from pathlib import Path

import numpy as np
from manim import *

config.pixel_width = 2560
config.pixel_height = 1440
config.frame_rate = 60
config.background_color = "#0f1115"
config.media_dir = str(Path(__file__).resolve().parent.parent / "animaciones")

try:
    config.no_latex_cleanup = True
except Exception:
    pass

try:
    tex_dir = Path(tempfile.gettempdir()) / f"manim_tex_halfline_{os.getpid()}"
    tex_dir.mkdir(parents=True, exist_ok=True)
    config.tex_dir = str(tex_dir)
except Exception:
    pass


class ReflexionImparSemirrecta(Scene):
    def construct(self):
        title = Text(
            "Ecuación de onda en la semirrecta",
            font_size=46,
        ).to_edge(UP)

        problem = MathTex(
            r"u_{tt}-c^2u_{xx}=0,\qquad x>0,\qquad u(0,t)=0"
        ).scale(0.86).next_to(title, DOWN, buff=0.35)

        self.play(Write(title), Write(problem), run_time=2.4)
        self.wait(1.5)

        axes = Axes(
            x_range=[-5, 7, 1],
            y_range=[-1.2, 1.2, 0.5],
            x_length=11.2,
            y_length=4.4,
            axis_config={"include_tip": True},
        ).shift(DOWN * 0.9)

        labels = axes.get_axis_labels(MathTex("x"), MathTex("u"))

        wall = Line(
            axes.c2p(0, -1.15),
            axes.c2p(0, 1.15),
            stroke_width=6,
        )
        wall_label = MathTex(r"x=0").next_to(wall, UP)

        self.play(Create(axes), Write(labels), Create(wall), Write(wall_label), run_time=2.5)

        def bump(x, center=2.2, radius=0.72):
            q = (x-center)/radius
            if abs(q) >= 1:
                return 0.0
            return np.exp(-1/(1-q*q))/np.exp(-1)

        profile = axes.plot(
            lambda x: bump(x),
            x_range=[0, 7],
        )
        profile_label = MathTex(r"f(x)").next_to(profile, UP, buff=0.2)

        self.play(Create(profile), Write(profile_label), run_time=2.2)

        # Perfil inicial visible antes de iniciar cualquier movimiento.
        self.wait(4.0)

        extension_label = MathTex(
            r"\widetilde f(x)=\begin{cases}"
            r"f(x),&x\geq0,\\"
            r"-f(-x),&x<0"
            r"\end{cases}"
        ).scale(0.74).to_corner(UL).shift(DOWN * 1.4)

        mirror_profile = axes.plot(
            lambda x: -bump(-x),
            x_range=[-5, 0],
        )

        self.play(Write(extension_label), run_time=2.2)
        self.play(Create(mirror_profile), run_time=2.4)
        self.wait(3.0)

        formula = MathTex(
            r"u(x,t)=\frac12\bigl[\widetilde f(x-ct)+\widetilde f(x+ct)\bigr]"
        ).scale(0.78).to_edge(DOWN)

        self.play(Write(formula), run_time=2.4)
        self.wait(2.0)

        incident = profile.copy()
        reflected = mirror_profile.copy()

        self.play(
            incident.animate.shift(LEFT * 2.2),
            reflected.animate.shift(RIGHT * 2.2),
            run_time=4.8,
        )
        self.wait(2.5)

        inversion = MathTex(
            r"\text{la reflexión en }x=0\text{ invierte el signo}"
        ).scale(0.78).next_to(formula, UP, buff=0.35)

        self.play(Write(inversion), run_time=2.0)
        self.wait(4.0)


# Render directo al ejecutar:
# manim -qh --fps 60 reflexion_impar_semirrecta.py ReflexionImparSemirrecta
````
