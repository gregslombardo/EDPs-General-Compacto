# 2.6.1. Problema mixto en un intervalo

## **Definición 2.6.1 (Problema de Dirichlet homogéneo en un intervalo).**

Sean $L>0$ y

$$
f\in C^2([0,L]),
\qquad
g\in C^1([0,L]).
$$

Se considera

$$
\begin{cases}
u_{tt}-c^2u_{xx}=0,
&x\in(0,L),\ t>0,\\
u(x,0)=f(x),
&x\in[0,L],\\
u_t(x,0)=g(x),
&x\in[0,L],\\
u(0,t)=0,
&t>0,\\
u(L,t)=0,
&t>0.
\end{cases}
$$

Los dos extremos permanecen fijos.

## **Definición 2.6.2 (Condiciones de compatibilidad).**

En las notas se escriben

$$
f(0)=f(L)=0,
$$

y

$$
g(0)=g(L)=0.
$$

Estas condiciones garantizan que los datos iniciales concuerdan con la condición de frontera en las esquinas.

### Complemento para regularidad clásica

Para que la extensión impar periódica de $f$ sea de clase $C^2$, se requiere además

$$
f''(0)=f''(L)=0.
$$

# 2.6.2. Extensión impar periódica

## **Definición 2.6.3 (Extensión impar $2L$-periódica).**

Primero definimos en $[-L,L]$

$$
\widetilde f(x)
=
\begin{cases}
f(x),&0\leq x\leq L,\\
-f(-x),&-L\leq x<0,
\end{cases}
$$

y después imponemos

$$
\widetilde f(x+2L)=\widetilde f(x).
$$

Se define $\widetilde g$ de la misma manera.

La extensión es impar respecto de $0$ y también respecto de $L$:

$$
\widetilde f(-x)=-\widetilde f(x),
$$

$$
\widetilde f(L+x)=-\widetilde f(L-x).
$$

## **Proposición 2.6.4 (Regularidad de la extensión periódica).**

Si

$$
f\in C^2([0,L]),
$$

y

$$
f(0)=f(L)=f''(0)=f''(L)=0,
$$

entonces

$$
\widetilde f\in C^2(\mathbb R).
$$

Si

$$
g\in C^1([0,L]),
$$

y

$$
g(0)=g(L)=0,
$$

entonces

$$
\widetilde g\in C^1(\mathbb R).
$$

### Demostración

La regularidad se verifica en los puntos de reflexión $kL$. En cada extremo, la continuidad de la función exige la anulación del valor, mientras que la continuidad de la segunda derivada de una extensión impar exige la anulación de la segunda derivada.

$\square$

# 2.6.3. Solución por reflexiones múltiples

## **Teorema 2.6.5 (Método de reflexión periódica para extremos fijos).**

Sean $\widetilde f$ y $\widetilde g$ las extensiones impares $2L$-periódicas de los datos.

Entonces

$$
\boxed{
u(x,t)
=
\frac12
\left[
\widetilde f(x-ct)+\widetilde f(x+ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}
\widetilde g(y)\,dy,
}
$$

restringida a

$$
0\leq x\leq L,
$$

satisface el problema de Dirichlet homogéneo.

### Demostración

La fórmula resuelve la ecuación de onda en toda la recta. La imparidad respecto de $0$ implica

$$
u(0,t)=0.
$$

La imparidad respecto de $L$ implica

$$
u(L,t)=0.
$$

En $t=0$ se recuperan $f$ y $g$ por restricción.

$\square$

## **Observación 2.6.6 (Interpretación geométrica).**

La extensión periódica reemplaza las reflexiones sucesivas por una propagación libre en una sucesión infinita de copias del intervalo.

Cada choque con un extremo fijo invierte el signo del pulso.

# 2.6.4. Escena Manim

La escena:

1. muestra primero el intervalo y el perfil inicial;
2. mantiene el perfil visible durante cuatro segundos;
3. introduce las condiciones de compatibilidad;
4. muestra la extensión periódica;
5. anima dos reflexiones sucesivas con inversión de signo.

El video no se compila en esta entrega.

Script:

`manim/reflexiones_en_intervalo.py`

Render directo:

```bash
manim -qh --fps 60 manim/reflexiones_en_intervalo.py ReflexionesEnIntervalo
```

````python
from __future__ import annotations

import os
import tempfile
from pathlib import Path

import numpy as np
from manim import *

config.pixel_width = 2560
config.pixel_height = 1440
config.frame_rate = 60
config.background_color = "#0f1115"
config.media_dir = str(Path(__file__).resolve().parent.parent / "animaciones")

try:
    config.no_latex_cleanup = True
except Exception:
    pass

try:
    tex_dir = Path(tempfile.gettempdir()) / f"manim_tex_interval_{os.getpid()}"
    tex_dir.mkdir(parents=True, exist_ok=True)
    config.tex_dir = str(tex_dir)
except Exception:
    pass


class ReflexionesEnIntervalo(Scene):
    def construct(self):
        title = Text(
            "Onda en un intervalo con extremos fijos",
            font_size=44,
        ).to_edge(UP)

        problem = MathTex(
            r"u_{tt}-c^2u_{xx}=0,\qquad 0<x<L,\qquad u(0,t)=u(L,t)=0"
        ).scale(0.78).next_to(title, DOWN, buff=0.35)

        self.play(Write(title), Write(problem), run_time=2.4)
        self.wait(1.5)

        axes = Axes(
            x_range=[-0.3, 6.3, 1],
            y_range=[-1.2, 1.2, 0.5],
            x_length=10.8,
            y_length=4.4,
            axis_config={"include_tip": True},
        ).shift(DOWN * 0.9)

        labels = axes.get_axis_labels(MathTex("x"), MathTex("u"))

        L = 6.0
        left_wall = Line(axes.c2p(0, -1.15), axes.c2p(0, 1.15), stroke_width=6)
        right_wall = Line(axes.c2p(L, -1.15), axes.c2p(L, 1.15), stroke_width=6)

        self.play(
            Create(axes),
            Write(labels),
            Create(left_wall),
            Create(right_wall),
            run_time=2.5,
        )

        def bump(x, center=1.7, radius=0.72):
            q = (x-center)/radius
            if abs(q) >= 1:
                return 0.0
            return np.exp(-1/(1-q*q))/np.exp(-1)

        profile = axes.plot(
            lambda x: bump(x),
            x_range=[0, L],
        )
        profile_label = MathTex(r"f(x)").next_to(profile, UP, buff=0.2)

        self.play(Create(profile), Write(profile_label), run_time=2.2)

        # Perfil inicial visible antes de la evolución.
        self.wait(4.0)

        compatibility = MathTex(
            r"f(0)=f(L)=g(0)=g(L)=0"
        ).scale(0.78).to_corner(UL).shift(DOWN * 1.3)

        self.play(Write(compatibility), run_time=2.0)
        self.wait(2.0)

        extension = MathTex(
            r"\widetilde f(x+2L)=\widetilde f(x),"
            r"\qquad \widetilde f(-x)=-\widetilde f(x)"
        ).scale(0.72).to_edge(DOWN)

        self.play(Write(extension), run_time=2.4)
        self.wait(2.0)

        pulse = profile.copy()

        self.play(pulse.animate.shift(RIGHT * 5.1), run_time=4.8)
        self.wait(1.5)

        reflected = pulse.copy().rotate(PI, axis=RIGHT).set_opacity(1.0)
        self.play(Transform(pulse, reflected), run_time=1.2)
        self.play(pulse.animate.shift(LEFT * 5.2), run_time=4.8)
        self.wait(1.5)

        reflected_again = pulse.copy().rotate(PI, axis=RIGHT).set_opacity(1.0)
        self.play(Transform(pulse, reflected_again), run_time=1.2)
        self.play(pulse.animate.shift(RIGHT * 3.2), run_time=3.2)

        statement = MathTex(
            r"\text{cada reflexión en un extremo fijo cambia el signo}"
        ).scale(0.72).next_to(extension, UP, buff=0.35)

        self.play(Write(statement), run_time=2.0)
        self.wait(4.0)


# Render directo:
# manim -qh --fps 60 reflexiones_en_intervalo.py ReflexionesEnIntervalo
````
