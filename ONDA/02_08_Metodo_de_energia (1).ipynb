# 2.4.1. Problema no homogéneo

## **Definición 2.4.1 (Problema global de Cauchy no homogéneo).**

Sean

$$
f\in C^2(\mathbb R),
\qquad
g\in C^1(\mathbb R),
$$

y sea

$$
h:\mathbb R\times[0,\infty)\to\mathbb R.
$$

Se considera

$$
\begin{cases}
u_{tt}-c^2u_{xx}=h(x,t),
&x\in\mathbb R,\ t>0,\\[1mm]
u(x,0)=f(x),\\[1mm]
u_t(x,0)=g(x).
\end{cases}
$$

La función $h$ se denomina **término fuente** o **fuerza externa**.

# 2.4.2. Problema auxiliar iniciado en $t=s$

## **Definición 2.4.2 (Problema elemental de Duhamel).**

Para cada $s\geq0$, sea $w=w(x,t;s)$ la solución de

$$
\begin{cases}
w_{tt}-c^2w_{xx}=0,
&t>s,\\[1mm]
w(x,s;s)=0,\\[1mm]
w_t(x,s;s)=h(x,s).
\end{cases}
$$

La fuente en el instante $s$ se interpreta como una velocidad inicial aplicada en ese instante.

## **Proposición 2.4.3 (Solución del problema elemental).**

Para $t\geq s$,

$$
\boxed{
w(x,t;s)
=
\frac{1}{2c}
\int_{x-c(t-s)}^{x+c(t-s)}
h(y,s)\,dy.
}
$$

### Demostración

Se aplica la fórmula de d’Alembert con tiempo inicial $s$, desplazamiento inicial nulo y velocidad inicial $h(\cdot,s)$.

$\square$

# 2.4.3. Superposición temporal

## **Teorema 2.4.4 (Principio de Duhamel).**

Supongamos que $h$ posee la regularidad necesaria para diferenciar bajo el signo integral. Definamos

$$
u_p(x,t)
=
\int_0^t w(x,t;s)\,ds.
$$

Entonces $u_p$ satisface

$$
\begin{cases}
(u_p)_{tt}-c^2(u_p)_{xx}=h(x,t),\\[1mm]
u_p(x,0)=0,\\[1mm]
(u_p)_t(x,0)=0.
\end{cases}
$$

En particular,

$$
\boxed{
u_p(x,t)
=
\frac{1}{2c}
\int_0^t
\int_{x-c(t-s)}^{x+c(t-s)}
h(y,s)\,dy\,ds.
}
$$

### Demostración

Como

$$
u_p(x,t)=\int_0^t w(x,t;s)\,ds,
$$

la regla de Leibniz da

$$
(u_p)_t(x,t)
=
w(x,t;t)
+
\int_0^t w_t(x,t;s)\,ds.
$$

Puesto que

$$
w(x,t;t)=0,
$$

resulta

$$
(u_p)_t(x,t)
=
\int_0^t w_t(x,t;s)\,ds.
$$

Derivando otra vez,

$$
(u_p)_{tt}(x,t)
=
w_t(x,t;t)
+
\int_0^t w_{tt}(x,t;s)\,ds.
$$

Por la condición inicial del problema elemental,

$$
w_t(x,t;t)=h(x,t).
$$

Además,

$$
w_{tt}=c^2w_{xx}.
$$

Por tanto,

$$
(u_p)_{tt}
=
h(x,t)
+
c^2\int_0^t w_{xx}(x,t;s)\,ds.
$$

La integral es precisamente $(u_p)_{xx}$. Así,

$$
(u_p)_{tt}-c^2(u_p)_{xx}=h.
$$

Los datos iniciales nulos se obtienen directamente de la definición.

$\square$

## **Corolario 2.4.5 (Fórmula completa para el problema no homogéneo).**

La solución del problema con datos $(f,g)$ y fuente $h$ es

$$
\boxed{
\begin{aligned}
u(x,t)
&=
\frac12\bigl[f(x-ct)+f(x+ct)\bigr]\\
&\quad+
\frac{1}{2c}
\int_{x-ct}^{x+ct}g(y)\,dy\\
&\quad+
\frac{1}{2c}
\int_0^t
\int_{x-c(t-s)}^{x+c(t-s)}
h(y,s)\,dy\,ds.
\end{aligned}
}
$$

La primera línea corresponde al desplazamiento inicial, la segunda a la velocidad inicial y la tercera a la fuente.

## **Definición 2.4.6 (Dominio de dependencia de la fuente).**

Para un punto $(x,t)$, definimos

$$
\boxed{
D(x,t)
=
\left\{
(y,s):
0\leq s\leq t,\ 
|x-y|\leq c(t-s)
\right\}.
}
$$

La contribución de la fuente a $u(x,t)$ depende únicamente de los valores de $h$ dentro de este triángulo característico.

# 2.4.4. Visualización con Manim

La escena se ha diseñado con ritmo lento:

1. presenta primero el perfil espacial $h(x,s)$;
2. lo deja visible durante cuatro segundos;
3. introduce el problema elemental;
4. muestra la separación del impulso en dos ondas;
5. superpone los impulsos aplicados en todos los tiempos $s$;
6. termina con la fórmula integral y el dominio de dependencia.

El video no se compila en esta entrega.

Script:

`manim/principio_de_Duhamel.py`

Render de alta calidad:

```bash
manim -qh --fps 60 manim/principio_de_Duhamel.py PrincipioDeDuhamel
```

````python
from __future__ import annotations

import os
import tempfile
from pathlib import Path

import numpy as np
from manim import *

config.pixel_width = 2560
config.pixel_height = 1440
config.frame_rate = 60
config.background_color = "#0f1115"
config.media_dir = str(Path(__file__).resolve().parent.parent / "animaciones")

try:
    config.no_latex_cleanup = True
except Exception:
    pass

try:
    tex_dir = Path(tempfile.gettempdir()) / f"manim_tex_duhamel_{os.getpid()}"
    tex_dir.mkdir(parents=True, exist_ok=True)
    config.tex_dir = str(tex_dir)
except Exception:
    pass


class PrincipioDeDuhamel(Scene):
    def construct(self):
        title = Text(
            "Principio de Duhamel para la ecuación de onda",
            font_size=44,
        ).to_edge(UP)

        problem = MathTex(
            r"u_{tt}-c^2u_{xx}=h(x,t)"
        ).next_to(title, DOWN, buff=0.35)

        self.play(Write(title), Write(problem), run_time=2.4)
        self.wait(2.0)

        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-0.2, 1.3, 0.5],
            x_length=10.5,
            y_length=3.0,
            axis_config={"include_tip": True},
        ).shift(DOWN * 1.35)

        labels = axes.get_axis_labels(
            MathTex("x"),
            MathTex("h(x,s)"),
        )

        self.play(Create(axes), Write(labels), run_time=2.2)

        source_graph = axes.plot(
            lambda x: np.exp(-2.2 * (x + 1.2) ** 2),
            x_range=[-5, 5],
        )

        source_label = MathTex(
            r"h(x,s)"
        ).next_to(source_graph, UP, buff=0.2)

        self.play(Create(source_graph), Write(source_label), run_time=2.5)

        # Pausa larga para observar el perfil fuente antes de la dinámica.
        self.wait(4.0)

        aux_problem = VGroup(
            MathTex(r"w_{tt}-c^2w_{xx}=0"),
            MathTex(r"w(x,s;s)=0"),
            MathTex(r"w_t(x,s;s)=h(x,s)"),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.25).scale(0.8)
        aux_problem.to_corner(UL).shift(DOWN * 1.55)

        self.play(Write(aux_problem), run_time=3.2)
        self.wait(2.5)

        impulse_text = MathTex(
            r"\text{impulso aplicado en el instante }s"
        ).scale(0.75).to_edge(DOWN)

        self.play(Write(impulse_text), run_time=1.8)
        self.wait(2.0)

        # Reemplazamos el perfil fuente por dos pulsos que se separan.
        left_pulse = axes.plot(
            lambda x: 0.5 * np.exp(-2.2 * (x + 1.2) ** 2),
            x_range=[-5, 5],
        )
        right_pulse = left_pulse.copy()

        self.play(
            Transform(source_graph, left_pulse),
            FadeOut(source_label),
            run_time=1.6,
        )

        self.play(
            source_graph.animate.shift(LEFT * 2.3),
            FadeIn(right_pulse),
            right_pulse.animate.shift(RIGHT * 2.3),
            run_time=4.5,
        )
        self.wait(3.0)

        formula_w = MathTex(
            r"w(x,t;s)=\frac{1}{2c}"
            r"\int_{x-c(t-s)}^{x+c(t-s)}h(y,s)\,dy"
        ).scale(0.82).to_edge(DOWN)

        self.play(
            ReplacementTransform(impulse_text, formula_w),
            run_time=2.6,
        )
        self.wait(4.0)

        self.play(
            FadeOut(axes),
            FadeOut(labels),
            FadeOut(source_graph),
            FadeOut(right_pulse),
            FadeOut(aux_problem),
            run_time=1.8,
        )

        superposition = MathTex(
            r"u_p(x,t)=\int_0^t w(x,t;s)\,ds"
        ).scale(1.0)

        self.play(Write(superposition), run_time=2.5)
        self.wait(3.0)

        final_formula = MathTex(
            r"\boxed{"
            r"u_p(x,t)=\frac{1}{2c}"
            r"\int_0^t"
            r"\int_{x-c(t-s)}^{x+c(t-s)}"
            r"h(y,s)\,dy\,ds"
            r"}"
        ).scale(0.88)

        self.play(
            TransformMatchingTex(superposition, final_formula),
            FadeOut(formula_w),
            run_time=3.4,
        )
        self.wait(5.0)

        cone = VGroup(
            Line(DOWN * 2.2 + LEFT * 3.2, UP * 1.3),
            Line(DOWN * 2.2 + RIGHT * 3.2, UP * 1.3),
            Line(DOWN * 2.2 + LEFT * 3.2, DOWN * 2.2 + RIGHT * 3.2),
        ).shift(DOWN * 0.6)

        cone_label = MathTex(
            r"D(x,t)=\{(y,s):0\leq s\leq t,\ |x-y|\leq c(t-s)\}"
        ).scale(0.67).to_edge(DOWN)

        self.play(final_formula.animate.to_edge(UP).shift(DOWN * 0.9), run_time=2.0)
        self.play(Create(cone), Write(cone_label), run_time=3.0)
        self.wait(4.0)
````
