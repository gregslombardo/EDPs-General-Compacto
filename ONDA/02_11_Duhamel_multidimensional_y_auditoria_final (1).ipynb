# 2.1.1. Problema global de Cauchy

## **Definición 2.1.1 (Problema global de Cauchy para la ecuación de onda).**

En las notas se plantea

$$
\begin{cases}
u_{tt}-c^2u_{xx}=0,
& x\in\mathbb R,\ t>0,\\[1mm]
u(x,0)=f(x),\\[1mm]
u_t(x,0)=g(x),
\end{cases}
\tag{2.1}
$$

con $f$ y $g$ funciones conocidas.

### Aclaración de regularidad

En el encabezado manuscrito parece indicarse que ambas funciones son de clase $C^2(\mathbb R)$. Más adelante, el **Teorema 1** de las notas usa las hipótesis

$$
f\in C^2(\mathbb R),
\qquad
g\in C^1(\mathbb R).
$$

Conservaremos estas últimas hipótesis al formular el teorema.

En las notas se factoriza el operador como

$$
u_{tt}-c^2u_{xx}
=
(\partial_t+c\partial_x)
(\partial_t-c\partial_x)u,
$$

y se anota que se trata de una factorización **bicaracterística**.

## **Proposición 2.1.2 (Factorización bicaracterística del operador de onda).**

Sea $u\in C^2(\mathbb R\times(0,\infty))$. Entonces

$$
u_{tt}-c^2u_{xx}
=
(\partial_t+c\partial_x)
(\partial_t-c\partial_x)u.
$$

### Demostración

Como las derivadas mixtas conmutan,

$$
\begin{aligned}
(\partial_t+c\partial_x)
(\partial_t-c\partial_x)u
&=
\partial_t(u_t-cu_x)
+
c\partial_x(u_t-cu_x)\\
&=
u_{tt}
-cu_{xt}
+cu_{tx}
-c^2u_{xx}\\
&=
u_{tt}-c^2u_{xx}.
\end{aligned}
$$

$\square$

# 2.1.2. Reducción a una ecuación de transporte

Las notas definen

$$
w(x,t):=u_t(x,t)-cu_x(x,t).
$$

En $t=0$,

$$
w(x,0)=g(x)-cf'(x).
$$

La factorización anterior muestra que $w$ satisface

$$
\begin{cases}
w_t+cw_x=0,
&x\in\mathbb R,\ t>0,\\[1mm]
w(x,0)=g(x)-cf'(x),
&x\in\mathbb R.
\end{cases}
\tag{2.2}
$$

La solución escrita en las notas es

$$
w(x,t)
=
g(x-ct)-cf'(x-ct).
\tag{2.3}
$$

Sustituyendo la definición de $w$, se obtiene el problema de primer orden para $u$:

$$
\begin{cases}
u_t-cu_x
=
g(x-ct)-cf'(x-ct),
&x\in\mathbb R,\ t>0,\\[1mm]
u(x,0)=f(x).
\end{cases}
\tag{2.4}
$$

## **Proposición 2.1.3 (Reducción del problema de onda a transporte).**

Sean

$$
f\in C^2(\mathbb R),
\qquad
g\in C^1(\mathbb R),
$$

y sea

$$
u\in
C^2(\mathbb R\times(0,\infty))
\cap
C^1(\mathbb R\times[0,\infty))
$$

una solución de (2.1).

Entonces la función

$$
w=u_t-cu_x
$$

satisface (2.2), y por tanto está dada por (2.3).

### Demostración

Por la factorización,

$$
0
=
u_{tt}-c^2u_{xx}
=
(\partial_t+c\partial_x)(u_t-cu_x).
$$

Esto equivale a

$$
w_t+cw_x=0.
$$

Además,

$$
w(x,0)
=
u_t(x,0)-cu_x(x,0)
=
g(x)-cf'(x).
$$

La ecuación de transporte conserva el dato sobre las rectas

$$
x-ct=\text{constante}.
$$

Por ello,

$$
w(x,t)
=
w(x-ct,0)
=
g(x-ct)-cf'(x-ct).
$$

$\square$

# 2.1.3. Fórmula de d’Alembert

En la parte inferior de la página aparece

$$
\boxed{
u(x,t)
=
\frac12
\left[
f(x+ct)+f(x-ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}g(y)\,dy
}
\tag{2.5}
$$

con la anotación **“Fórmula de d’Alembert”**.

Las notas interpretan:

- el primer término como el promedio de $f$ en los extremos característicos;
- el segundo término como el promedio integral de $g$ sobre el intervalo característico.

## **Teorema 2.1.4 (Fórmula de d’Alembert; Teorema 1 de las notas).**

Sean

$$
f\in C^2(\mathbb R),
\qquad
g\in C^1(\mathbb R).
$$

Definamos

$$
u(x,t)
=
\frac12
\left[
f(x-ct)+f(x+ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}g(y)\,dy.
$$

Entonces

$$
u\in
C^2(\mathbb R\times(0,\infty))
\cap
C^1(\mathbb R\times[0,\infty))
$$

y $u$ es solución del problema global de Cauchy (2.1).

Además,

$$
u(x,t)=W(x-ct)+V(x+ct),
$$

donde

$$
W(z)
=
\frac12f(z)
-
\frac{1}{2c}\int_0^z g(y)\,dy,
$$

$$
V(z)
=
\frac12f(z)
+
\frac{1}{2c}\int_0^z g(y)\,dy.
$$

### Demostración

Partimos de

$$
u_t-cu_x
=
g(x-ct)-cf'(x-ct).
$$

Fijemos $(x,t)$ y parametrizamos la característica retrógrada mediante

$$
X(s)=x+c(t-s),
\qquad
0\leq s\leq t.
$$

Entonces

$$
X(t)=x,
\qquad
X(0)=x+ct,
$$

y

$$
\frac{d}{ds}u(X(s),s)
=
u_t(X(s),s)-cu_x(X(s),s).
$$

Por tanto,

$$
\frac{d}{ds}u(X(s),s)
=
g(x+ct-2cs)-cf'(x+ct-2cs).
$$

Integramos de $0$ a $t$:

$$
u(x,t)-f(x+ct)
=
\int_0^t
\left[
g(x+ct-2cs)-cf'(x+ct-2cs)
\right]ds.
$$

Con el cambio

$$
y=x+ct-2cs,
\qquad
dy=-2c\,ds,
$$

obtenemos

$$
u(x,t)-f(x+ct)
=
\frac{1}{2c}
\int_{x-ct}^{x+ct}g(y)\,dy
-
\frac12
\int_{x-ct}^{x+ct}f'(y)\,dy.
$$

Finalmente,

$$
\int_{x-ct}^{x+ct}f'(y)\,dy
=
f(x+ct)-f(x-ct),
$$

y se obtiene la fórmula indicada.

$\square$

## **Resultado añadido 2.1.5 (Unicidad de la solución clásica).**

Bajo las hipótesis del Teorema 2.1.4, la solución dada por la fórmula de d’Alembert es única en la clase

$$
C^2(\mathbb R\times(0,\infty))
\cap
C^1(\mathbb R\times[0,\infty)).
$$

### Comentario

La página 1 construye la solución, pero no desarrolla explícitamente la unicidad. Se añade porque existencia y unicidad forman parte del problema global de Cauchy.

### Demostración

Si $u_1$ y $u_2$ son soluciones, entonces $z=u_1-u_2$ tiene datos iniciales nulos. La fórmula de d’Alembert aplicada a $z$ da

$$
z(x,t)=0.
$$

Luego $u_1=u_2$.

$\square$

# 2.1.4. Simulación — descomposición en ondas viajeras

La simulación muestra

$$
u(x,t)=W(x-ct)+V(x+ct).
$$

La animación se genera automáticamente al ejecutar la celda. No hay que activar ninguna bandera como `VIDEO=True`.

En esta fórmula explícita unidimensional, NumPy es más eficiente que trasladar datos continuamente entre GPU y CPU. CUDA se reservará para simulaciones multidimensionales y barridos donde sí aporte una mejora real.

```python
from __future__ import annotations

import shutil
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import Video, display

FIG_DIR = Path("figuras")
ANIM_DIR = Path("animaciones")
FIG_DIR.mkdir(exist_ok=True)
ANIM_DIR.mkdir(exist_ok=True)

c = 1.0
x_min, x_max = -7.0, 7.0
x = np.linspace(x_min, x_max, 1800)

def bump(z, center=0.0, radius=1.0):
    z = np.asarray(z, dtype=float)
    q = (z-center)/radius
    out = np.zeros_like(q)
    mask = np.abs(q) < 1.0
    out[mask] = np.exp(-1.0/(1.0-q[mask]**2))
    if np.any(mask):
        out /= np.exp(-1.0)
    return out

def f(z):
    return bump(z, center=0.0, radius=1.0)

def g(z):
    h = 1.0e-4
    return 0.32*(
        bump(z+h, 0.0, 0.8)-bump(z-h, 0.0, 0.8)
    )/(2*h)

x_ext = np.linspace(-12.0, 12.0, 14000)
g_ext = g(x_ext)
G_ext = np.zeros_like(x_ext)
G_ext[1:] = np.cumsum(
    0.5*(g_ext[:-1]+g_ext[1:])*np.diff(x_ext)
)

def primitive_g(z):
    return np.interp(
        z,
        x_ext,
        G_ext,
        left=G_ext[0],
        right=G_ext[-1],
    )

def W(z):
    return 0.5*f(z)-primitive_g(z)/(2*c)

def V(z):
    return 0.5*f(z)+primitive_g(z)/(2*c)

def u(x_values, t):
    return W(x_values-c*t)+V(x_values+c*t)

print("Datos y solución de d'Alembert construidos.")
```

```python
dx = x[1]-x[0]
dt = 0.45*dx/c

u0 = u(x, 0.0)
ut0 = (u(x, dt)-u(x, -dt))/(2*dt)

error_f = np.max(np.abs(u0-f(x)))
error_g = np.max(np.abs(ut0-g(x)))

t_test = 1.15
u_mid = u(x, t_test)
utt = (u(x, t_test+dt)-2*u_mid+u(x, t_test-dt))/dt**2
uxx = np.zeros_like(x)
uxx[1:-1] = (
    u_mid[2:]-2*u_mid[1:-1]+u_mid[:-2]
)/dx**2
residual = utt-c**2*uxx

print(f"max |u(x,0)-f(x)|   = {error_f:.3e}")
print(f"max |u_t(x,0)-g(x)| = {error_g:.3e}")
print(
    "max interior |u_tt-c^2u_xx| = "
    f"{np.max(np.abs(residual[20:-20])):.3e}"
)
```

```python
t_snapshot = 2.1

plt.figure(figsize=(12.8, 7.2))
plt.plot(x, W(x-c*t_snapshot), linewidth=2.0, label=r"$W(x-ct)$")
plt.plot(x, V(x+c*t_snapshot), linewidth=2.0, label=r"$V(x+ct)$")
plt.plot(x, u(x, t_snapshot), linewidth=2.6, label=r"$u(x,t)$")
plt.xlabel("x")
plt.ylabel("amplitud")
plt.title(f"Descomposición de d'Alembert en t={t_snapshot}")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

figure_path = FIG_DIR/"02_01_descomposicion_ondas_viajeras.png"
plt.savefig(figure_path, dpi=200, bbox_inches="tight")
plt.show()

print(f"Figura guardada en: {figure_path}")
```

```python
video_path = ANIM_DIR/"02_01_dAlembert_2K_120fps.mp4"

if not video_path.exists():
    fps = 120
    frames = 180
    times = np.linspace(0.0, 4.2, frames)

    fig, ax = plt.subplots(figsize=(12.8, 7.2), dpi=200)

    line_w, = ax.plot([], [], linewidth=1.8, label=r"$W(x-ct)$")
    line_v, = ax.plot([], [], linewidth=1.8, label=r"$V(x+ct)$")
    line_u, = ax.plot([], [], linewidth=2.8, label=r"$u(x,t)$")
    time_text = ax.text(0.02, 0.93, "", transform=ax.transAxes)

    ax.set_xlim(x_min, x_max)
    ax.set_ylim(-0.35, 1.15)
    ax.set_xlabel("x")
    ax.set_ylabel("amplitud")
    ax.set_title("Fórmula de d'Alembert: dos ondas viajeras")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper right")

    def init():
        for line in (line_w, line_v, line_u):
            line.set_data([], [])
        time_text.set_text("")
        return line_w, line_v, line_u, time_text

    def update(frame):
        t = times[frame]
        line_w.set_data(x, W(x-c*t))
        line_v.set_data(x, V(x+c*t))
        line_u.set_data(x, u(x, t))
        time_text.set_text(f"t = {t:.3f}")
        return line_w, line_v, line_u, time_text

    animation = FuncAnimation(
        fig,
        update,
        init_func=init,
        frames=frames,
        interval=1000/fps,
        blit=True,
    )

    if shutil.which("ffmpeg") is None:
        raise RuntimeError("FFmpeg no está disponible.")

    writer = FFMpegWriter(
        fps=fps,
        bitrate=8000,
        metadata={"title": "D'Alembert — ondas viajeras"},
    )
    animation.save(video_path, writer=writer, dpi=200)
    plt.close(fig)

display(
    Video(
        str(video_path),
        embed=True,
        html_attributes="controls autoplay loop muted",
    )
)
print(f"Animación guardada en: {video_path}")
```
