# Propiedad del promedio para funciones armónicas

Este notebook contiene el enunciado formal y cuatro escenas Manim en resolución 2K, a 120 fps y con duración amplia. Las escenas no se renderizan al abrir el notebook; cada una se genera únicamente al ejecutar su celda de render.

## Enunciado

Sea $\Omega\subset\mathbb{R}^n$ un abierto y sea $u\in C^2(\Omega)$ una función armónica, es decir,

$$
\Delta u=0
\qquad\text{en }\Omega.
$$

Si $x_0\in\Omega$ y $r>0$ son tales que

$$
\overline{B_r(x_0)}\subset\Omega,
$$

entonces

$$
u(x_0)
=
\frac{1}{|\partial B_r(x_0)|}
\int_{\partial B_r(x_0)}u\,dS.
$$

Además,

$$
u(x_0)
=
\frac{1}{|B_r(x_0)|}
\int_{B_r(x_0)}u(x)\,dx.
$$

Por tanto, el valor de una función armónica en el centro de una bola coincide tanto con el promedio de sus valores sobre la esfera como con el promedio de sus valores en toda la bola.

## Interpretación

La propiedad afirma que una función armónica no puede tener en el interior un valor aislado que sea mayor o menor que todos los valores cercanos. El valor central está exactamente equilibrado por los valores que lo rodean.

En dimensión dos, para un disco $B_r(a)$,

$$
u(a)
=
\frac{1}{2\pi r}
\int_{\partial B_r(a)}u\,ds
=
\frac{1}{\pi r^2}
\int_{B_r(a)}u(x,y)\,dx\,dy.
$$

Las animaciones usan la función

$$
u(x,y)=x^2-y^2+0.6x-0.4y+1.2,
$$

para la cual

$$
\Delta u=2-2=0.
$$

También se compara con

$$
v(x,y)=x^2+y^2+0.25x-0.15y+0.8,
$$

que satisface

$$
\Delta v=4
$$

y no cumple la propiedad del promedio.

## Render 1 — Promedio sobre circunferencias

Muestra cómo el radio cambia mientras el promedio sobre la circunferencia permanece igual al valor en el centro.

```python
import subprocess, sys

subprocess.run(
    [
        sys.executable,
        "-m",
        "manim",
        "-qh",
        "--resolution", "2560,1440",
        "--fps", "120",
        str(SCRIPT),
        "PromedioCircunferencia",
    ],
    check=True,
)
```

## Render 3 — Comparación armónica y no armónica

La función armónica conserva el promedio para todo radio. La función no armónica produce un promedio que cambia como $r^2$.

## Render 4 — Resumen formal

## Render de todas las escenas

Esta celda genera las cuatro animaciones consecutivamente.

```python
import subprocess, sys

for scene in [
    "PromedioCircunferencia",
    "PromedioDisco",
    "ComparacionArmonicaNoArmonica",
    "ResumenPropiedadPromedio",
]:
    subprocess.run(
        [
            sys.executable,
            "-m",
            "manim",
            "-qh",
            "--resolution", "2560,1440",
            "--fps", "120",
            str(SCRIPT),
            scene,
        ],
        check=True,
    )
```

## Qué observar

1. Para la función armónica, el promedio sobre cada circunferencia centrada en $a$ coincide con $u(a)$, independientemente del radio.
2. El promedio sobre el disco también coincide con $u(a)$.
3. Para la función no armónica, el promedio sobre la circunferencia es

$$
\operatorname{prom}_{\partial B_r(a)}v
=
v(a)+r^2,
$$

por lo que la discrepancia aumenta con el radio.
4. La propiedad del promedio no es solamente una consecuencia de la harmonicidad: bajo hipótesis adecuadas, también caracteriza a las funciones armónicas.
