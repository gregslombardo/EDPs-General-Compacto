## Núcleo teórico

El núcleo fundamental es
\[
\psi(x,t)=(4\pi t)^{-n/2}e^{-|x|^2/(4t)},\qquad t>0.
\]
La solución global es \(u(x,t)=(\psi_t*g)(x)\). El núcleo tiene masa uno,
escala parabólicamente y aproxima la identidad cuando \(t\downarrow0\).
Además \(\|\psi_t\|_\infty=(4\pi t)^{-n/2}\).

## Interpretación geométrica o física

El calor tiene propagación instantánea: una masa puntual se vuelve una
gaussiana positiva en todo el espacio. Su anchura crece como \(\sqrt t\) y su
altura decrece como \(t^{-n/2}\).

## Ejemplo guiado

Por simetría, \(\int x_i\psi\,dx=0\). Cada coordenada tiene varianza
\(2t\), por lo que \(\int|x|^2\psi\,dx=2nt\).

## Práctica esencial

La cuenta es literal: dos ejercicios normales. Sólo los temas marcados llevan un tercero de nivel General.

### Ejercicio normal 04.01.1

Demuestra \(\int x_i\psi(x,t)\,dx=0\) y \(\int|x|^2\psi(x,t)\,dx=2nt\).

### Ejercicio normal 04.01.2

Prueba \(\|u(\cdot,t)\|_\infty\le(4\pi t)^{-n/2}\|g\|_1\).

### Problema tipo General 04.01.G

Sea \(g\in L^1(\mathbb R^n)\) con masa \(M\) y primer momento finito.
Demuestre que \(t^{n/2}u(\sqrt t\,y,t)\) converge al perfil gaussiano de masa
\(M\), y obtenga una estimación del error usando el primer momento.
