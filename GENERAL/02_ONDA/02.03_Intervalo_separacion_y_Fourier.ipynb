## Núcleo teórico

En \(0<x<L\) con extremos fijos, \(u(0,t)=u(L,t)=0\), se busca
\(u=X(x)T(t)\). Resulta
\[
X_n=\sin(n\pi x/L),\qquad \omega_n=cn\pi/L,
\]
\[
u=\sum_{n\ge1}\left(A_n\cos\omega_nt+B_n\sin\omega_nt\right)
\sin(n\pi x/L).
\]
Los coeficientes se obtienen de las series seno de \(f\) y \(g\).

## Interpretación geométrica o física

Los modos normales son patrones estacionarios con nodos en la frontera.
Cada modo oscila con frecuencia propia; la solución total es su
superposición.

## Ejemplo guiado

Si \(f(x)=\sin(\pi x/L)\), \(g=0\), sólo se excita el primer modo:
\(u=\cos(c\pi t/L)\sin(\pi x/L)\).
