# 02.04 — Problema no homogéneo y principio de Duhamel

**Prioridad para el General:** MEDIA  
**Patrón histórico:** Duhamel conecta las versiones no homogéneas en una y varias dimensiones.

> Notebook canónico de la ruta compacta. Para estudiar el examen, usa éste en lugar de las versiones extensas o legadas.

## Núcleo teórico

Para
\[
u_{tt}-c^2u_{xx}=F(x,t),\qquad u(\cdot,0)=u_t(\cdot,0)=0,
\]
Duhamel superpone soluciones homogéneas creadas en cada tiempo \(s\):
\[
u(x,t)=\frac1{2c}\int_0^t\int_{x-c(t-s)}^{x+c(t-s)}
F(y,s)\,dy\,ds.
\]
Con datos no nulos se suma la solución de d'Alembert.

## Interpretación geométrica o física

Cada punto de la fuerza emite una onda elemental. Sólo contribuyen los
puntos del soporte de \(F\) contenidos en el cono pasado de \((x,t)\).

## Ejemplo guiado

Para \(F(x,t)=e^{-t}\cos(kx)\), se usa la identidad trigonométrica
en la integral espacial y queda una integral temporal escalar con frecuencia
\(ck\).
