# 2.1.2. Transcripción de las notas

## Problema mixto

Se considera

$$
\begin{cases}
u_t+a\,u_x=0,\\
u(x,0)=u_0(x),\\
u(0,t)=g(t),
\end{cases}
\qquad x>0,\quad t>0.
$$

Para $a>0$, la característica que pasa por $(x,t)$ está dada por

$$
x-a t=\text{constante}.
$$

La recta

$$
x=at
$$

divide el dominio en dos regiones.

La solución tiene la forma

$$
u(x,t)=
\begin{cases}
u_0(x-at), & x>at,\\
g\!\left(t-\dfrac{x}{a}\right), & x<at.
\end{cases}
$$

La función $u$ es continua en $x=at$ si

$$
g(0)=u_0(0).
$$

En las notas se plantea la pregunta:

> ¿Qué falta para que $u\in C^1$?

---

## Principio universal de balance

En la misma página se introduce la identidad

$$
\int_V \left(u_t+\nabla\cdot F-g\right)\,dV=0,
$$

y, mediante localización,

$$
u_t+\nabla\cdot F=g.
$$

Este tema se desarrollará por separado en la sección 2.1.3.

# 2.1.3. Formulación rigurosa del problema mixto

## Definición 2.1.9 — Problema mixto de transporte

Sea $a>0$ y sea

$$
Q:=\{(x,t)\in\mathbb{R}^2:x>0,\ t>0\}.
$$

Dados

$$
u_0:[0,\infty)\longrightarrow\mathbb{R},
\qquad
g:[0,\infty)\longrightarrow\mathbb{R},
$$

el problema mixto consiste en encontrar $u$ tal que

$$
\begin{cases}
u_t+a\,u_x=0, & (x,t)\in Q,\\
u(x,0)=u_0(x), & x\ge 0,\\
u(0,t)=g(t), & t\ge 0.
\end{cases}
$$

---

## Aclaración 2.1.10 — Por qué se supone $a>0$

Cuando $a>0$, las características avanzan hacia la derecha:

$$
X'(s)=a>0.
$$

Por ello, al retroceder desde un punto interior $(x,t)$, una característica puede alcanzar:

- el eje inicial $t=0$;
- o la frontera de entrada $x=0$.

La frontera $x=0$ aporta información al dominio.

Si $a<0$, la velocidad apunta hacia la izquierda y $x=0$ es una frontera de salida. En ese caso, imponer simultáneamente un dato de frontera arbitrario suele sobreespecificar el problema.

---

## Proposición 2.1.11 — Punto de procedencia de una característica

La característica que pasa por $(x,t)\in Q$ es

$$
X(s)=x+a(s-t).
$$

Al retroceder:

- alcanza $t=0$ en el punto $x-at$;
- alcanza $x=0$ en el tiempo $t-\dfrac{x}{a}$.

### Demostración

La condición $X(t)=x$ y la ecuación $X'(s)=a$ dan

$$
X(s)=x+a(s-t).
$$

En $s=0$,

$$
X(0)=x-at.
$$

Por otro lado, $X(s)=0$ si y sólo si

$$
0=x+a(s-t),
$$

es decir,

$$
s=t-\frac{x}{a}.
$$

---

## Proposición 2.1.12 — División del dominio

Para $a>0$:

- si $x>at$, entonces $x-at>0$ y la característica llega primero al eje $t=0$;
- si $x<at$, entonces $t-\dfrac{x}{a}>0$ y la característica llega primero a la frontera $x=0$;
- si $x=at$, la característica llega a la esquina $(0,0)$.

La recta $x=at$ es, por tanto, la frontera entre las regiones de influencia de los dos conjuntos de datos.

---

## Teorema 2.1.13 — Fórmula de representación

Supongamos $a>0$. Si existe una solución clásica del problema mixto, necesariamente satisface

$$
u(x,t)=
\begin{cases}
u_0(x-at), & x\ge at,\\
g\!\left(t-\dfrac{x}{a}\right), & 0\le x\le at.
\end{cases}
$$

### Demostración

Sea $(x,t)\in Q$.

Si $x>at$, la característica retrocede hasta $(x-at,0)$. Como $u$ es constante sobre ella,

$$
u(x,t)=u(x-at,0)=u_0(x-at).
$$

Si $x<at$, la característica retrocede hasta

$$
\left(0,t-\frac{x}{a}\right).
$$

Por constancia característica,

$$
u(x,t)
=
u\!\left(0,t-\frac{x}{a}\right)
=
g\!\left(t-\frac{x}{a}\right).
$$

Sobre $x=at$, ambas expresiones remiten a la esquina $(0,0)$.

---

## Proposición 2.1.14 — Compatibilidad de orden cero

La solución por partes es continua sobre la recta $x=at$ si y sólo si

$$
\boxed{g(0)=u_0(0).}
$$

### Demostración

Al aproximarse a $x=at$ desde la región $x>at$,

$$
u(x,t)\longrightarrow u_0(0).
$$

Al aproximarse desde la región $x<at$,

$$
u(x,t)\longrightarrow g(0).
$$

Los límites coinciden exactamente cuando

$$
g(0)=u_0(0).
$$

---

## Resultado añadido 2.1.15 — Compatibilidad de orden uno

Supongamos

$$
u_0\in C^1([0,\infty)),
\qquad
g\in C^1([0,\infty)).
$$

La solución por partes pertenece a $C^1$ a través de la recta $x=at$ si y sólo si

$$
\boxed{g(0)=u_0(0)}
$$

y

$$
\boxed{g'(0)=-a\,u_0'(0).}
$$

### Demostración

En la región $x>at$,

$$
u(x,t)=u_0(x-at),
$$

de donde

$$
u_x=u_0'(x-at),
\qquad
u_t=-a\,u_0'(x-at).
$$

Sobre la recta $x=at$, los límites desde esta región son

$$
u_x\longrightarrow u_0'(0),
\qquad
u_t\longrightarrow -a\,u_0'(0).
$$

En la región $x<at$,

$$
u(x,t)=g\!\left(t-\frac{x}{a}\right),
$$

por lo que

$$
u_x
=
-\frac{1}{a}
g'\!\left(t-\frac{x}{a}\right),
$$

y

$$
u_t
=
g'\!\left(t-\frac{x}{a}\right).
$$

Sobre $x=at$,

$$
u_x\longrightarrow-\frac{1}{a}g'(0),
\qquad
u_t\longrightarrow g'(0).
$$

La continuidad de $u_x$ equivale a

$$
u_0'(0)=-\frac{1}{a}g'(0),
$$

es decir,

$$
g'(0)=-a\,u_0'(0).
$$

La misma condición garantiza la continuidad de $u_t$.

---

## Teorema 2.1.16 — Existencia y unicidad clásica

Sean

$$
u_0\in C^1([0,\infty)),
\qquad
g\in C^1([0,\infty)),
$$

y supongamos que

$$
g(0)=u_0(0),
\qquad
g'(0)=-a\,u_0'(0).
$$

Entonces la función

$$
u(x,t)=
\begin{cases}
u_0(x-at), & x\ge at,\\
g\!\left(t-\dfrac{x}{a}\right), & 0\le x\le at,
\end{cases}
$$

es la única solución de clase $C^1$ del problema mixto.

### Demostración

Cada rama es $C^1$ en su región y satisface

$$
u_t+a\,u_x=0.
$$

Las condiciones de compatibilidad garantizan que tanto $u$ como sus derivadas de primer orden coinciden sobre $x=at$. Por tanto, $u\in C^1$ en todo el dominio cerrado.

La unicidad se obtiene siguiendo hacia atrás la única característica que pasa por cada punto interior. Esa característica alcanza exactamente uno de los conjuntos de datos, salvo sobre $x=at$, donde ambos datos coinciden por compatibilidad.

$$

````python
%%manim -v WARNING -r 2560,1440 --fps 120 ProblemaMixtoCompatibilidad

from manim import *
import numpy as np

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_width = "100%"
config.media_embed = True

A = 1.0
T_FINAL = 4.0


def u0(x):
    return np.exp(-(x**2))


def g_compatible(t):
    return np.exp(-(t**2))


def g_incompatible(t):
    return 0.5 * np.exp(-(t**2))


class ProblemaMixtoCompatibilidad(Scene):
    def construct(self):
        self.camera.background_color = "#0B1020"

        COLOR_AXIS = GREY_B
        COLOR_INITIAL = TEAL_C
        COLOR_BOUNDARY = ORANGE
        COLOR_DIVIDER = YELLOW_C
        COLOR_REGION_INITIAL = BLUE_D
        COLOR_REGION_BOUNDARY = PURPLE_D
        COLOR_INCOMPATIBLE = RED_C

        title = Text(
            "Problema mixto para la ecuación de transporte",
            font_size=38,
            weight=BOLD,
            color=WHITE,
        ).to_edge(UP, buff=0.20)

        equation = MathTex(
            r"u_t+a\,u_x=0",
            r"\qquad",
            r"u(x,0)=u_0(x)",
            r"\qquad",
            r"u(0,t)=g(t)",
            font_size=33,
        )
        equation[2].set_color(COLOR_INITIAL)
        equation[4].set_color(COLOR_BOUNDARY)
        equation.next_to(title, DOWN, buff=0.14)

        self.play(
            FadeIn(title, shift=0.12 * DOWN),
            Write(equation),
            run_time=1.5,
        )

        # ========================================================
        # 1. PLANO ESPACIO-TIEMPO
        # ========================================================

        xt_axes = Axes(
            x_range=[0, 5.2, 1],
            y_range=[0, 4.4, 1],
            x_length=6.0,
            y_length=4.7,
            tips=False,
            axis_config={
                "color": COLOR_AXIS,
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 21,
            },
        ).move_to(3.45 * LEFT + 0.45 * DOWN)

        xt_labels = xt_axes.get_axis_labels(
            MathTex(r"x", font_size=30),
            MathTex(r"t", font_size=30),
        )

        xt_heading = Text(
            "Dominio de influencia",
            font_size=28,
            weight=SEMIBOLD,
            color=WHITE,
        ).next_to(xt_axes, UP, buff=0.18)

        divider = Line(
            xt_axes.c2p(0, 0),
            xt_axes.c2p(A*T_FINAL, T_FINAL),
            color=COLOR_DIVIDER,
            stroke_width=6,
        )

        divider_label = MathTex(
            r"x=at",
            font_size=27,
            color=COLOR_DIVIDER,
        ).next_to(
            xt_axes.c2p(3.2, 3.2),
            RIGHT,
            buff=0.10,
        )

        initial_region = Polygon(
            xt_axes.c2p(0, 0),
            xt_axes.c2p(5.0, 0),
            xt_axes.c2p(5.0, 4.0),
            xt_axes.c2p(4.0, 4.0),
            fill_color=COLOR_REGION_INITIAL,
            fill_opacity=0.22,
            stroke_opacity=0,
        )

        boundary_region = Polygon(
            xt_axes.c2p(0, 0),
            xt_axes.c2p(4.0, 4.0),
            xt_axes.c2p(0, 4.0),
            fill_color=COLOR_REGION_BOUNDARY,
            fill_opacity=0.22,
            stroke_opacity=0,
        )

        initial_region_label = VGroup(
            Text(
                "Dato inicial",
                font_size=24,
                color=COLOR_INITIAL,
            ),
            MathTex(
                r"u_0(x-at)",
                font_size=26,
                color=COLOR_INITIAL,
            ),
        ).arrange(DOWN, buff=0.06)
        initial_region_label.move_to(xt_axes.c2p(4.05, 1.45))

        boundary_region_label = VGroup(
            Text(
                "Dato de frontera",
                font_size=24,
                color=COLOR_BOUNDARY,
            ),
            MathTex(
                r"g\!\left(t-\frac{x}{a}\right)",
                font_size=26,
                color=COLOR_BOUNDARY,
            ),
        ).arrange(DOWN, buff=0.06)
        boundary_region_label.move_to(xt_axes.c2p(1.15, 2.75))

        initial_data_line = Line(
            xt_axes.c2p(0, 0),
            xt_axes.c2p(5.0, 0),
            color=COLOR_INITIAL,
            stroke_width=7,
        )

        boundary_data_line = Line(
            xt_axes.c2p(0, 0),
            xt_axes.c2p(0, 4.0),
            color=COLOR_BOUNDARY,
            stroke_width=7,
        )

        # ========================================================
        # 2. PANEL DE PERFILES
        # ========================================================

        profile_axes = Axes(
            x_range=[0, 5.2, 1],
            y_range=[0, 1.2, 0.25],
            x_length=6.0,
            y_length=4.7,
            tips=False,
            axis_config={
                "color": COLOR_AXIS,
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 21,
            },
        ).move_to(3.45 * RIGHT + 0.45 * DOWN)

        profile_labels = profile_axes.get_axis_labels(
            MathTex(r"x", font_size=30),
            MathTex(r"u", font_size=30),
        )

        profile_heading = Text(
            "Solución compatible",
            font_size=28,
            weight=SEMIBOLD,
            color=WHITE,
        ).next_to(profile_axes, UP, buff=0.18)

        initial_profile = profile_axes.plot(
            u0,
            x_range=[0, 5.0],
            color=COLOR_INITIAL,
            stroke_width=5,
            use_smoothing=True,
        )

        initial_profile_label = MathTex(
            r"u_0(x)=e^{-x^2}",
            font_size=26,
            color=COLOR_INITIAL,
        ).move_to(profile_axes.c2p(2.0, 1.02))

        boundary_dot = Dot(
            profile_axes.c2p(0, g_compatible(0)),
            radius=0.08,
            color=COLOR_BOUNDARY,
        )

        boundary_label = MathTex(
            r"g(0)=u_0(0)=1",
            font_size=25,
            color=COLOR_BOUNDARY,
        ).move_to(profile_axes.c2p(3.7, 0.78))

        # Present the initial geometry and data before evolution.
        self.play(
            FadeIn(xt_axes),
            FadeIn(xt_labels),
            FadeIn(xt_heading),
            FadeIn(profile_axes),
            FadeIn(profile_labels),
            FadeIn(profile_heading),
            run_time=1.4,
        )

        self.play(
            FadeIn(initial_region),
            FadeIn(boundary_region),
            Create(initial_data_line),
            Create(boundary_data_line),
            Create(divider),
            FadeIn(divider_label),
            FadeIn(initial_region_label),
            FadeIn(boundary_region_label),
            Create(initial_profile),
            FadeIn(initial_profile_label),
            FadeIn(boundary_dot),
            FadeIn(boundary_label),
            run_time=2.0,
        )

        # Required pause: the viewer sees the initial data before motion.
        self.wait(4.0)

        # ========================================================
        # 3. EVOLUCIÓN DE LA SOLUCIÓN
        # ========================================================

        time = ValueTracker(0.0)

        time_tex = MathTex(r"t=", font_size=27, color=GREY_A)
        time_number = DecimalNumber(
            0.0,
            num_decimal_places=2,
            font_size=28,
            color=WHITE,
        )
        time_number.add_updater(
            lambda mob: mob.set_value(time.get_value())
        )

        time_display = VGroup(time_tex, time_number)
        time_display.arrange(RIGHT, buff=0.05)
        time_display.move_to(0.18 * DOWN)

        solution_curve = always_redraw(
            lambda: profile_axes.plot(
                lambda x: (
                    u0(x - A*time.get_value())
                    if x >= A*time.get_value()
                    else g_compatible(time.get_value() - x/A)
                ),
                x_range=[0, 5.0],
                color=YELLOW_C,
                stroke_width=5,
                use_smoothing=True,
            )
        )

        moving_interface = always_redraw(
            lambda: DashedLine(
                profile_axes.c2p(A*time.get_value(), 0),
                profile_axes.c2p(A*time.get_value(), 1.12),
                color=COLOR_DIVIDER,
                stroke_width=2,
                dash_length=0.09,
            )
        )

        interface_tex = MathTex(
            r"x=at",
            font_size=25,
            color=COLOR_DIVIDER,
        )

        interface_number = DecimalNumber(
            0.0,
            num_decimal_places=2,
            font_size=25,
            color=COLOR_DIVIDER,
        )
        interface_number.add_updater(
            lambda mob: mob.set_value(A*time.get_value())
        )

        interface_display = VGroup(
            interface_tex,
            interface_number,
        ).arrange(RIGHT, buff=0.08)
        interface_display.next_to(profile_axes, DOWN, buff=0.18)

        self.play(
            FadeOut(initial_profile_label),
            FadeIn(solution_curve),
            FadeIn(moving_interface),
            FadeIn(time_display),
            FadeIn(interface_display),
            run_time=1.0,
        )

        self.play(
            time.animate.set_value(T_FINAL),
            run_time=12.0,
            rate_func=linear,
        )

        self.wait(2.0)

        time_number.clear_updaters()
        interface_number.clear_updaters()

        # ========================================================
        # 4. CASO INCOMPATIBLE
        # ========================================================

        incompatible_title = Text(
            "Dato incompatible en la esquina",
            font_size=30,
            weight=SEMIBOLD,
            color=COLOR_INCOMPATIBLE,
        ).move_to(profile_heading)

        incompatible_formula = MathTex(
            r"\widetilde g(0)=\frac12\ne u_0(0)=1",
            font_size=30,
            color=COLOR_INCOMPATIBLE,
        ).next_to(profile_axes, DOWN, buff=0.18)

        jump_marker = Line(
            profile_axes.c2p(A*T_FINAL, 0.5),
            profile_axes.c2p(A*T_FINAL, 1.0),
            color=COLOR_INCOMPATIBLE,
            stroke_width=7,
        )

        self.play(
            Transform(profile_heading, incompatible_title),
            FadeOut(interface_display),
            FadeIn(incompatible_formula),
            Create(jump_marker),
            run_time=1.4,
        )

        self.wait(4.0)

        # ========================================================
        # 5. CONCLUSIÓN
        # ========================================================

        all_objects = VGroup(
            xt_axes,
            xt_labels,
            xt_heading,
            initial_region,
            boundary_region,
            initial_data_line,
            boundary_data_line,
            divider,
            divider_label,
            initial_region_label,
            boundary_region_label,
            profile_axes,
            profile_labels,
            profile_heading,
            initial_profile,
            boundary_dot,
            boundary_label,
            solution_curve,
            moving_interface,
            time_display,
            incompatible_formula,
            jump_marker,
        )

        self.play(
            FadeOut(all_objects),
            FadeOut(equation),
            run_time=1.2,
        )

        conclusion = VGroup(
            MathTex(
                r"u(x,t)=u_0(x-at),\qquad x\ge at",
                font_size=40,
                color=COLOR_INITIAL,
            ),
            MathTex(
                r"u(x,t)=g\!\left(t-\frac{x}{a}\right),\qquad x\le at",
                font_size=40,
                color=COLOR_BOUNDARY,
            ),
            MathTex(
                r"C^0:\quad g(0)=u_0(0)",
                font_size=42,
                color=YELLOW_C,
            ),
            MathTex(
                r"C^1:\quad g'(0)=-a\,u_0'(0)",
                font_size=42,
                color=GREEN_C,
            ),
        ).arrange(DOWN, buff=0.38)

        box = RoundedRectangle(
            corner_radius=0.16,
            width=conclusion.width + 0.90,
            height=conclusion.height + 0.78,
            stroke_color=BLUE_B,
            stroke_width=2,
            fill_color="#10182B",
            fill_opacity=0.68,
        ).move_to(conclusion)

        self.play(
            FadeIn(box),
            LaggedStart(
                *[Write(line) for line in conclusion],
                lag_ratio=0.22,
            ),
            run_time=3.8,
        )

        self.wait(3.0)
````

## Ejercicio 2.1.6 — Solución por regiones

Sea $a=2$ y considere

$$
\begin{cases}
u_t+2u_x=0, & x>0,\ t>0,\\
