# 2.6.1. Transcripción organizada de las notas

## Condición de entropía de Lax

Una discontinuidad que se propaga sobre

$$
\Sigma=\{(x,t):x=\sigma t\}
$$

satisface la condición de entropía de Lax si

$$
\boxed{f'(u_R)\leq\sigma\leq f'(u_L)}.
$$

En las notas, una solución débil por pedazos suaves es entrópica si la condición anterior se cumple en cada discontinuidad.

Para el frente

$$
u(x,t)=
\begin{cases}
u_L, & x<\sigma t,\\
u_R, & x>\sigma t,
\end{cases}
$$

la velocidad viene dada por Rankine–Hugoniot:

$$
\sigma=\frac{[f(u)]}{[u]}=
\frac{f(u_R)-f(u_L)}{u_R-u_L}.
$$

Si el frente satisface Lax, se denomina onda de choque.

## Flujo sin cambio de convexidad

Las notas consideran el intervalo entre ambos estados y señalan que, bajo convexidad o concavidad estricta, las desigualdades son estrictas:

$$
f'(u_R)<\sigma<f'(u_L).
$$

Para un flujo estrictamente convexo,

$$
f''(u)\geq\delta>0,
$$

el frente satisface Lax si y sólo si

$$
u_L>u_R.
$$

El caso estrictamente cóncavo es análogo.

## Condición de Oleinik

Sea $f\in C^2(\mathbb{R})$ estrictamente convexa. Una solución débil satisface la condición de Oleinik si existe una constante uniforme $C>0$ tal que

$$
\boxed{
\frac{u(x+a,t)-u(x,t)}{a}\leq\frac{C}{t}
}
$$

para todo

$$
a>0,\qquad x\in\mathbb{R},\qquad t>0.
$$

En las notas se registra

$$
\text{Oleinik}\Longrightarrow\text{Lax}.
$$

# 2.6.2. Aclaraciones de notación e hipótesis

## Aclaración 2.6.1 — Estados y saltos

En esta sección $u_L$ y $u_R$ son las trazas izquierda y derecha, respectivamente, y

$$
[u]=u_R-u_L.
$$

Por tanto,

$$
\sigma=\frac{f(u_R)-f(u_L)}{u_R-u_L}.
$$

## Aclaración 2.6.2 — Desigualdades estrictas

Para asegurar

$$
f'(u_R)<\sigma<f'(u_L)
$$

no basta una formulación ambigua sobre puntos de inflexión. Es suficiente que $f'$ sea estrictamente monótona entre los estados. En particular, basta

$$
f''\geq\delta>0
$$

o

$$
f''\leq-\delta<0.
$$

## Aclaración 2.6.3 — Alcance de Lax

La definición de las notas se aplica a soluciones $C^1$ por pedazos con discontinuidades identificables. Para soluciones generales en $L^\infty$ se requerirán desigualdades entrópicas más robustas, que aparecerán al estudiar contracción en $L^1$.

# 2.6.3. Condición de Lax e interpretación geométrica

## Definición 2.6.4 — Choque de Lax

Un frente de Rankine–Hugoniot es un choque de Lax si

$$
f'(u_R)\leq\sigma\leq f'(u_L).
$$

Es estrictamente compresivo cuando

$$
f'(u_R)<\sigma<f'(u_L).
$$

## Interpretación 2.6.5 — Características que entran al frente

En una región constante, las características satisfacen

$$
\frac{dx}{dt}=f'(u).
$$

A la izquierda del frente su velocidad es $f'(u_L)$ y a la derecha es $f'(u_R)$. Si

$$
f'(u_R)<\sigma<f'(u_L),
$$

las características izquierdas alcanzan al frente y el frente alcanza a las características derechas. La información entra desde ambos lados.

## Definición 2.6.6 — Discontinuidad expansiva

Un frente de Rankine–Hugoniot para el cual

$$
f'(u_L)<\sigma<f'(u_R)
$$

es expansivo. Las características se alejan del frente. Es solución débil, pero no satisface Lax.

# 2.6.4. Flujo estrictamente convexo

## Lema 2.6.7 — Pendiente de la cuerda

Sea $f\in C^2(\mathbb{R})$ y supongamos $f''>0$ entre $u_L$ y $u_R$.

Si $u_R<u_L$, entonces

$$
f'(u_R)
<
\frac{f(u_L)-f(u_R)}{u_L-u_R}
<
f'(u_L).
$$

Si $u_L<u_R$, entonces

$$
f'(u_L)
<
\frac{f(u_R)-f(u_L)}{u_R-u_L}
<
f'(u_R).
$$

### Demostración

Para $u_R<u_L$,

$$
f(u_L)-f(u_R)=\int_{u_R}^{u_L}f'(v)\,dv.
$$

Como $f'$ es estrictamente creciente,

$$
f'(u_R)<f'(v)<f'(u_L)
$$

para $v\in(u_R,u_L)$. Integramos y dividimos por $u_L-u_R>0$. El segundo caso es análogo.

## Teorema 2.6.8 — Criterio convexo

Para un flujo estrictamente convexo, un frente de Rankine–Hugoniot satisface Lax si y sólo si

$$
u_L>u_R.
$$

### Demostración

La velocidad puede escribirse

$$
\sigma=\frac{f(u_L)-f(u_R)}{u_L-u_R}.
$$

Si $u_L>u_R$, el lema da

$$
f'(u_R)<\sigma<f'(u_L).
$$

Si $u_L<u_R$, se obtiene

$$
f'(u_L)<\sigma<f'(u_R),
$$

que corresponde a una discontinuidad expansiva.

## Corolario 2.6.9 — Burgers

Para

$$
f(u)=\frac{u^2}{2},
$$

se tiene

$$
f'(u)=u,
\qquad
\sigma=\frac{u_L+u_R}{2}.
$$

El frente es entrópico exactamente cuando

$$
u_L>u_R.
$$

## Resultado añadido 2.6.10 — Flujo cóncavo

Si $f''<0$ entre los estados, el frente satisface Lax exactamente cuando

$$
u_L<u_R.
$$

# 2.6.5. Condición unilateral de Oleinik

## Definición 2.6.11 — Oleinik

Una solución débil satisface Oleinik si existe $C>0$ tal que

$$
\frac{u(x+a,t)-u(x,t)}{a}
\leq
\frac{C}{t}
$$

para todo $a>0$, $x\in\mathbb{R}$ y $t>0$.

La desigualdad controla únicamente incrementos hacia la derecha y por ello es unilateral.

## Complemento 2.6.12 — Convexidad uniforme

Cuando

$$
f''\geq\delta>0,
$$

la cota natural para soluciones entrópicas es

$$
\frac{u(x+a,t)-u(x,t)}{a}
\leq
\frac{1}{\delta t}.
$$

La demostración global se reserva para el bloque de existencia y unicidad.

## Proposición 2.6.13 — Oleinik excluye saltos crecientes

Supongamos que en $(x_0,t_0)$ existe un salto con

$$
u_R>u_L.
$$

Entonces Oleinik no puede cumplirse.

### Demostración

Tomamos $x<x_0<x+a$ y aproximamos ambos puntos al frente desde lados opuestos. Entonces

$$
u(x+a,t_0)-u(x,t_0)\longrightarrow u_R-u_L>0,
$$

mientras

$$
a\longrightarrow0^+.
$$

Por tanto,

$$
\frac{u(x+a,t_0)-u(x,t_0)}{a}\longrightarrow+\infty,
$$

lo que contradice una cota finita $C/t_0$.

## Teorema 2.6.14 — Oleinik implica Lax

Sea $f$ estrictamente convexa y sea $u$ una solución débil $C^1$ por pedazos que satisface Oleinik. Entonces cada salto no trivial cumple

$$
u_L>u_R
$$

y, por tanto,

$$
f'(u_R)<\sigma<f'(u_L).
$$

### Demostración

Oleinik excluye $u_R>u_L$. Como el salto es no trivial, necesariamente $u_L>u_R$. El Teorema 2.6.8 concluye la condición de Lax.

## Observación 2.6.15 — Diferencia entre ambos criterios

Lax es una condición local sobre discontinuidades. Oleinik es una estimación espacial válida a todas las escalas positivas y contiene información adicional sobre las regiones continuas.

# 2.6.6. Ejemplos y contraejemplos

## Ejemplo 2.6.16 — Choque compresivo de Burgers

Sean

$$
u_L=2,
\qquad
u_R=0.
$$

Entonces

$$
\sigma=1,
$$

y

$$
0=f'(u_R)<1<2=f'(u_L).
$$

El frente es un choque de Lax.

## Contraejemplo 2.6.17 — Choque expansivo débil

Sean

$$
u_L=0,
\qquad
u_R=2.
$$

Rankine–Hugoniot también da

$$
\sigma=1.
$$

Sin embargo,

$$
0=f'(u_L)<1<2=f'(u_R).
$$

La función por estados constantes es débil, pero no entrópica. El salto creciente también viola Oleinik.

## Ejemplo 2.6.18 — Modelo LWR

Para

$$
f(\rho)=\rho(1-\rho),
$$

se tiene

$$
f''(\rho)=-2<0.
$$

La orientación admisible se invierte. Para $\rho_L=0$ y $\rho_R=1$,

$$
\sigma=0,
$$

y

$$
f'(\rho_R)=-1<0<1=f'(\rho_L).
$$

El frente satisface Lax.
