# 2.2.2. Transcripción organizada de las notas

En las notas se considera el problema

$$
\begin{cases}
u_t+a(x,t)\cdot\nabla u=g(x,t),\\
u(x,0)=f(x).
\end{cases}
$$

Las curvas características satisfacen

$$
\frac{dX}{ds}=a(X(s),s),
$$

con un dato inicial arbitrario

$$
X(0)=x_0.
$$

Definiendo

$$
\widetilde u(s)=u(X(s),s)
$$

y

$$
\widetilde g(s)=g(X(s),s),
$$

la regla de la cadena da

$$
\frac{d\widetilde u}{ds}
=
u_t(X(s),s)
+
a(X(s),s)\cdot\nabla u(X(s),s).
$$

Por la EDP,

$$
\boxed{
\frac{d\widetilde u}{ds}
=
\widetilde g(s)
}.
$$

En el caso homogéneo,

$$
g=0,
$$

la función $u$ es constante sobre cada característica.

# 2.2.3. Flujo y fórmula de representación

## Definición 2.2.7 — Flujo directo

Sea

$$
a:\mathbb{R}^d\times[0,T]\longrightarrow\mathbb{R}^d
$$

continua y localmente Lipschitz respecto de $x$.

Para cada $x_0\in\mathbb{R}^d$, la curva

$$
s\longmapsto X(s;x_0)
$$

se define mediante

$$
\begin{cases}
X'(s;x_0)=a(X(s;x_0),s),\\
X(0;x_0)=x_0.
\end{cases}
$$

## Aclaración 2.2.8 — Existencia y unicidad

Bajo las hipótesis anteriores, el teorema de Picard–Lindelöf garantiza una única característica local. Si el campo tiene crecimiento a lo sumo lineal, las características existen durante todo el intervalo considerado.

## Definición 2.2.9 — Característica retrógrada

Para un punto final $(x,t)$ se define

$$
X(s;t,x)
$$

por

$$
\begin{cases}
\dfrac{dX}{ds}(s;t,x)=a(X(s;t,x),s),\\
X(t;t,x)=x.
\end{cases}
$$

La curva retrógrada permite determinar qué punto del dato inicial influye en $(x,t)$.

## Teorema 2.2.10 — Fórmula de representación

Sea $u\in C^1$ una solución del problema de Cauchy. Entonces

$$
\boxed{
u(x,t)
=
f(X(0;t,x))
+
\int_0^t
g(X(s;t,x),s)\,ds
}.
$$

### Demostración

Sobre la característica retrógrada,

$$
\frac{d}{ds}
u(X(s;t,x),s)
=
g(X(s;t,x),s).
$$

Integrando entre $0$ y $t$,

$$
u(X(t;t,x),t)
-
u(X(0;t,x),0)
=
\int_0^t
g(X(s;t,x),s)\,ds.
$$

Como

$$
X(t;t,x)=x
$$

y

$$
u(X(0;t,x),0)=f(X(0;t,x)),
$$

se obtiene la fórmula.

## Corolario 2.2.11 — Ecuación homogénea

Si

$$
g=0,
$$

entonces

$$
u(x,t)=f(X(0;t,x)).
$$

## Ejemplo 2.2.12 — Campo constante

Si

$$
a(x,t)=a_0,
$$

entonces

$$
X(s;t,x)=x-a_0(t-s).
$$

Por tanto,

$$
u(x,t)
=
f(x-a_0t)
+
\int_0^t
g(x-a_0(t-s),s)\,ds.
$$

# 2.2.4. Ejemplo rotacional con fuente

En $\mathbb{R}^2$ sea

$$
a(x,y)=(-y,x).
$$

Las características son rotaciones:

$$
X(s)=
\begin{pmatrix}
\cos s & -\sin s\\
\sin s & \cos s
\end{pmatrix}
x_0.
$$

Si

$$
g(x,y,t)=e^{-t}(x^2+y^2),
$$

entonces $x^2+y^2$ permanece constante sobre las trayectorias. La fórmula de representación produce

$$
u(x,t)
=
f(R_{-t}x)
+
|x|^2
\int_0^t e^{-s}\,ds,
$$

es decir,

$$
u(x,t)
=
f(R_{-t}x)
+
|x|^2(1-e^{-t}).
$$

# 2.2.5. Visualización Manim

La escena muestra:

- el campo rotacional $a(x,y)=(-y,x)$;
- varias trayectorias características;
- un conjunto de partículas;
- la acumulación de la fuente
  $$
  \int_0^t e^{-s}|X(s)|^2\,ds;
  $$
- el tiempo mediante `DecimalNumber`.

Los ejes, el campo y los datos iniciales permanecen visibles durante cuatro segundos antes de comenzar la dinámica.

```python
import os
import shutil
import tempfile
import uuid
from pathlib import Path

def miktex_candidates():
    candidates = []
    for base in (
        os.environ.get("LOCALAPPDATA"),
        os.environ.get("ProgramFiles"),
        os.environ.get("ProgramFiles(x86)"),
    ):
        if not base:
            continue
        root = Path(base)
        candidates.extend([
            root / "Programs" / "MiKTeX" / "miktex" / "bin" / "x64",
            root / "Programs" / "MiKTeX" / "miktex" / "bin",
            root / "MiKTeX" / "miktex" / "bin" / "x64",
            root / "MiKTeX" / "miktex" / "bin",
        ])
    return candidates

for candidate in miktex_candidates():
    if candidate.exists():
        os.environ["PATH"] = (
            str(candidate)
            + os.pathsep
            + os.environ.get("PATH", "")
        )

latex_exe = shutil.which("latex")
dvisvgm_exe = shutil.which("dvisvgm")

if latex_exe is None or dvisvgm_exe is None:
    raise RuntimeError(
        "No se encontraron latex.exe y dvisvgm.exe. "
        "Abre MiKTeX Console, completa la instalación y reinicia el kernel."
    )

from manim import config, __version__ as manim_version

TEX_RUNTIME_DIR = (
    Path(tempfile.gettempdir())
    / f"manim_tex_recuperado_{os.getpid()}_{uuid.uuid4().hex[:10]}"
)
TEX_RUNTIME_DIR.mkdir(parents=True, exist_ok=True)

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_embed = True
config.media_width = "100%"

print(f"Manim Community: {manim_version}")
print("latex   :", latex_exe)
print("dvisvgm :", dvisvgm_exe)
print("TeX dir :", TEX_RUNTIME_DIR)
print("Render preparado: 2560×1440, 120 fps")
print("La animación no ha sido compilada.")
```

```python
%%manim -v WARNING -r 2560,1440 --fps 120 FlujoCaracteristicoRepresentacion

from manim import *
import numpy as np

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_embed = True
config.media_width = "100%"


class FlujoCaracteristicoRepresentacion(Scene):
    def construct(self):
        self.camera.background_color = "#0B1020"

        title = Text(
            "Flujo característico y fórmula de representación",
            font_size=39,
            weight=BOLD,
        ).to_edge(UP, buff=0.18)

        formula = MathTex(
            r"u(x,t)=f(X(0;t,x))+\int_0^t g(X(s;t,x),s)\,ds",
            font_size=35,
            color=YELLOW_C,
        ).next_to(title, DOWN, buff=0.14)

        axes = Axes(
            x_range=[-3.5, 3.5, 1],
            y_range=[-3.5, 3.5, 1],
            x_length=7.1,
            y_length=7.1,
            tips=False,
        ).shift(2.5 * LEFT + 0.35 * DOWN)

        field = ArrowVectorField(
            lambda p: np.array([-p[1], p[0], 0.0]),
            x_range=[-3, 3, 0.8],
            y_range=[-3, 3, 0.8],
            length_func=lambda norm: 0.34,
            colors=[BLUE_D, BLUE_B],
        ).move_to(axes)

        radii = [0.8, 1.4, 2.0, 2.6]
        orbits = VGroup(*[
            Circle(
                radius=axes.x_axis.unit_size * radius,
                color=GREY_B,
                stroke_opacity=0.45,
                stroke_width=2,
            ).move_to(axes.c2p(0, 0))
            for radius in radii
        ])

        angles = np.linspace(0.0, 2 * np.pi, 12, endpoint=False)
        initial_points = [
            (1.8 * np.cos(theta), 1.8 * np.sin(theta))
            for theta in angles
        ]

        particles = VGroup(*[
            Dot(axes.c2p(x, y), radius=0.065, color=ORANGE)
            for x, y in initial_points
        ])

        tracker = ValueTracker(0.0)

        for dot, (x0, y0) in zip(particles, initial_points):
            dot.add_updater(
                lambda mob, x0=x0, y0=y0: mob.move_to(
                    axes.c2p(
                        np.cos(tracker.get_value()) * x0
                        - np.sin(tracker.get_value()) * y0,
                        np.sin(tracker.get_value()) * x0
                        + np.cos(tracker.get_value()) * y0,
                    )
                )
            )

        time_number = DecimalNumber(
            0.0,
            num_decimal_places=2,
            font_size=31,
            color=GREEN_C,
        )
        time_number.add_updater(
            lambda mob: mob.set_value(tracker.get_value())
        )

        source_number = DecimalNumber(
            0.0,
            num_decimal_places=3,
            font_size=31,
            color=ORANGE,
        )
        source_number.add_updater(
            lambda mob: mob.set_value(
                1.8**2 * (1.0 - np.exp(-tracker.get_value()))
            )
        )

        panel = VGroup(
            VGroup(
                MathTex(r"t=", font_size=31, color=GREEN_C),
                time_number,
            ).arrange(RIGHT, buff=0.05),
            MathTex(
                r"a(x,y)=(-y,x)",
                font_size=34,
                color=TEAL_C,
            ),
            VGroup(
                MathTex(
                    r"\int_0^t e^{-s}|X(s)|^2\,ds=",
                    font_size=30,
                    color=ORANGE,
                ),
                source_number,
            ).arrange(RIGHT, buff=0.06),
            MathTex(
                r"|X(s)|=|X(0)|",
                font_size=33,
                color=YELLOW_C,
            ),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.35)
        panel.move_to(3.5 * RIGHT + 0.15 * DOWN)

        self.play(
            FadeIn(title),
            Write(formula),
            Create(axes),
            FadeIn(field),
            Create(orbits),
            FadeIn(particles),
            FadeIn(panel),
            run_time=3.0,
        )

        self.wait(4.0)

        self.play(
            tracker.animate.set_value(2 * np.pi),
            run_time=14.0,
            rate_func=linear,
        )

        self.wait(3.0)

        for dot in particles:
            dot.clear_updaters()
        time_number.clear_updaters()
        source_number.clear_updaters()
```
