# Capítulo 2. Ecuaciones de primer orden
## Sección 2.4.2. Condiciones de compatibilidad y transversalidad generalizada

**Fuente principal:** página 8 de `Ecuacione Lineales De Primer Orden.pdf`.

Se estudia cómo elegir

$$
\widetilde p(\xi)
$$

y

$$
\widetilde q(\xi)
$$

de modo compatible con el dato de Cauchy y con la ecuación

$$
F(x,y,u,p,q)=0.
$$

# 2.4.2. Transcripción organizada de las notas

Los datos iniciales deben satisfacer

$$
F(
\widetilde x(\xi),
\widetilde y(\xi),
f(\xi),
\widetilde p(\xi),
\widetilde q(\xi)
)=0.
$$

Además, como

$$
f(\xi)
=
u(\widetilde x(\xi),\widetilde y(\xi)),
$$

al derivar se obtiene

$$
\boxed{
\widetilde p(\xi)\widetilde x'(\xi)
+
\widetilde q(\xi)\widetilde y'(\xi)
=
f'(\xi)
}.
$$

La condición de transversalidad generalizada se expresa mediante

$$
\boxed{
\det
\begin{pmatrix}
\widetilde x'(\xi_0)
&
F_p(P_0,\widetilde p(\xi_0),\widetilde q(\xi_0))
\\
\widetilde y'(\xi_0)
&
F_q(P_0,\widetilde p(\xi_0),\widetilde q(\xi_0))
\end{pmatrix}
\neq0
},
$$

donde

$$
P_0
=
(
\widetilde x(\xi_0),
\widetilde y(\xi_0),
f(\xi_0)
).
$$

# 2.4.3. Sistema de compatibilidad

## Definición 2.4.6 — Ecuaciones para las pendientes iniciales

Para cada $\xi$ se consideran

$$
H_1(p,q)
=
\widetilde x'(\xi)p
+
\widetilde y'(\xi)q
-
f'(\xi),
$$

y

$$
H_2(p,q)
=
F(
\widetilde x(\xi),
\widetilde y(\xi),
f(\xi),
p,
q
).
$$

Las pendientes compatibles satisfacen

$$
H_1(p,q)=0,
$$

$$
H_2(p,q)=0.
$$

## Proposición 2.4.7 — Jacobiano del sistema de compatibilidad

La matriz derivada respecto de $(p,q)$ es

$$
D_{(p,q)}H
=
\begin{pmatrix}
\widetilde x' & \widetilde y'\\
F_p & F_q
\end{pmatrix}.
$$

Su determinante es

$$
\boxed{
\widetilde x'F_q
-
\widetilde y'F_p
}.
$$

## Teorema 2.4.8 — Pendientes compatibles locales

Si

$$
\widetilde x'F_q-\widetilde y'F_p\neq0
$$

en un punto compatible, entonces el teorema de la función implícita determina localmente de manera única las pendientes iniciales.

## Aclaración 2.4.9 — Significado geométrico

La proyección de la dirección característica es

$$
(F_p,F_q).
$$

La condición

$$
\det
\begin{pmatrix}
\widetilde x' & F_p\\
\widetilde y' & F_q
\end{pmatrix}
\neq0
$$

significa que la curva inicial no es tangente a la proyección característica.

## Teorema 2.4.10 — Existencia local

Supongamos

$$
F\in C^2,
$$

que los datos son de clase $C^1$, que existe una elección compatible de $(\widetilde p,\widetilde q)$ y que se cumple la transversalidad generalizada.

Entonces el sistema de Charpit determina localmente una única solución clásica del problema de Cauchy.

> **Complemento.** La demostración combina Picard–Lindelöf para el sistema de cinco EDO con el teorema de la función inversa aplicado a la proyección $(s,\xi)\mapsto(x,y)$.

# 2.4.4. Ejemplo eikonal sobre una recta

Considere

$$
u_x^2+u_y^2=1
$$

con datos sobre

$$
(\widetilde x(\xi),\widetilde y(\xi))=(\xi,0)
$$

y

$$
u(\xi,0)=f(\xi).
$$

La identidad de contacto da

$$
\widetilde p(\xi)=f'(\xi).
$$

La ecuación eikonal exige

$$
\widetilde p(\xi)^2+\widetilde q(\xi)^2=1.
$$

Por tanto,

$$
\widetilde q(\xi)
=
\pm\sqrt{1-f'(\xi)^2},
$$

si

$$
|f'(\xi)|\leq1.
$$

La transversalidad es

$$
F_q=2\widetilde q\neq0.
$$

Así:

- si $|f'|<1$, existen dos ramas locales;
- si $|f'|=1$, la intersección es tangente y falla la transversalidad;
- si $|f'|>1$, no existen pendientes reales compatibles.

# 2.4.5. Visualización Manim

En el plano $(p,q)$ se representan:

$$
p^2+q^2=1
$$

y

$$
p=m.
$$

Al variar $m$ se observan dos intersecciones, una intersección tangente o ninguna intersección real. El parámetro $m$ se actualiza mediante `DecimalNumber`. Los ejes y datos permanecen visibles durante cuatro segundos.

```python
import os
import shutil
import tempfile
import uuid
from pathlib import Path

def miktex_candidates():
    candidates = []
    for base in (
        os.environ.get("LOCALAPPDATA"),
        os.environ.get("ProgramFiles"),
        os.environ.get("ProgramFiles(x86)"),
    ):
        if not base:
            continue
        root = Path(base)
        candidates.extend([
            root / "Programs" / "MiKTeX" / "miktex" / "bin" / "x64",
            root / "Programs" / "MiKTeX" / "miktex" / "bin",
            root / "MiKTeX" / "miktex" / "bin" / "x64",
            root / "MiKTeX" / "miktex" / "bin",
        ])
    return candidates

for candidate in miktex_candidates():
    if candidate.exists():
        os.environ["PATH"] = (
            str(candidate)
            + os.pathsep
            + os.environ.get("PATH", "")
        )

latex_exe = shutil.which("latex")
dvisvgm_exe = shutil.which("dvisvgm")

if latex_exe is None or dvisvgm_exe is None:
    raise RuntimeError(
        "No se encontraron latex.exe y dvisvgm.exe. "
        "Abre MiKTeX Console, completa la instalación y reinicia el kernel."
    )

from manim import config, __version__ as manim_version

TEX_RUNTIME_DIR = (
    Path(tempfile.gettempdir())
    / f"manim_tex_recuperado_{os.getpid()}_{uuid.uuid4().hex[:10]}"
)
TEX_RUNTIME_DIR.mkdir(parents=True, exist_ok=True)

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_embed = True
config.media_width = "100%"

print(f"Manim Community: {manim_version}")
print("latex   :", latex_exe)
print("dvisvgm :", dvisvgm_exe)
print("TeX dir :", TEX_RUNTIME_DIR)
print("Render preparado: 2560×1440, 120 fps")
print("La animación no ha sido compilada.")
```

```python
%%manim -v WARNING -r 2560,1440 --fps 120 CompatibilidadEikonal

from manim import *
import numpy as np

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_embed = True
config.media_width = "100%"


class CompatibilidadEikonal(Scene):
    def construct(self):
        self.camera.background_color = "#0B1020"

        title = Text(
            "Compatibilidad y transversalidad generalizada",
            font_size=39,
            weight=BOLD,
        ).to_edge(UP, buff=0.18)

        formula = MathTex(
            r"p^2+q^2=1",
            r"\qquad",
            r"p=f'(\xi)=m",
            font_size=36,
            color=YELLOW_C,
        ).next_to(title, DOWN, buff=0.12)

        axes = Axes(
            x_range=[-1.5, 1.5, 0.5],
            y_range=[-1.5, 1.5, 0.5],
            x_length=7.0,
            y_length=7.0,
            tips=False,
        ).shift(2.5 * LEFT + 0.35 * DOWN)

        p_label = MathTex(r"p", font_size=29).next_to(
            axes.x_axis.get_end(),
            RIGHT,
        )
        q_label = MathTex(r"q", font_size=29).next_to(
            axes.y_axis.get_end(),
            UP,
        )

        circle = Circle(
            radius=axes.x_axis.unit_size,
            color=BLUE_B,
            stroke_width=6,
        ).move_to(axes.c2p(0, 0))

        tracker = ValueTracker(-0.8)

        vertical_line = always_redraw(
            lambda: Line(
                axes.c2p(tracker.get_value(), -1.4),
                axes.c2p(tracker.get_value(), 1.4),
                color=ORANGE,
                stroke_width=5,
            )
        )

        intersections = always_redraw(
            lambda: VGroup(
                *(
                    [
                        Dot(
                            axes.c2p(
                                tracker.get_value(),
                                np.sqrt(
                                    max(
                                        0.0,
                                        1.0 - tracker.get_value()**2,
                                    )
                                ),
                            ),
                            radius=0.085,
                            color=GREEN_C,
                        ),
                        Dot(
                            axes.c2p(
                                tracker.get_value(),
                                -np.sqrt(
                                    max(
                                        0.0,
                                        1.0 - tracker.get_value()**2,
                                    )
                                ),
                            ),
                            radius=0.085,
                            color=GREEN_C,
                        ),
                    ]
                    if abs(tracker.get_value()) < 1.0
                    else (
                        [
                            Dot(
                                axes.c2p(
                                    np.sign(tracker.get_value()),
                                    0.0,
                                ),
                                radius=0.09,
                                color=YELLOW_C,
                            )
                        ]
                        if abs(tracker.get_value()) <= 1.0001
                        else []
                    )
                )
            )
        )

        m_number = DecimalNumber(
            -0.8,
            num_decimal_places=2,
            font_size=31,
            color=ORANGE,
        )
        m_number.add_updater(
            lambda mob: mob.set_value(tracker.get_value())
        )

        determinant_number = DecimalNumber(
            1.2,
            num_decimal_places=3,
            font_size=31,
            color=GREEN_C,
        )
        determinant_number.add_updater(
            lambda mob: mob.set_value(
                2.0
                * np.sqrt(
                    max(0.0, 1.0 - tracker.get_value()**2)
                )
            )
        )

        panel = VGroup(
            VGroup(
                MathTex(r"m=", font_size=31, color=ORANGE),
                m_number,
            ).arrange(RIGHT, buff=0.05),
            MathTex(
                r"q=\pm\sqrt{1-m^2}",
                font_size=34,
                color=GREEN_C,
            ),
            VGroup(
                MathTex(
                    r"|\det D_{(p,q)}H|=2|q|=",
                    font_size=29,
                    color=GREEN_C,
                ),
                determinant_number,
            ).arrange(RIGHT, buff=0.05),
            MathTex(
                r"|m|<1:\ 2\ \text{ramas}",
                font_size=30,
            ),
            MathTex(
                r"|m|=1:\ \text{tangencia}",
                font_size=30,
            ),
            MathTex(
                r"|m|>1:\ \text{sin solución real}",
                font_size=30,
            ),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.28)
        panel.move_to(3.5 * RIGHT + 0.10 * DOWN)

        self.play(
            FadeIn(title),
            Write(formula),
            Create(axes),
            FadeIn(p_label),
            FadeIn(q_label),
            Create(circle),
            FadeIn(vertical_line),
            FadeIn(intersections),
            FadeIn(panel),
            run_time=3.0,
        )

        self.wait(4.0)

        self.play(
            tracker.animate.set_value(1.2),
            run_time=14.0,
            rate_func=linear,
        )

        self.wait(3.0)

        m_number.clear_updaters()
        determinant_number.clear_updaters()
```
