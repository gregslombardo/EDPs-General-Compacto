# 2.1.1. Transcripción de las notas

## Ecuación de transporte

**Ecuación (2.1.1).**

$$
u_t+a\,u_x=0.
$$

EDP más simple: $u=u(x,t)$.

$(1,a)$ es la dirección característica; la información se propaga en esta dirección.

---

### Proposición 2.1.1 — Curvas características asociadas a (2.1.1)

Las curvas de la forma

$$
\left\{
\begin{aligned}
&(x(t),t)\in \mathbb{R}\times(0,\infty),\\
&x(t)=at+x_0
\end{aligned}
\right.
$$

se denominan curvas características asociadas a la ecuación (2.1.1).

---

### Proposición 2.1.2 — Constancia sobre características

Si

$$
u=u(x,t)\in C^1\!\left(\mathbb{R}\times[0,\infty)\right)
$$

es solución de (2.1.1), entonces $u$ es constante sobre cualquier característica:

$$
\frac{du}{dt}=0.
$$

---

## Solución del problema de Cauchy

Para cada $(x,t)\in\mathbb{R}\times(0,\infty)$, existe una única característica que pasa por $(x,t)$:

$$
\Sigma_{(x,t)}
=
\left\{
\bigl(x+a(s-t),s\bigr)\in\mathbb{R}^2:s\ge 0
\right\}.
$$

La intersección con los datos iniciales está en $s=0$. Por tanto,

**Fórmula de representación (2.1.2).**

$$
u(x,t)=u(x-at,0)=u_0(x-at).
$$

Si $u_0$ es diferenciable, entonces

$$
u(x,t)=u_0(x-at).
$$

En particular,

$$
u(x,t)=u_0(x-at)
$$

es la única solución de clase $C^1$ si $u_0\in C^1(\mathbb{R})$.

---

### Observación 2.1.3 — Onda viajera

La solución del problema de Cauchy tiene forma de onda viajera: es una cantidad que depende de la traslación $x-at$,

$$
u(x,t)=u_0(x-at).
$$


# 2.1.2. Complemento formal

## Definición 2.1.4 — Solución clásica

Sea $T>0$ y sea

$$
Q_T:=\mathbb{R}\times(0,T).
$$

Una función

$$
u\in C^1\!\left(\mathbb{R}\times[0,T]\right)
$$

se denomina **solución clásica** del problema de Cauchy (2.1.3):

$$
\begin{cases}
u_t+a\,u_x=0, & (x,t)\in Q_T,\\
u(x,0)=u_0(x), & x\in\mathbb{R}.
\end{cases}
$$

si satisface la ecuación y el dato inicial punto a punto.

---

## Demostración de la constancia característica

Sea

$$
X(s)=x+a(s-t).
$$

Entonces $X(t)=x$, $X(0)=x-at$ y $X'(s)=a$. Por la regla de la cadena,

$$
\frac{d}{ds}u(X(s),s)
=
u_t(X(s),s)+X'(s)u_x(X(s),s).
$$

Como $X'(s)=a$,

$$
\frac{d}{ds}u(X(s),s)
=
u_t+a\,u_x
=
0.
$$

Por consiguiente, $u(X(s),s)$ es constante en $s$, y

$$
u(x,t)
=
u(X(t),t)
=
u(X(0),0)
=
u_0(x-at).
$$

---

## Teorema 2.1.5 — Existencia y unicidad clásica

Sea $a\in\mathbb{R}$, $T>0$ y $u_0\in C^1(\mathbb{R})$. Entonces el problema (2.1.3) posee una única solución clásica, dada por

$$
\boxed{u(x,t)=u_0(x-at).}
$$

### Demostración

Definimos

$$
u(x,t):=u_0(x-at).
$$

Como $u_0\in C^1(\mathbb{R})$, se tiene $u\in C^1(\mathbb{R}\times[0,T])$. Además,

$$
u_t(x,t)=-a\,u_0'(x-at),
$$

y

$$
u_x(x,t)=u_0'(x-at).
$$

Por tanto,

$$
u_t+a\,u_x
=
-a\,u_0'(x-at)+a\,u_0'(x-at)
=
0.
$$

Asimismo,

$$
u(x,0)=u_0(x).
$$

Esto prueba existencia.

Para la unicidad, sea $v$ otra solución clásica. La Proposición 2.1.2 implica que $v$ es constante sobre la característica que une $(x-at,0)$ con $(x,t)$. Luego,

$$
v(x,t)=v(x-at,0)=u_0(x-at)=u(x,t).
$$

Así, $v=u$.

---

## Proposición 2.1.6 — Conservación de regularidad

Si $u_0\in C^m(\mathbb{R})$, con $m\ge 1$, entonces

$$
u(x,t)=u_0(x-at)
$$

pertenece a $C^m(\mathbb{R}\times[0,T])$.

### Demostración

Toda derivada respecto de $x$ o $t$ es una combinación constante de derivadas de $u_0$ evaluadas en $x-at$. En particular,

$$
\partial_x^k\partial_t^\ell u(x,t)
=
(-a)^\ell u_0^{(k+\ell)}(x-at),
$$

siempre que $k+\ell\le m$.

---

## Proposición 2.1.7 — Transporte del soporte

Si

$$
\operatorname{supp}u_0\subset K,
$$

entonces, para cada $t\ge 0$,

$$
\operatorname{supp}u(\cdot,t)\subset K+at
:=
\{y+at:y\in K\}.
$$

Si $K$ es cerrado y $u_0$ no se anula idénticamente en ninguna vecindad de sus puntos de soporte, entonces

$$
\operatorname{supp}u(\cdot,t)=K+at.
$$

---

## Proposición 2.1.8 — Conservación de las normas $L^p$

Sea $1\le p<\infty$ y $u_0\in L^p(\mathbb{R})$. Entonces

$$
\|u(\cdot,t)\|_{L^p(\mathbb{R})}
=
\|u_0\|_{L^p(\mathbb{R})}.
$$

También,

$$
\|u(\cdot,t)\|_{L^\infty(\mathbb{R})}
=
\|u_0\|_{L^\infty(\mathbb{R})}.
$$

### Demostración para $1\le p<\infty$

Usando el cambio de variable $y=x-at$,

$$
\begin{aligned}
\|u(\cdot,t)\|_{L^p}^p
&=
\int_{\mathbb{R}}|u_0(x-at)|^p\,dx\\
&=
\int_{\mathbb{R}}|u_0(y)|^p\,dy\\
&=
\|u_0\|_{L^p}^p.
\end{aligned}
$$


# 2.1.3. Animación matemática

En el ejemplo se toma

$$
a=1,
\qquad
u_0(x)=\exp\!\left[-\left(\frac{x+2}{0.85}\right)^2\right].
$$

La característica destacada parte de $\xi=-2$:

$$
X(t)=-2+t.
$$

El máximo del perfil sigue esa trayectoria y conserva el valor

$$
u(X(t),t)=u_0(-2)=1.
$$

### Qué debe observarse

1. Todas las características son rectas paralelas.
2. Cada punto del dato inicial se transporta con velocidad $a$.
3. El perfil no cambia de forma.
4. La amplitud y las normas $L^p$ permanecen invariantes.
5. La información presente en $(x,t)$ proviene del punto $x-at$ del eje inicial.

````python
%%manim -v WARNING -r 2560,1440 --fps 120 TransporteConstanteCaracteristicas

from manim import *
import numpy as np

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_width = "100%"
config.media_embed = True

A = 1.0
XI_SELECTED = -2.0
T_FINAL = 4.0


def u0(x: float) -> float:
    return np.exp(-((x + 2.0) / 0.85) ** 2)


class TransporteConstanteCaracteristicas(Scene):
    def construct(self):
        self.camera.background_color = "#0B1020"

        COLOR_AXIS = GREY_B
        COLOR_CHARS = BLUE_D
        COLOR_SELECTED = YELLOW_C
        COLOR_INITIAL = GREY_C
        COLOR_SOLUTION = TEAL_C

        title = Text(
            "Ecuación de transporte con velocidad constante",
            font_size=38,
            weight=BOLD,
            color=WHITE,
        ).to_edge(UP, buff=0.20)

        formula = MathTex(
            r"u_t+a\,u_x=0",
            r"\qquad",
            r"X(t;\xi)=\xi+at",
            r"\qquad",
            r"u(x,t)=u_0(x-at)",
            font_size=34,
        )
        formula[2].set_color(COLOR_SELECTED)
        formula[4].set_color(COLOR_SOLUTION)
        formula.next_to(title, DOWN, buff=0.14)

        self.play(
            FadeIn(title, shift=0.12 * DOWN),
            Write(formula),
            run_time=1.5,
        )

        time = ValueTracker(0.0)

        time_tex = MathTex(r"t=", font_size=27, color=GREY_A)
        time_number = DecimalNumber(
            0.0,
            num_decimal_places=2,
            font_size=28,
            color=WHITE,
        )
        time_number.add_updater(lambda m: m.set_value(time.get_value()))
        time_display = VGroup(time_tex, time_number).arrange(RIGHT, buff=0.05)
        time_display.move_to(0.18 * DOWN)

        char_axes = Axes(
            x_range=[-6, 6, 2],
            y_range=[0, 4.4, 1],
            x_length=5.85,
            y_length=4.45,
            tips=False,
            axis_config={
                "color": COLOR_AXIS,
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 21,
            },
        ).move_to(3.55 * LEFT + 0.48 * DOWN)

        char_labels = char_axes.get_axis_labels(
            MathTex(r"x", font_size=30),
            MathTex(r"t", font_size=30),
        )

        char_heading = Text(
            "Plano de características",
            font_size=28,
            weight=SEMIBOLD,
            color=WHITE,
        ).next_to(char_axes, UP, buff=0.17)

        xi_values = np.linspace(-5.0, 1.0, 9)
        characteristic_lines = VGroup()

        for xi in xi_values:
            characteristic_lines.add(
                Line(
                    char_axes.c2p(xi, 0.0),
                    char_axes.c2p(xi + A*T_FINAL, T_FINAL),
                    color=COLOR_CHARS,
                    stroke_width=3,
                )
            )

        selected_line = Line(
            char_axes.c2p(XI_SELECTED, 0.0),
            char_axes.c2p(XI_SELECTED + A*T_FINAL, T_FINAL),
            color=COLOR_SELECTED,
            stroke_width=6,
        )

        xi_label = MathTex(
            r"\xi=-2",
            font_size=26,
            color=COLOR_SELECTED,
        ).next_to(char_axes.c2p(XI_SELECTED, 0.0), DOWN, buff=0.10)

        time_slice = always_redraw(
            lambda: DashedLine(
                char_axes.c2p(-6, time.get_value()),
                char_axes.c2p(6, time.get_value()),
                color=GREY_D,
                stroke_width=2,
                dash_length=0.10,
            )
        )

        char_particle = always_redraw(
            lambda: Dot(
                char_axes.c2p(
                    XI_SELECTED + A*time.get_value(),
                    time.get_value(),
                ),
                radius=0.075,
                color=COLOR_SELECTED,
            )
        )

        x_tex = MathTex(r"X(t)=", font_size=25, color=COLOR_SELECTED)
        x_number = DecimalNumber(
            XI_SELECTED,
            num_decimal_places=2,
            font_size=25,
            color=COLOR_SELECTED,
        )
        x_number.add_updater(
            lambda m: m.set_value(XI_SELECTED + A*time.get_value())
        )
        x_display = VGroup(x_tex, x_number).arrange(RIGHT, buff=0.05)
        x_display.next_to(char_axes, DOWN, buff=0.20)

        profile_axes = Axes(
            x_range=[-6, 6, 2],
            y_range=[0, 1.2, 0.25],
            x_length=5.85,
            y_length=4.45,
            tips=False,
            axis_config={
                "color": COLOR_AXIS,
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 21,
            },
        ).move_to(3.55 * RIGHT + 0.48 * DOWN)

        profile_labels = profile_axes.get_axis_labels(
            MathTex(r"x", font_size=30),
            MathTex(r"u", font_size=30),
        )

        profile_heading = Text(
            "Perfil transportado",
            font_size=28,
            weight=SEMIBOLD,
            color=WHITE,
        ).next_to(profile_axes, UP, buff=0.17)

        base_graph = profile_axes.plot(
            u0,
            x_range=[-6, 6],
            color=COLOR_INITIAL,
            stroke_width=3,
            use_smoothing=True,
        )
        initial_graph = base_graph.copy().set_opacity(0.52)

        initial_label = MathTex(
            r"u_0(x)",
            font_size=26,
            color=COLOR_INITIAL,
        ).move_to(profile_axes.c2p(-3.35, 0.92))

        moving_reference = (
            base_graph.copy()
            .set_color(COLOR_SOLUTION)
            .set_stroke(width=5)
        )
        moving_graph = moving_reference.copy()

        def update_profile(mobject):
            shift_vector = (
                profile_axes.c2p(A*time.get_value(), 0)
                - profile_axes.c2p(0, 0)
            )
            mobject.become(moving_reference.copy().shift(shift_vector))

        moving_graph.add_updater(update_profile)

        solution_label = MathTex(
            r"u(x,t)=u_0(x-at)",
            font_size=26,
            color=COLOR_SOLUTION,
        ).move_to(profile_axes.c2p(2.65, 0.92))

        profile_particle = always_redraw(
            lambda: Dot(
                profile_axes.c2p(
                    XI_SELECTED + A*time.get_value(),
                    u0(XI_SELECTED),
                ),
                radius=0.075,
                color=COLOR_SELECTED,
            )
        )

        profile_vertical = always_redraw(
            lambda: DashedLine(
                profile_axes.c2p(
                    XI_SELECTED + A*time.get_value(),
                    0,
                ),
                profile_axes.c2p(
                    XI_SELECTED + A*time.get_value(),
                    u0(XI_SELECTED),
                ),
                color=COLOR_SELECTED,
                stroke_width=2,
                dash_length=0.09,
            )
        )

        invariant_label = MathTex(
            r"u(X(t),t)=u_0(-2)=1",
            font_size=25,
            color=COLOR_SELECTED,
        ).next_to(profile_axes, DOWN, buff=0.20)

        self.play(
            FadeIn(char_axes),
            FadeIn(char_labels),
            FadeIn(char_heading),
            FadeIn(profile_axes),
            FadeIn(profile_labels),
            FadeIn(profile_heading),
            run_time=1.3,
        )

        self.play(
            LaggedStart(
                *[Create(line) for line in characteristic_lines],
                lag_ratio=0.06,
            ),
            run_time=1.8,
        )

        self.play(
            Create(selected_line),
            FadeIn(xi_label),
            Create(initial_graph),
            FadeIn(initial_label),
            Create(moving_graph),
            FadeIn(solution_label),
            run_time=1.5,
        )

        self.add(
            time_slice,
            char_particle,
            profile_vertical,
            profile_particle,
            time_display,
            x_display,
            invariant_label,
        )

        self.wait(0.5)

        self.play(
            time.animate.set_value(T_FINAL),
            run_time=10.0,
            rate_func=linear,
        )

        self.wait(0.8)

        moving_graph.clear_updaters()
        time_number.clear_updaters()
        x_number.clear_updaters()

        panel_objects = VGroup(
            char_axes,
            char_labels,
            char_heading,
            characteristic_lines,
            selected_line,
            xi_label,
            profile_axes,
            profile_labels,
            profile_heading,
            initial_graph,
            initial_label,
            moving_graph,
            solution_label,
        )

        dynamic_objects = VGroup(
            time_slice,
            char_particle,
            profile_vertical,
            profile_particle,
            time_display,
            x_display,
            invariant_label,
        )

        self.play(
            FadeOut(panel_objects),
            FadeOut(dynamic_objects),
            FadeOut(formula),
            run_time=1.1,
        )

        conclusion = VGroup(
            MathTex(
                r"X'(t)=a,\qquad X(0)=\xi",
                font_size=42,
            ),
            MathTex(
                r"X(t;\xi)=\xi+at",
                font_size=43,
                color=COLOR_SELECTED,
            ),
            MathTex(
                r"\frac{d}{dt}u(X(t;\xi),t)=u_t+a\,u_x=0",
                font_size=41,
            ),
            MathTex(
                r"u(X(t;\xi),t)=u_0(\xi)",
                font_size=43,
                color=COLOR_SOLUTION,
            ),
            MathTex(
                r"\boxed{u(x,t)=u_0(x-at)}",
                font_size=52,
                color=YELLOW_C,
            ),
        ).arrange(DOWN, buff=0.30)

        box = RoundedRectangle(
            corner_radius=0.16,
            width=conclusion.width + 0.90,
            height=conclusion.height + 0.75,
            stroke_color=BLUE_B,
            stroke_width=2,
            fill_color="#10182B",
            fill_opacity=0.65,
        ).move_to(conclusion)

        self.play(
            FadeIn(box),
            LaggedStart(
                *[Write(line) for line in conclusion],
                lag_ratio=0.18,
            ),
            run_time=3.5,
        )

        self.wait(2.0)
````

u_0(x)=\frac{1}{1+x^2}

$$
u(x,t)
