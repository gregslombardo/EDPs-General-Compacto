## Sección 2.6.2. Problema de Riemann para flujos estrictamente convexos

**Fuente principal:** página 12 de `Ecuacione Lineales De Primer Orden.pdf`.

Se estudia el problema de Riemann

$$
\begin{cases}
u_t+f(u)_x=0,
&
x\in\mathbb{R},
\quad
t>0,
\\
u(x,0)=
\begin{cases}
u_L, & x<0,\\
u_R, & x>0,
\end{cases}
\end{cases}
$$

bajo la hipótesis

$$
f\in C^2(\mathbb{R}),
$$

$$
f''(u)\geq\delta>0.
$$

### Objetivos

Al terminar esta sección se deberá poder:

1. definir el problema de Riemann;
2. justificar la autosimilitud de su solución entrópica;
3. construir el choque cuando $u_L>u_R$;
4. construir la rarefacción centrada cuando $u_L<u_R$;
5. derivar la velocidad de Rankine–Hugoniot;
6. derivar el perfil autosimilar de rarefacción;
7. verificar la EDP y las condiciones de entropía;
8. resolver explícitamente el problema para Burgers y para otros flujos convexos.

La escena Manim compara choque y rarefacción en el plano espacio-tiempo. Queda preparada en resolución $2560\times1440$ y $120$ fps, pero no se ejecuta ni se compila.

# 2.6.2. Transcripción organizada de las notas

## Teorema — Solución del problema de Riemann

Sea

$$
f\in C^2(\mathbb{R}),
$$

con

$$
f''(u)\geq\delta>0,
$$

y considérese

$$
u_t+f(u)_x=0,
$$

con dato de Riemann

$$
u(x,0)
=
\begin{cases}
u_L, & x<0,\\
u_R, & x>0,
\end{cases}
$$

donde

$$
u_L\neq u_R.
$$

Se define

$$
a(u)=f'(u).
$$

### Caso a. Choque

Si

$$
u_R<u_L,
$$

la única solución entrópica del problema de Riemann es la onda de choque

$$
u(x,t)
=
\begin{cases}
u_L, & x<\sigma t,\\
u_R, & x>\sigma t,
\end{cases}
$$

que se propaga con velocidad

$$
\boxed{
\sigma
=
\frac{[f(u)]}{[u]}
=
\frac{f(u_R)-f(u_L)}{u_R-u_L}
}.
$$

### Caso b. Rarefacción

Si

$$
u_L<u_R,
$$

la única solución entrópica es la onda de rarefacción centrada en $(0,0)$:

$$
u(x,t)
=
\begin{cases}
u_L,
&
x<a(u_L)t,
\\
g\left(\dfrac{x}{t}\right),
&
a(u_L)t\leq x<a(u_R)t,
\\
u_R,
&
x\geq a(u_R)t,
\end{cases}
$$

donde

$$
g=a^{-1}.
$$

# 2.6.3. Definiciones y preliminares

## Definición 2.6.19 — Problema de Riemann

Un problema de Riemann para una ley de conservación escalar es un problema de Cauchy cuyo dato inicial consta de dos estados constantes separados por un único salto:

$$
u_0(x)
=
\begin{cases}
u_L, & x<0,\\
u_R, & x>0.
\end{cases}
$$

El valor asignado en $x=0$ no afecta la solución débil.

## Definición 2.6.20 — Solución autosimilar

Una solución es autosimilar si existe una función $U$ tal que

$$
u(x,t)=U\left(\frac{x}{t}\right)
$$

para $t>0$.

La variable

$$
\xi=\frac{x}{t}
$$

se denomina variable autosimilar.

## Aclaración 2.6.21 — Invertibilidad de $a=f'$

La hipótesis

$$
f''(u)\geq\delta>0
$$

implica que

$$
a(u)=f'(u)
$$

es estrictamente creciente.

Por tanto, $a$ es invertible sobre su imagen y

$$
g=a^{-1}
$$

está bien definida en el intervalo

$$
[a(u_L),a(u_R)]
$$

cuando $u_L<u_R$.

# 2.6.4. Autosimilitud

## Resultado añadido 2.6.22 — Invariancia de escala

El problema de Riemann es invariante bajo la transformación

$$
(x,t)\longmapsto(\lambda x,\lambda t),
$$

con

$$
\lambda>0.
$$

En efecto, si $u$ es una solución, entonces

$$
u_\lambda(x,t)
=
u(\lambda x,\lambda t)
$$

satisface la misma ecuación y el mismo dato inicial.

## Proposición 2.6.23 — Autosimilitud por unicidad

Supongamos que el problema de Riemann posee una única solución entrópica. Entonces

$$
u(\lambda x,\lambda t)=u(x,t)
$$

para todo $\lambda>0$.

Por consiguiente,

$$
u(x,t)=U\left(\frac{x}{t}\right).
$$

### Demostración

La función $u_\lambda$ satisface el mismo problema de Riemann que $u$. La unicidad entrópica implica

$$
u_\lambda=u.
$$

Tomando

$$
\lambda=\frac{1}{t},
$$

se obtiene

$$
u(x,t)
=
u\left(\frac{x}{t},1\right).
$$

Definiendo

$$
U(\xi)=u(\xi,1),
$$

resulta

$$
u(x,t)=U(\xi).
$$

# 2.6.5. Ecuación autosimilar

## Proposición 2.6.24 — Reducción de la EDP

Sea

$$
u(x,t)=U(\xi),
$$

con

$$
\xi=\frac{x}{t}.
$$

En una región donde $U$ es diferenciable,

$$
u_t
=
-\frac{\xi}{t}U'(\xi),
$$

y

$$
u_x
=
\frac{1}{t}U'(\xi).
$$

Por tanto,

$$
u_t+f(u)_x
=
\frac{1}{t}
\left[
f'(U(\xi))-\xi
\right]
U'(\xi).
$$

La EDP se reduce a

$$
\boxed{
\left[
a(U(\xi))-\xi
\right]
U'(\xi)=0
}.
$$

## Interpretación 2.6.25

En cada intervalo conectado ocurre una de dos posibilidades:

1. $U'(\xi)=0$, por lo que el estado es constante;
2. $a(U(\xi))=\xi$, por lo que
   $$
   U(\xi)=a^{-1}(\xi)=g(\xi).
   $$

La solución del problema de Riemann se construye pegando estados constantes y, cuando es necesario, una rama no constante determinada por $g$.

# 2.6.6. Caso de choque: $u_L>u_R$

## Teorema 2.6.26 — Solución de choque

Supongamos

$$
u_L>u_R.
$$

Entonces la solución entrópica del problema de Riemann es

$$
u(x,t)
=
\begin{cases}
u_L, & x<\sigma t,\\
u_R, & x>\sigma t,
\end{cases}
$$

donde

$$
\sigma
=
\frac{f(u_R)-f(u_L)}{u_R-u_L}.
$$

### Demostración

La función es constante a cada lado del frente, por lo que satisface la EDP clásicamente en ambas regiones.

Para ser solución débil debe cumplir Rankine–Hugoniot:

$$
-\sigma[u]+[f(u)]=0.
$$

Como

$$
[u]=u_R-u_L\neq0,
$$

se obtiene

$$
\sigma
=
\frac{[f(u)]}{[u]}.
$$

La convexidad estricta implica

$$
f'(u_R)<\sigma<f'(u_L).
$$

Por tanto, las características entran al frente desde ambos lados y el salto satisface la condición estricta de Lax.

La unicidad se sigue de la contracción $L^1$ para soluciones entrópicas.

## Interpretación 2.6.27 — Compresión

Las características del estado izquierdo tienen velocidad

$$
a(u_L),
$$

y las del estado derecho tienen velocidad

$$
a(u_R).
$$

Como

$$
a(u_R)<\sigma<a(u_L),
$$

las características convergen hacia el frente. El choque concentra la información que, en la solución clásica, produciría un cruce multivaluado.

# 2.6.7. Caso de rarefacción: $u_L<u_R$

## Teorema 2.6.28 — Rarefacción centrada

Supongamos

$$
u_L<u_R.
$$

Entonces la solución entrópica es

$$
u(x,t)
=
\begin{cases}
u_L,
&
\dfrac{x}{t}<a(u_L),
\\
g\left(\dfrac{x}{t}\right),
&
a(u_L)\leq\dfrac{x}{t}<a(u_R),
\\
u_R,
&
\dfrac{x}{t}\geq a(u_R),
\end{cases}
$$

donde

$$
g=a^{-1}.
$$

### Demostración

En las regiones exteriores, $u$ es constante y satisface la EDP.

En el abanico

$$
a(u_L)<\frac{x}{t}<a(u_R),
$$

se define

$$
u(x,t)=g\left(\frac{x}{t}\right).
$$

Como

$$
a(g(\xi))=\xi,
$$

la ecuación autosimilar

$$
[a(U)-\xi]U'=0
$$

se satisface.

En los bordes del abanico,

$$
g(a(u_L))=u_L,
$$

y

$$
g(a(u_R))=u_R.
$$

Por tanto, la solución es continua a través de ambos bordes.

No aparecen saltos internos. La solución es clásica por pedazos, continua y débil.

Las características dentro del abanico son las rectas

$$
x=a(u)t,
$$

con estados intermedios

$$
u\in[u_L,u_R].
$$

La solución expande la discontinuidad inicial en un abanico continuo de características.

La unicidad se sigue nuevamente del teorema de contracción.

## Observación 2.6.29 — Centro de la rarefacción

Todas las rectas del abanico parten del punto

$$
(x,t)=(0,0).
$$

Por ello la rarefacción se denomina centrada.

# 2.6.8. Verificación entrópica

## Proposición 2.6.30 — El choque decreciente satisface Lax

Para $u_L>u_R$,

$$
f'(u_R)<\sigma<f'(u_L).
$$

Por tanto, el choque es entrópico.

## Proposición 2.6.31 — La rarefacción creciente satisface Oleinik

Dentro del abanico,

$$
u(x,t)=g\left(\frac{x}{t}\right).
$$

Derivando,

$$
u_x(x,t)
=
\frac{1}{t}
g'\left(\frac{x}{t}\right).
$$

Como

$$
g'(\xi)
=
\frac{1}{a'(g(\xi))}
=
\frac{1}{f''(g(\xi))},
$$

y

$$
f''\geq\delta,
$$

se obtiene

$$
0\leq u_x(x,t)\leq\frac{1}{\delta t}.
$$

En consecuencia,

$$
\frac{u(x+h,t)-u(x,t)}{h}
\leq
\frac{1}{\delta t}
$$

para todo $h>0$.

La rarefacción satisface la condición unilateral de Oleinik.

## Contraejemplo 2.6.32 — Salto creciente no entrópico

Si $u_L<u_R$, la función

$$
u(x,t)
=
\begin{cases}
u_L, & x<\sigma t,\\
u_R, & x>\sigma t,
\end{cases}
$$

con velocidad de Rankine–Hugoniot es una solución débil.

Sin embargo,

$$
f'(u_L)<\sigma<f'(u_R).
$$

Las características se alejan del frente y el salto viola Oleinik. La solución entrópica correcta es la rarefacción centrada.

# 2.6.9. Ejemplos

## Ejemplo 2.6.33 — Burgers: choque

Para

$$
f(u)=\frac{u^2}{2},
$$

se tiene

$$
a(u)=u.
$$

Si

$$
u_L=2,
$$

$$
u_R=0,
$$

la velocidad es

$$
\sigma
=
\frac{u_L+u_R}{2}
=
1.
$$

La solución es

$$
u(x,t)
=
\begin{cases}
2, & x<t,\\
0, & x>t.
\end{cases}
$$

## Ejemplo 2.6.34 — Burgers: rarefacción

Si

$$
u_L=0,
$$

$$
u_R=2,
$$

entonces

$$
g(\xi)=\xi.
$$

La solución es

$$
u(x,t)
=
\begin{cases}
0, & x<0,\\
\dfrac{x}{t}, & 0\leq x<2t,\\
2, & x\geq2t.
\end{cases}
$$

## Ejemplo 2.6.35 — Flujo exponencial

Sea

$$
f(u)=e^u.
$$

Entonces

$$
a(u)=e^u,
$$

y

$$
g(\xi)=\log\xi.
$$

Para estados

$$
u_L<u_R,
$$

la rarefacción es

$$
u(x,t)
=
\begin{cases}
u_L,
&
x<e^{u_L}t,
\\
\log\left(\dfrac{x}{t}\right),
&
e^{u_L}t\leq x<e^{u_R}t,
\\
u_R,
&
x\geq e^{u_R}t.
\end{cases}
$$

```python
%%manim -v WARNING -r 2560,1440 --fps 120 RiemannChoqueRarefaccion

from manim import *
import numpy as np

config.no_latex_cleanup = True
config.tex_dir = str(TEX_RUNTIME_DIR)
config.media_embed = True
config.media_width = "100%"

T_FINAL = 3.0


class RiemannChoqueRarefaccion(Scene):
    def construct(self):
        self.camera.background_color = "#0B1020"

        color_left = ORANGE
        color_right = TEAL_C
        color_shock = YELLOW_C
        color_fan = BLUE_B
        color_time = GREEN_C

        title = Text(
            "Problema de Riemann: choque y rarefacción",
            font_size=40,
            weight=BOLD,
            color=WHITE,
        ).to_edge(UP, buff=0.18)

        equation = MathTex(
            r"u_t+\left(\frac{u^2}{2}\right)_x=0",
            r"\qquad",
            r"u_0(x)=\begin{cases}u_L,&x<0,\\u_R,&x>0,\end{cases}",
            font_size=33,
        ).next_to(title, DOWN, buff=0.12)

        self.play(
            FadeIn(title, shift=0.10 * DOWN),
            Write(equation),
            run_time=1.7,
        )

        time = ValueTracker(0.0)

        shock_axes = Axes(
            x_range=[-3, 6, 1],
            y_range=[0, T_FINAL, 0.5],
            x_length=5.8,
            y_length=4.9,
            tips=False,
            axis_config={
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 18,
            },
        ).move_to(3.55 * LEFT + 0.55 * DOWN)

        rare_axes = Axes(
            x_range=[-3, 6, 1],
            y_range=[0, T_FINAL, 0.5],
            x_length=5.8,
            y_length=4.9,
            tips=False,
            axis_config={
                "stroke_width": 2,
                "include_numbers": True,
                "font_size": 18,
            },
        ).move_to(3.55 * RIGHT + 0.55 * DOWN)

        shock_labels = shock_axes.get_axis_labels(
            MathTex(r"x", font_size=27),
            MathTex(r"t", font_size=27),
        )

        rare_labels = rare_axes.get_axis_labels(
            MathTex(r"x", font_size=27),
            MathTex(r"t", font_size=27),
        )

        shock_heading = Text(
            "Choque:  u_L=2 > u_R=0",
            font_size=27,
            weight=SEMIBOLD,
            color=color_shock,
        ).next_to(shock_axes, UP, buff=0.12)

        rare_heading = Text(
            "Rarefacción:  u_L=0 < u_R=2",
            font_size=27,
            weight=SEMIBOLD,
            color=color_fan,
        ).next_to(rare_axes, UP, buff=0.12)

        shock_region_left = Polygon(
            shock_axes.c2p(-3, 0),
            shock_axes.c2p(0, 0),
            shock_axes.c2p(3, T_FINAL),
            shock_axes.c2p(-3, T_FINAL),
            stroke_width=0,
            fill_color=color_left,
            fill_opacity=0.10,
        )

        shock_region_right = Polygon(
            shock_axes.c2p(0, 0),
            shock_axes.c2p(6, 0),
            shock_axes.c2p(6, T_FINAL),
            shock_axes.c2p(3, T_FINAL),
            stroke_width=0,
            fill_color=color_right,
            fill_opacity=0.10,
        )

        shock_line = Line(
            shock_axes.c2p(0, 0),
            shock_axes.c2p(T_FINAL, T_FINAL),
            color=color_shock,
            stroke_width=6,
        )

        shock_formula = MathTex(
            r"x=\sigma t=t",
            font_size=27,
            color=color_shock,
        ).next_to(
            shock_axes.c2p(2.0, 2.0),
            RIGHT,
            buff=0.06,
        )

        shock_characteristics = VGroup()

        for xi in np.linspace(-2.5, -0.5, 5):
            hit_time = -xi
            shock_characteristics.add(
                Line(
                    shock_axes.c2p(xi, 0),
                    shock_axes.c2p(hit_time, hit_time),
                    color=color_left,
                    stroke_width=2.5,
                    stroke_opacity=0.72,
                )
            )

        for xi in np.linspace(0.5, 2.5, 5):
            hit_time = xi
            shock_characteristics.add(
                Line(
                    shock_axes.c2p(xi, 0),
                    shock_axes.c2p(hit_time, hit_time),
                    color=color_right,
                    stroke_width=2.5,
                    stroke_opacity=0.72,
                )
            )

        rare_region_left = Polygon(
            rare_axes.c2p(-3, 0),
            rare_axes.c2p(0, 0),
            rare_axes.c2p(0, T_FINAL),
            rare_axes.c2p(-3, T_FINAL),
            stroke_width=0,
            fill_color=color_left,
            fill_opacity=0.10,
        )

        rare_region_fan = Polygon(
            rare_axes.c2p(0, 0),
            rare_axes.c2p(0, T_FINAL),
            rare_axes.c2p(2 * T_FINAL, T_FINAL),
            stroke_width=0,
            fill_color=color_fan,
            fill_opacity=0.13,
        )

        rare_region_right = Polygon(
            rare_axes.c2p(0, 0),
            rare_axes.c2p(6, 0),
            rare_axes.c2p(6, T_FINAL),
            stroke_width=0,
            fill_color=color_right,
            fill_opacity=0.10,
        )

        fan_boundaries = VGroup(
            Line(
                rare_axes.c2p(0, 0),
                rare_axes.c2p(0, T_FINAL),
                color=color_fan,
                stroke_width=4,
            ),
            Line(
                rare_axes.c2p(0, 0),
                rare_axes.c2p(2 * T_FINAL, T_FINAL),
                color=color_fan,
                stroke_width=4,
            ),
        )

        rare_characteristics = VGroup()

        for xi in np.linspace(-2.5, -0.5, 5):
            rare_characteristics.add(
                Line(
                    rare_axes.c2p(xi, 0),
                    rare_axes.c2p(xi, T_FINAL),
                    color=color_left,
                    stroke_width=2.5,
                    stroke_opacity=0.72,
                )
            )

        for state in np.linspace(0.2, 1.8, 9):
            rare_characteristics.add(
                Line(
                    rare_axes.c2p(0, 0),
                    rare_axes.c2p(state * T_FINAL, T_FINAL),
                    color=color_fan,
                    stroke_width=2.3,
                    stroke_opacity=0.68,
                )
            )

        for xi in np.linspace(0.5, 2.0, 4):
            rare_characteristics.add(
                Line(
                    rare_axes.c2p(xi, 0),
                    rare_axes.c2p(xi + 2 * T_FINAL, T_FINAL),
                    color=color_right,
                    stroke_width=2.5,
                    stroke_opacity=0.72,
                )
            )

        rare_formula = MathTex(
            r"u=\frac{x}{t}",
            r"\quad",
            r"0\leq\frac{x}{t}\leq2",
            font_size=26,
            color=color_fan,
        ).move_to(rare_axes.c2p(2.9, 1.55))

        shock_slice = always_redraw(
            lambda: DashedLine(
                shock_axes.c2p(-3, time.get_value()),
                shock_axes.c2p(6, time.get_value()),
                color=color_time,
                stroke_width=2,
                dash_length=0.08,
            )
        )

        rare_slice = always_redraw(
            lambda: DashedLine(
                rare_axes.c2p(-3, time.get_value()),
                rare_axes.c2p(6, time.get_value()),
                color=color_time,
                stroke_width=2,
                dash_length=0.08,
            )
        )

        shock_point = always_redraw(
            lambda: Dot(
                shock_axes.c2p(
                    time.get_value(),
                    time.get_value(),
                ),
                radius=0.07,
                color=color_shock,
            )
        )

        time_tex = MathTex(
            r"t=",
            font_size=29,
            color=color_time,
        )

        time_number = DecimalNumber(
            0.0,
            num_decimal_places=2,
            font_size=29,
            color=color_time,
        )
        time_number.add_updater(
            lambda mob: mob.set_value(time.get_value())
        )

        time_display = VGroup(
            time_tex,
            time_number,
        ).arrange(
            RIGHT,
            buff=0.05,
        ).move_to(0.18 * DOWN)

        shock_lax = MathTex(
            r"0=f'(u_R)<\sigma=1<f'(u_L)=2",
            font_size=25,
            color=color_shock,
        ).next_to(shock_axes, DOWN, buff=0.13)

        rare_bounds = MathTex(
            r"a(u_L)t=0",
            r"\qquad",
            r"a(u_R)t=2t",
            font_size=25,
            color=color_fan,
        ).next_to(rare_axes, DOWN, buff=0.13)

        self.play(
            FadeIn(shock_axes),
            FadeIn(rare_axes),
            FadeIn(shock_labels),
            FadeIn(rare_labels),
            FadeIn(shock_heading),
            FadeIn(rare_heading),
            FadeIn(shock_region_left),
            FadeIn(shock_region_right),
            FadeIn(rare_region_left),
            FadeIn(rare_region_fan),
            FadeIn(rare_region_right),
            Create(shock_line),
            FadeIn(shock_formula),
            FadeIn(fan_boundaries),
            FadeIn(rare_formula),
            LaggedStart(
                *[Create(line) for line in shock_characteristics],
                lag_ratio=0.05,
            ),
            LaggedStart(
                *[Create(line) for line in rare_characteristics],
                lag_ratio=0.035,
            ),
            FadeIn(shock_lax),
            FadeIn(rare_bounds),
            FadeIn(time_display),
            run_time=3.2,
        )

        # Los datos y las geometrías se observan antes de la evolución.
        self.wait(4.0)

        self.add(
            shock_slice,
            rare_slice,
            shock_point,
        )

        self.play(
            time.animate.set_value(T_FINAL),
            run_time=12.0,
            rate_func=linear,
        )

        self.wait(2.0)

        time_number.clear_updaters()

        conclusion = VGroup(
            MathTex(
                r"u_L>u_R",
                r"\Longrightarrow",
                r"\text{choque}",
                font_size=42,
            ),
            MathTex(
                r"\sigma=\frac{f(u_R)-f(u_L)}{u_R-u_L}",
                font_size=39,
                color=color_shock,
            ),
            MathTex(
                r"u_L<u_R",
                r"\Longrightarrow",
                r"\text{rarefacción}",
                font_size=42,
            ),
            MathTex(
                r"u(x,t)=a^{-1}\left(\frac{x}{t}\right)",
                font_size=39,
                color=color_fan,
            ),
        ).arrange(DOWN, buff=0.27)

        conclusion[0][0].set_color(color_shock)
        conclusion[0][2].set_color(color_shock)
        conclusion[2][0].set_color(color_fan)
        conclusion[2][2].set_color(color_fan)

        conclusion_box = SurroundingRectangle(
            conclusion,
            color=BLUE_B,
            stroke_width=3,
            buff=0.38,
            corner_radius=0.12,
        )

        panel_objects = VGroup(
            shock_axes,
            rare_axes,
            shock_labels,
            rare_labels,
            shock_heading,
            rare_heading,
            shock_region_left,
            shock_region_right,
            rare_region_left,
            rare_region_fan,
            rare_region_right,
            shock_line,
            shock_formula,
            fan_boundaries,
            rare_formula,
            shock_characteristics,
            rare_characteristics,
            shock_lax,
            rare_bounds,
            time_display,
            shock_slice,
            rare_slice,
            shock_point,
            equation,
        )

        self.play(
            FadeOut(panel_objects),
            run_time=1.4,
        )

        self.play(
            FadeIn(conclusion_box),
            LaggedStart(
                *[Write(item) for item in conclusion],
                lag_ratio=0.18,
            ),
            run_time=3.6,
        )

        self.wait(3.0)
```
